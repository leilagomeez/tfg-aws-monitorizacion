```python
# === Global configuration ===
import re

SERVICES = ['Overview', 'Lambda', 'DynamoDB', 'API Gateway', 'SQS', 'SNS', 'S3',
            'RDS', 'Cognito', 'Step Functions', 'WAF', 'Elastic Beanstalk',
            'ElastiCache', 'ECR', 'CodeCommit']

TABS = ['Discovery', 'Health', 'Metrics', 'Logs', 'Security']

import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd

# Time range (configurable)
RANGO = 'ultimo_mes'  # ultima_hora | ultimo_dia | ultima_semana | ultimo_mes

_rangos = {
    'ultima_hora':   (timedelta(hours=1), 60),
    'ultimo_dia':    (timedelta(days=1), 300),
    'ultima_semana': (timedelta(weeks=1), 3600),
    'ultimo_mes':    (timedelta(days=30), 86400),
}
_delta, PERIOD = _rangos.get(RANGO, _rangos['ultimo_dia'])
end_time = datetime.now(tz=None)
start_time = end_time - _delta

cloudwatch = clients['cloudwatch']

def get_metric(namespace, metric_name, dimensions, stat, default=0):
    """Query a single CloudWatch metric and return one aggregated value."""
    try:
        response = cloudwatch.get_metric_statistics(
            Namespace=namespace, MetricName=metric_name,
            Dimensions=dimensions, StartTime=start_time, EndTime=end_time,
            Period=PERIOD, Statistics=[stat]
        )
        dps = response.get('Datapoints', [])
        if not dps:
            return default
        if stat == 'Sum':
            return round(sum(dp['Sum'] for dp in dps), 2)
        if stat == 'Average':
            return round(sum(dp['Average'] for dp in dps) / len(dps), 2)
        if stat == 'Maximum':
            return round(max(dp['Maximum'] for dp in dps), 2)
        return default
    except Exception:
        return default


def get_metric_series(namespace, metric_name, dimensions, stat):
    """Return sorted CloudWatch datapoints for time-series charts."""
    try:
        response = cloudwatch.get_metric_statistics(
            Namespace=namespace, MetricName=metric_name,
            Dimensions=dimensions, StartTime=start_time, EndTime=end_time,
            Period=PERIOD, Statistics=[stat]
        )
        datapoints = []
        for dp in response.get('Datapoints', []):
            if stat in dp:
                datapoints.append((dp['Timestamp'].replace(tzinfo=None), round(dp[stat], 2)))
        return sorted(datapoints, key=lambda item: item[0])
    except Exception:
        return []

print(f"Metrics config: range={RANGO}, period={PERIOD}s, from={start_time.strftime('%Y-%m-%d %H:%M')} to={end_time.strftime('%Y-%m-%d %H:%M')}")
```

```python
# === Visual helpers ===
# --- HTML rendering helpers ---
def html_title(text):
    return f'<h3 style="color: #2E5090; border-bottom: 2px solid #2E5090; padding-bottom: 6px;">{text}</h3>'

def html_section(text):
    return f'<h4 style="color: #333; margin-top: 22px; margin-bottom: 8px; font-size: 18px; font-weight: 700;">{text}</h4>'


def html_subsection(text):
    return f'<h5 style="color: #555; margin-top: 16px; margin-bottom: 6px; font-size: 15px; font-weight: 700;">{text}</h5>'


def html_table(headers, rows, widths=None):
    """Builds a stable, left-aligned HTML table inside the notebook."""
    th_style = (
        'style="background: #2E5090; color: white; padding: 9px 12px; '
        'text-align: left; vertical-align: middle; font-size: 13px; font-weight: 700;"'
    )
    td_base = (
        'padding: 8px 12px; border-bottom: 1px solid #e8e8e8; font-size: 13px; '
        'text-align: left; vertical-align: middle; overflow-wrap: anywhere;'
    )
    table_style = (
        'border-collapse: collapse; width: 100%; max-width: 100%; table-layout: fixed; '
        'margin: 10px 0 16px 0; direction: ltr; font-family: Arial, sans-serif;'
    )

    html = f'<table style="{table_style}">'
    if widths:
        html += '<colgroup>' + ''.join(f'<col style="width: {width};">' for width in widths) + '</colgroup>'
    html += '<thead><tr>' + ''.join(f'<th {th_style}>{h}</th>' for h in headers) + '</tr></thead><tbody>'
    for i, row in enumerate(rows):
        bg = ' background: #f7f7f7;' if i % 2 == 1 else ''
        td_style = f'style="{td_base}{bg}"'
        html += '<tr>' + ''.join(f'<td {td_style}>{cell}</td>' for cell in row) + '</tr>'
    html += '</tbody></table>'
    return html

def html_status(level, text, display=None):
    """Builds a compact status badge."""
    colors = {
        'HEALTHY': ('#27ae60', '#e8f8f0'), 'ACTIVE': ('#27ae60', '#e8f8f0'),
        'AVAILABLE': ('#27ae60', '#e8f8f0'), 'OK': ('#27ae60', '#e8f8f0'),
        'ENABLED': ('#27ae60', '#e8f8f0'), 'DEPLOYED': ('#27ae60', '#e8f8f0'),
        'GREEN': ('#27ae60', '#e8f8f0'), 'NO FAILURES': ('#27ae60', '#e8f8f0'),
        'ACCESSIBLE': ('#27ae60', '#e8f8f0'), 'WEBSITE ACTIVE': ('#27ae60', '#e8f8f0'),
        'WARNING': ('#e67e22', '#fef5e7'), 'WARN': ('#e67e22', '#fef5e7'),
        'DISABLED': ('#e67e22', '#fef5e7'), 'NOT DEPLOYED': ('#e67e22', '#fef5e7'),
        'EMPTY': ('#e67e22', '#fef5e7'), 'YELLOW': ('#e67e22', '#fef5e7'),
        'CREATING': ('#e67e22', '#fef5e7'), 'UPDATING': ('#e67e22', '#fef5e7'),
        'MODIFYING': ('#e67e22', '#fef5e7'), 'BACKING-UP': ('#e67e22', '#fef5e7'),
        'STARTING': ('#e67e22', '#fef5e7'), 'STOPPING': ('#e67e22', '#fef5e7'),
        'STOPPED': ('#e67e22', '#fef5e7'), 'PENDING': ('#e67e22', '#fef5e7'),
        'CRIT': ('#c0392b', '#fce4e4'), 'CRITICAL': ('#c0392b', '#fce4e4'),
        'ALARM': ('#c0392b', '#fce4e4'),
        'UNHEALTHY': ('#c0392b', '#fce4e4'), 'FAILED': ('#c0392b', '#fce4e4'),
        'RED': ('#c0392b', '#fce4e4'), 'NOT ACCESSIBLE': ('#c0392b', '#fce4e4'),
        'DELETING': ('#c0392b', '#fce4e4'), 'DELETE_FAILED': ('#c0392b', '#fce4e4'),
        'INFO': ('#7f8c8d', '#f0f0f0'), 'NO ACCESS': ('#7f8c8d', '#f0f0f0'),
        'INSUFFICIENT_DATA': ('#7f8c8d', '#f0f0f0'), 'GREY': ('#7f8c8d', '#f0f0f0'),
        'GRAY': ('#7f8c8d', '#f0f0f0'), 'UNKNOWN': ('#7f8c8d', '#f0f0f0'),
        'N/A': ('#7f8c8d', '#f0f0f0'),
    }
    key = (level or 'INFO').upper()
    fg, bg = colors.get(key, ('#7f8c8d', '#f0f0f0'))
    shown = display if display is not None else level
    return f'<span style="background:{bg}; color:{fg}; padding: 2px 8px; border-radius: 3px; font-size: 12px; font-weight: bold; white-space: nowrap; display: inline-block; line-height: 1.35;">{shown}</span> {text}'

def health_category(status):
    status = (status or 'INFO').upper()
    if status in ('HEALTHY', 'ACTIVE', 'AVAILABLE', 'OK', 'ENABLED', 'DEPLOYED', 'GREEN', 'NO FAILURES', 'ACCESSIBLE', 'WEBSITE ACTIVE'):
        return 'Healthy'
    if status in ('WARNING', 'WARN', 'DISABLED', 'NOT DEPLOYED', 'EMPTY', 'YELLOW', 'CREATING', 'UPDATING', 'MODIFYING', 'BACKING-UP', 'STARTING', 'STOPPING', 'STOPPED', 'PENDING'):
        return 'Warning'
    if status in ('UNHEALTHY', 'FAILED', 'RED', 'NOT ACCESSIBLE', 'DELETING', 'DELETE_FAILED'):
        return 'Critical'
    return 'Info'

def render_health_donut(category_counts):
    total = sum(category_counts.values())
    if total == 0:
        return ''

    colors = {
        'Healthy': '#2FA36B',
        'Warning': '#E39A2D',
        'Critical': '#C94A44',
        'Info': '#7F8C8D',
    }
    labels = ['Healthy', 'Warning', 'Critical', 'Info']
    radius = 58
    circumference = 2 * 3.141592653589793 * radius
    offset = 0
    segments = []

    for label in labels:
        value = category_counts.get(label, 0)
        if value <= 0:
            continue
        length = circumference * value / total
        segments.append(
            f'<circle cx="80" cy="80" r="{radius}" fill="none" '
            f'stroke="{colors[label]}" stroke-width="24" '
            f'stroke-dasharray="{length:.2f} {circumference - length:.2f}" '
            f'stroke-dashoffset="{-offset:.2f}" stroke-linecap="butt" '
            f'transform="rotate(-90 80 80)" />'
        )
        offset += length

    legend_items = ''.join(
        f'<div style="display:flex; align-items:center; gap:8px; margin:4px 0; font-size:13px;">'
        f'<span style="width:10px; height:10px; background:{colors[label]}; display:inline-block; border-radius:2px;"></span>'
        f'<span style="min-width:70px;">{label}</span><strong>{category_counts.get(label, 0)}</strong>'
        f'</div>'
        for label in labels
    )

    return (
        '<div style="display:flex; align-items:center; gap:28px; margin: 8px 0 18px 0; '
        'padding: 14px 16px; border:1px solid #e6e6e6; background:#fafafa;">'
        '<svg width="160" height="160" viewBox="0 0 160 160" role="img" aria-label="Health status distribution">'
        '<circle cx="80" cy="80" r="58" fill="none" stroke="#eeeeee" stroke-width="24" />'
        + ''.join(segments) +
        f'<text x="80" y="76" text-anchor="middle" style="font-size:24px; font-weight:700; fill:#2E5090;">{total}</text>'
        '<text x="80" y="96" text-anchor="middle" style="font-size:11px; fill:#777;">services</text>'
        '</svg>'
        '<div>'
        '<div style="font-weight:700; color:#2E5090; margin-bottom:8px;">Health status distribution</div>'
        + legend_items +
        '</div>'
        '</div>'
    )

def html_metric_range(include_period=False):
    labels = {
        'ultima_hora': 'Last hour',
        'ultimo_dia': 'Last day',
        'ultima_semana': 'Last week',
        'ultimo_mes': 'Last month',
    }
    label = labels.get(RANGO, RANGO)
    detail = f' | Period: {PERIOD}s' if include_period else ''
    return f'<p style="color:#777; font-size:12px; margin: 4px 0 12px 0;">Range: {label}{detail}</p>'


def html_empty(message):
    return f'<p style="color:#777; font-size:13px; margin: 8px 0 14px 0;">{message}</p>'


def html_success(message):
    return f'<p style="color:#2FA36B; font-size:13px; font-weight:600; margin: 8px 0 14px 0;">{message}</p>'


def html_note(message):
    return f'<p style="color:#777; font-size:12px; margin: 4px 0 12px 0;">Note: {message}</p>'

def html_escape(value):
    text = str(value)
    return text.replace('&', '&amp;').replace('<', '&lt;').replace('>', '&gt;')


def details_list(summary, items):
    clean_items = [html_escape(item) for item in items if item]
    if not clean_items:
        return html_escape(summary)
    body = ''.join(f'<div style="padding:2px 0;">{item}</div>' for item in clean_items)
    return (
        '<details style="margin:0;">'
        f'<summary style="cursor:pointer; color:#2E5090; font-weight:600;">{html_escape(summary)}</summary>'
        f'<div style="margin-top:6px; color:#333; line-height:1.35;">{body}</div>'
        '</details>'
    )
```

```python
# === Health collectors ===

def health_lambda(c):
    fn_list = c['lambda'].list_functions()['Functions']
    functions = []
    for fn in fn_list:
        name = fn['FunctionName']
        try:
            config = c['lambda'].get_function_configuration(FunctionName=name)
            state = config.get('State', 'N/A')
            last_update = config.get('LastUpdateStatus', 'N/A')
            if state == 'Active' and last_update == 'Successful':
                functions.append({'status': 'HEALTHY', 'resource': name, 'detail': ''})
            elif state == 'Active':
                functions.append({'status': 'WARNING', 'resource': name, 'detail': f'last update: {last_update}'})
            else:
                functions.append({'status': 'UNHEALTHY', 'resource': name, 'detail': state})
        except Exception:
            continue

    mappings = []
    for m in c['lambda'].list_event_source_mappings()['EventSourceMappings']:
        fn_name = m.get('FunctionArn', '').split(':')[-1]
        source = m.get('EventSourceArn', '').split(':')[-1]
        state = m.get('State', 'N/A')
        mappings.append({
            'status': 'ENABLED' if state == 'Enabled' else 'DISABLED',
            'function': fn_name,
            'source': source,
        })
    return {'functions': functions, 'mappings': mappings}


def health_dynamodb(c):
    result = []
    tables = c['dynamodb'].list_tables()['TableNames']
    for table in tables:
        try:
            info = c['dynamodb'].describe_table(TableName=table)['Table']
            status = info['TableStatus']
            items = info.get('ItemCount', 0)
            try:
                backups = c['dynamodb'].describe_continuous_backups(TableName=table)['ContinuousBackupsDescription']
                pitr = backups.get('PointInTimeRecoveryDescription', {}).get('PointInTimeRecoveryStatus', 'DISABLED')
                pitr_detail = 'PITR disabled' if pitr == 'DISABLED' else 'PITR enabled'
            except Exception:
                pitr_detail = ''
            detail = f'{items} items'
            if pitr_detail:
                detail += f' | {pitr_detail}'
            result.append({'status': 'ACTIVE' if status == 'ACTIVE' else status, 'resource': table, 'detail': detail})
        except Exception:
            result.append({'status': 'NO ACCESS', 'resource': table, 'detail': ''})
    return result


def health_apigateway(c):
    result = []
    apis = c['apigateway'].get_rest_apis()['items']
    for api in apis:
        api_name = api['name']
        api_id = api['id']
        checks = []
        try:
            stages = c['apigateway'].get_stages(restApiId=api_id)['item']
            if stages:
                for stage in stages:
                    date = str(stage.get('lastUpdatedDate', ''))[:10]
                    checks.append({'status': 'DEPLOYED', 'detail': f'stage: {stage["stageName"]} (updated: {date})'})
            else:
                checks.append({'status': 'NOT DEPLOYED', 'detail': 'no stages'})

            resources = c['apigateway'].get_resources(restApiId=api_id, limit=500)['items']
            ok, fail = 0, 0
            for rec in resources:
                for method in rec.get('resourceMethods', {}):
                    if method == 'OPTIONS':
                        continue
                    try:
                        integ = c['apigateway'].get_integration(
                            restApiId=api_id, resourceId=rec['id'], httpMethod=method
                        )
                        if integ.get('uri') or integ.get('type') == 'MOCK':
                            ok += 1
                        else:
                            fail += 1
                    except Exception:
                        fail += 1
            if fail:
                checks.append({'status': 'WARNING', 'detail': f'{ok} connected, {fail} missing'})
            else:
                checks.append({'status': 'OK', 'detail': f'{ok} endpoints connected'})
        except Exception:
            checks.append({'status': 'NO ACCESS', 'detail': ''})
        result.append({'api_name': api_name, 'checks': checks})
    return result


def health_sqs(c):
    result = []
    urls = c['sqs'].list_queues().get('QueueUrls', [])
    for url in urls:
        name = url.split('/')[-1]
        try:
            attrs = c['sqs'].get_queue_attributes(
                QueueUrl=url,
                AttributeNames=['ApproximateNumberOfMessages', 'ApproximateNumberOfMessagesNotVisible'],
            )['Attributes']
            visible = int(attrs.get('ApproximateNumberOfMessages', 0))
            in_flight = int(attrs.get('ApproximateNumberOfMessagesNotVisible', 0))
            is_dlq = 'dlq' in name.lower()
            if is_dlq and visible > 0:
                result.append({'status': 'FAILED', 'resource': name, 'detail': f'{visible} failed messages'})
            elif is_dlq:
                result.append({'status': 'NO FAILURES', 'resource': name, 'detail': ''})
            elif visible > 10:
                result.append({'status': 'WARNING', 'resource': name, 'detail': f'{visible} pending'})
            else:
                result.append({'status': 'OK', 'resource': name, 'detail': f'{visible} pending, {in_flight} in flight'})
        except Exception:
            result.append({'status': 'NO ACCESS', 'resource': name, 'detail': ''})
    return result


def health_sns(c):
    result = []
    topics = c['sns'].list_topics()['Topics']
    for topic in topics:
        name = topic['TopicArn'].split(':')[-1]
        try:
            attrs = c['sns'].get_topic_attributes(TopicArn=topic['TopicArn'])['Attributes']
            confirmed = int(attrs.get('SubscriptionsConfirmed', 0))
            pending = int(attrs.get('SubscriptionsPending', 0))
            if confirmed > 0 and pending == 0:
                result.append({'status': 'OK', 'resource': name, 'detail': f'{confirmed} active subscriptions'})
            elif pending > 0:
                result.append({'status': 'WARNING', 'resource': name, 'detail': f'{pending} pending confirmation'})
            else:
                result.append({'status': 'WARNING', 'resource': name, 'detail': 'no subscriptions'})
        except Exception:
            pass
    return result


def health_s3(c):
    result = []
    buckets = c['s3'].list_buckets()['Buckets']
    for b in buckets:
        name = b['Name']
        checks = []
        try:
            c['s3'].head_bucket(Bucket=name)
            objs = c['s3'].list_objects_v2(Bucket=name, MaxKeys=1)
            has_content = objs.get('KeyCount', 0) > 0
            checks.append({
                'status': 'ACCESSIBLE' if has_content else 'EMPTY',
                'detail': 'Bucket accessible with objects' if has_content else 'Bucket accessible, empty',
            })
        except Exception:
            checks.append({'status': 'NOT ACCESSIBLE', 'detail': 'Bucket not accessible'})
        try:
            c['s3'].get_bucket_website(Bucket=name)
            checks.append({'status': 'WEBSITE ACTIVE', 'detail': 'Static website hosting enabled'})
        except Exception:
            pass
        result.append({'bucket_name': name, 'checks': checks})
    return result


def health_rds(c):
    result = []
    try:
        for cluster in c['rds'].describe_db_clusters()['DBClusters']:
            s = cluster['Status']
            result.append({
                'status': 'AVAILABLE' if s == 'available' else s.upper(),
                'resource': cluster['DBClusterIdentifier'],
                'type': 'cluster',
            })
        for instance in c['rds'].describe_db_instances()['DBInstances']:
            s = instance['DBInstanceStatus']
            result.append({
                'status': 'AVAILABLE' if s == 'available' else s.upper(),
                'resource': instance['DBInstanceIdentifier'],
                'type': 'instance',
            })
    except Exception:
        pass
    return result


def health_cognito(c):
    result = []
    try:
        pools = c['cognito'].list_user_pools(MaxResults=60)['UserPools']
        for pool in pools:
            try:
                d = c['cognito'].describe_user_pool(UserPoolId=pool['Id'])['UserPool']
                result.append({
                    'status': 'ACTIVE',
                    'resource': pool['Name'],
                    'detail': f"{d.get('EstimatedNumberOfUsers', 0)} users",
                })
            except Exception:
                result.append({'status': 'NO ACCESS', 'resource': pool['Name'], 'detail': ''})
    except Exception:
        pass
    return result


def health_stepfunctions(c):
    result = []
    try:
        for sm in c['stepfunctions'].list_state_machines()['stateMachines']:
            try:
                d = c['stepfunctions'].describe_state_machine(stateMachineArn=sm['stateMachineArn'])
                s = d.get('status', 'N/A')
                result.append({
                    'status': 'ACTIVE' if s == 'ACTIVE' else s,
                    'resource': sm['name'],
                    'detail': sm['type'],
                })
            except Exception:
                result.append({'status': 'NO ACCESS', 'resource': sm['name'], 'detail': ''})
    except Exception:
        pass
    return result


def health_waf(c):
    result = []
    try:
        for acl in c['wafv2'].list_web_acls(Scope='REGIONAL')['WebACLs']:
            resources = []
            for rt in ['APPLICATION_LOAD_BALANCER', 'API_GATEWAY']:
                try:
                    resources.extend(
                        c['wafv2'].list_resources_for_web_acl(WebACLArn=acl['ARN'], ResourceType=rt)['ResourceArns']
                    )
                except Exception:
                    pass
            if resources:
                result.append({'status': 'ACTIVE', 'resource': acl['Name'], 'detail': f'protecting {len(resources)} resource(s)'})
            else:
                result.append({'status': 'WARNING', 'resource': acl['Name'], 'detail': 'not protecting any resource'})
    except Exception:
        pass
    return result


def health_elasticbeanstalk(c):
    result = []
    try:
        for env in c['elasticbeanstalk'].describe_environments()['Environments']:
            h = env.get('Health', 'N/A')
            hs = env.get('HealthStatus', 'N/A')
            result.append({'status': h.upper(), 'resource': env['EnvironmentName'], 'detail': hs})
    except Exception:
        pass
    return result


def health_elasticache(c):
    result = []
    try:
        for cluster in c['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True)['CacheClusters']:
            s = cluster.get('CacheClusterStatus', 'N/A')
            nodes = cluster.get('CacheNodes', [])
            ok = sum(1 for n in nodes if n.get('CacheNodeStatus') == 'available')
            result.append({
                'status': 'AVAILABLE' if s == 'available' else s.upper(),
                'resource': cluster['CacheClusterId'],
                'detail': f"{cluster.get('Engine', '')}, {ok}/{len(nodes)} nodes",
            })
    except Exception:
        pass
    return result


def health_ecr(c):
    result = []
    try:
        for repo in c['ecr'].describe_repositories()['repositories']:
            try:
                imgs = c['ecr'].describe_images(repositoryName=repo['repositoryName'])['imageDetails']
                if imgs:
                    latest = max(imgs, key=lambda x: str(x.get('imagePushedAt', '')))
                    tags = latest.get('imageTags', ['<untagged>'])
                    result.append({
                        'status': 'OK',
                        'resource': repo['repositoryName'],
                        'detail': f"{len(imgs)} image(s), latest: {', '.join(tags)}",
                    })
                else:
                    result.append({'status': 'EMPTY', 'resource': repo['repositoryName'], 'detail': ''})
            except Exception:
                result.append({'status': 'NO ACCESS', 'resource': repo['repositoryName'], 'detail': ''})
    except Exception:
        pass
    return result


def health_codecommit(c):
    result = []
    try:
        for r in c['codecommit'].list_repositories()['repositories']:
            try:
                meta = c['codecommit'].get_repository(repositoryName=r['repositoryName'])['repositoryMetadata']
                branch = meta.get('defaultBranch', 'N/A')
                result.append({'status': 'OK', 'resource': r['repositoryName'], 'detail': f'branch: {branch}'})
            except Exception:
                result.append({'status': 'NO ACCESS', 'resource': r['repositoryName'], 'detail': ''})
    except Exception:
        pass
    return result


HEALTH_FUNCTIONS = {
    'Lambda': health_lambda, 'DynamoDB': health_dynamodb, 'API Gateway': health_apigateway,
    'SQS': health_sqs, 'SNS': health_sns, 'S3': health_s3, 'RDS': health_rds,
    'Cognito': health_cognito, 'Step Functions': health_stepfunctions, 'WAF': health_waf,
    'Elastic Beanstalk': health_elasticbeanstalk, 'ElastiCache': health_elasticache,
    'ECR': health_ecr, 'CodeCommit': health_codecommit,
}

print("Health collectors loaded")
```

```python
# === Health renderers ===

def render_lambda_health(data):
    html = html_title('Lambda — Health')
    html += html_section('Functions')
    rows = [
        (html_status(item['status'], ''), item['resource'])
        for item in data.get('functions', [])
    ]
    if rows:
        html += html_table(['Status', 'Function'], rows, widths=['140px', 'auto'])
    else:
        html += html_empty('No Lambda functions found.')

    html += html_section('Event source mappings')
    mapping_rows = [
        (html_status(item['status'], ''), item['function'], item['source'])
        for item in data.get('mappings', [])
    ]
    if mapping_rows:
        html += html_table(['Status', 'Target function', 'Event source'], mapping_rows, widths=['140px', '240px', 'auto'])
    else:
        html += html_empty('No event source mappings found.')
    return html


def render_api_gateway_health(data):
    html = html_title('API Gateway — Health')
    if not data:
        html += html_empty('No APIs found.')
        return html
    for api in data:
        html += html_section(api['api_name'])
        rows = []
        for check in api['checks']:
            status = check['status']
            detail = check['detail']
            if status == 'DEPLOYED':
                stage_name = detail.replace('stage: ', '')
                rows.append((html_status(status, ''), 'Deployed stage', stage_name))
            elif status == 'NOT DEPLOYED':
                rows.append((html_status(status, ''), 'Deployment', 'No deployed stages'))
            elif status == 'OK':
                rows.append((html_status(status, ''), 'Configured integrations', detail))
            elif status == 'WARNING':
                rows.append((html_status(status, ''), 'Incomplete integrations', detail))
            else:
                rows.append((html_status(status, ''), 'Access', detail or status))
        if rows:
            html += html_table(['Status', 'Element', 'Result'], rows, widths=['140px', '220px', 'auto'])
    return html


def render_s3_health(data):
    html = html_title('S3 — Health')
    if not data:
        html += html_empty('No S3 buckets found.')
        return html
    for bucket in data:
        html += html_section(bucket['bucket_name'])
        rows = [(html_status(check['status'], ''), check.get('detail') or check['status'].capitalize()) for check in bucket['checks']]
        if rows:
            html += html_table(['Status', 'Check'], rows, widths=['140px', 'auto'])
        else:
            html += html_empty('No checks available.')
    return html



def render_rds_health(data):
    html = html_title('RDS — Health')
    if not data:
        html += html_empty('No RDS resources found.')
        return html
    rows = []
    for item in data:
        resource = item['resource']
        if item.get('type'):
            resource = f"{item['type'].capitalize()}: {resource}"
        rows.append((html_status(item['status'], ''), f'{html_escape(resource)}'))
    html += html_table(['Status', 'Resource'], rows, widths=['140px', 'auto'])
    return html

def render_resource_health(service, data):
    html = html_title(f'{service} — Health')
    if not data:
        html += html_empty(f'No {service} resources found.')
        return html
    rows = []
    for item in data:
        resource = item['resource']
        if item.get('type'):
            resource = f"{item['type'].capitalize()}: {resource}"
        detail = item.get('detail', '')
        rows.append((html_status(item['status'], ''), f'<strong>{resource}</strong>', detail))
    html += html_table(['Status', 'Resource', 'Details'], rows, widths=['140px', '280px', 'auto'])
    return html


def render_health_output(service, data):
    if service == 'Lambda':
        return render_lambda_health(data)
    if service == 'API Gateway':
        return render_api_gateway_health(data)
    if service == 'S3':
        return render_s3_health(data)
    if service == 'RDS':
        return render_rds_health(data)
    return render_resource_health(service, data)
```

```python
# === Health summaries ===

# --- Health summary functions (for overview table) ---
def _health_rollup(healthy=0, warning=0, critical=0, info=0, ok_status='OK', empty_status='INFO'):
    if critical > 0:
        return 'FAILED'
    if warning > 0:
        return 'WARNING'
    if healthy > 0:
        return ok_status
    return empty_status

def health_summary_lambda(c):
    fns = c['lambda'].list_functions()['Functions']
    checked = 0
    healthy = 0

    for fn in fns:
        try:
            cfg = c['lambda'].get_function_configuration(FunctionName=fn['FunctionName'])
            checked += 1
            if cfg.get('State') == 'Active' and cfg.get('LastUpdateStatus') == 'Successful':
                healthy += 1
        except Exception:
            continue

    if checked == 0:
        return ('INFO', 'No accessible functions')

    status = 'HEALTHY' if healthy == checked else 'WARNING'
    return (status, f'{healthy}/{checked} accessible functions healthy')

def health_summary_dynamodb(c):
    tables = c['dynamodb'].list_tables()['TableNames']
    if not tables:
        return ('INFO', 'No tables')
    active = 0
    for t in tables:
        try:
            if c['dynamodb'].describe_table(TableName=t)['Table']['TableStatus'] == 'ACTIVE':
                active += 1
        except Exception:
            pass
    return ('ACTIVE' if active == len(tables) else 'WARNING', f'{active}/{len(tables)} tables active')

def health_summary_apigateway(c):
    try:
        apis = c['apigateway'].get_rest_apis()['items']
        if not apis:
            return ('WARNING', 'No APIs')
        healthy = 0
        warning = 0
        checked = 0
        for api in apis:
            try:
                api_id = api['id']
                stages = c['apigateway'].get_stages(restApiId=api_id)['item']
                recursos = c['apigateway'].get_resources(restApiId=api_id, limit=500)['items']
                missing = 0
                for rec in recursos:
                    for method in rec.get('resourceMethods', {}):
                        if method == 'OPTIONS':
                            continue
                        try:
                            integ = c['apigateway'].get_integration(restApiId=api_id, resourceId=rec['id'], httpMethod=method)
                            if not (integ.get('uri') or integ.get('type') == 'MOCK'):
                                missing += 1
                        except Exception:
                            missing += 1
                checked += 1
                if stages and missing == 0:
                    healthy += 1
                else:
                    warning += 1
            except Exception:
                continue
        if checked == 0:
            return ('INFO', 'No accessible APIs')
        status = _health_rollup(healthy=healthy, warning=warning, ok_status='DEPLOYED')
        return (status, f'{healthy}/{checked} APIs deployed with integrations')
    except Exception:
        return ('INFO', 'No API Gateway resources')

def health_summary_sqs(c):
    urls = c['sqs'].list_queues().get('QueueUrls', [])
    if not urls:
        return ('INFO', 'No queues')
    ok = 0
    warning = 0
    critical = 0
    checked = 0
    dlq_msgs = 0
    for u in urls:
        name = u.split('/')[-1]
        try:
            attrs = c['sqs'].get_queue_attributes(QueueUrl=u, AttributeNames=['ApproximateNumberOfMessages', 'ApproximateNumberOfMessagesNotVisible'])['Attributes']
            visible = int(attrs.get('ApproximateNumberOfMessages', 0))
            checked += 1
            if 'dlq' in name.lower():
                dlq_msgs += visible
                if visible > 0:
                    critical += 1
                else:
                    ok += 1
            elif visible > 10:
                warning += 1
            else:
                ok += 1
        except Exception:
            continue
    if checked == 0:
        return ('INFO', 'No accessible queues')
    status = _health_rollup(healthy=ok, warning=warning, critical=critical)
    return (status, f'{ok}/{checked} queues OK, {dlq_msgs} DLQ messages')

def health_summary_sns(c):
    topics = c['sns'].list_topics()['Topics']
    checked = 0
    ok = 0
    warning = 0
    for t in topics:
        try:
            a = c['sns'].get_topic_attributes(TopicArn=t['TopicArn'])['Attributes']
            checked += 1
            confirmed = int(a.get('SubscriptionsConfirmed', 0))
            pending = int(a.get('SubscriptionsPending', 0))
            if confirmed > 0 and pending == 0:
                ok += 1
            else:
                warning += 1
        except Exception:
            continue
    if checked == 0:
        return ('INFO', 'No accessible topics')
    status = _health_rollup(healthy=ok, warning=warning)
    return (status, f'{ok}/{checked} topics with active subscriptions')

def health_summary_s3(c):
    buckets = c['s3'].list_buckets()['Buckets']
    if not buckets:
        return ('INFO', 'No buckets')
    accessible = 0
    empty = 0
    inaccessible = 0
    for b in buckets:
        try:
            c['s3'].head_bucket(Bucket=b['Name'])
            objs = c['s3'].list_objects_v2(Bucket=b['Name'], MaxKeys=1)
            accessible += 1
            if objs.get('KeyCount', 0) == 0:
                empty += 1
        except Exception:
            inaccessible += 1
    status = _health_rollup(healthy=accessible - empty, warning=empty, critical=inaccessible)
    return (status, f'{accessible}/{len(buckets)} accessible, {empty} empty')

def health_summary_rds(c):
    try:
        clusters = c['rds'].describe_db_clusters()['DBClusters']
    except Exception:
        clusters = []
    try:
        insts = c['rds'].describe_db_instances()['DBInstances']
    except Exception:
        insts = []
    total = len(clusters) + len(insts)
    if total == 0:
        return ('INFO', 'No RDS resources')
    ok_clusters = sum(1 for cl in clusters if cl.get('Status') == 'available')
    ok_insts = sum(1 for i in insts if i.get('DBInstanceStatus') == 'available')
    ok = ok_clusters + ok_insts
    status = 'AVAILABLE' if ok == total else 'WARNING'
    return (status, f'{ok}/{total} RDS resources available')

def health_summary_cognito(c):
    try:
        pools = c['cognito'].list_user_pools(MaxResults=60)['UserPools']
        if not pools:
            return ('INFO', 'No pools')
        accessible = 0
        for p in pools:
            try:
                c['cognito'].describe_user_pool(UserPoolId=p['Id'])
                accessible += 1
            except Exception:
                continue
        if accessible == 0:
            return ('INFO', 'No accessible pools')
        status = 'ACTIVE' if accessible == len(pools) else 'INFO'
        return (status, f'{accessible}/{len(pools)} pools accessible')
    except Exception:
        return ('INFO', 'No Cognito resources')

def health_summary_stepfunctions(c):
    try:
        sms = c['stepfunctions'].list_state_machines()['stateMachines']
        if not sms:
            return ('INFO', 'No state machines')
        active = 0
        checked = 0
        for sm in sms:
            try:
                d = c['stepfunctions'].describe_state_machine(stateMachineArn=sm['stateMachineArn'])
                checked += 1
                if d.get('status') == 'ACTIVE':
                    active += 1
            except Exception:
                continue
        if checked == 0:
            return ('INFO', 'No accessible state machines')
        return ('ACTIVE' if active == checked else 'WARNING', f'{active}/{checked} state machines active')
    except Exception:
        return ('INFO', 'No Step Functions resources')

def health_summary_waf(c):
    try:
        acls = c['wafv2'].list_web_acls(Scope='REGIONAL')['WebACLs']
        if not acls:
            return ('INFO', 'No Web ACLs')
        protected = 0
        for acl in acls:
            resources = []
            for rt in ['APPLICATION_LOAD_BALANCER', 'API_GATEWAY']:
                try:
                    resources.extend(c['wafv2'].list_resources_for_web_acl(WebACLArn=acl['ARN'], ResourceType=rt)['ResourceArns'])
                except Exception:
                    pass
            if resources:
                protected += 1
        status = 'ACTIVE' if protected == len(acls) else 'WARNING'
        return (status, f'{protected}/{len(acls)} Web ACL(s) protecting resources')
    except Exception:
        return ('INFO', 'No WAF resources')

def health_summary_eb(c):
    try:
        envs = c['elasticbeanstalk'].describe_environments()['Environments']
        if not envs:
            return ('INFO', 'No environments')
        green = sum(1 for env in envs if env.get('Health') == 'Green')
        red = sum(1 for env in envs if env.get('Health') == 'Red')
        yellow = sum(1 for env in envs if env.get('Health') == 'Yellow')
        if red > 0:
            status = 'RED'
        elif yellow > 0 or green < len(envs):
            status = 'WARNING'
        else:
            status = 'GREEN'
        return (status, f'{green}/{len(envs)} environments green')
    except Exception:
        return ('INFO', 'No Elastic Beanstalk resources')

def health_summary_elasticache(c):
    try:
        cls = c['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True)['CacheClusters']
        if not cls:
            return ('INFO', 'No clusters')
        ok = 0
        for cl in cls:
            nodes = cl.get('CacheNodes', [])
            nodes_ok = sum(1 for n in nodes if n.get('CacheNodeStatus') == 'available')
            if cl.get('CacheClusterStatus') == 'available' and nodes_ok == len(nodes):
                ok += 1
        return ('AVAILABLE' if ok == len(cls) else 'WARNING', f'{ok}/{len(cls)} clusters available')
    except Exception:
        return ('INFO', 'No ElastiCache resources')

def health_summary_ecr(c):
    try:
        repos = c['ecr'].describe_repositories()['repositories']
        if not repos:
            return ('INFO', 'No repositories')
        ok = 0
        empty = 0
        checked = 0
        for repo in repos:
            try:
                imgs = c['ecr'].describe_images(repositoryName=repo['repositoryName'])['imageDetails']
                checked += 1
                if imgs:
                    ok += 1
                else:
                    empty += 1
            except Exception:
                continue
        if checked == 0:
            return ('INFO', 'No accessible repositories')
        status = _health_rollup(healthy=ok, warning=empty)
        return (status, f'{ok}/{checked} repositories with images')
    except Exception:
        return ('INFO', 'No ECR resources')

def health_summary_codecommit(c):
    try:
        repos = c['codecommit'].list_repositories()['repositories']
        if not repos:
            return ('INFO', 'No repositories')
        accessible = 0
        for r in repos:
            try:
                c['codecommit'].get_repository(repositoryName=r['repositoryName'])
                accessible += 1
            except Exception:
                continue
        if accessible == 0:
            return ('INFO', 'No accessible repositories')
        return ('OK', f'{accessible}/{len(repos)} repositories accessible')
    except Exception:
        return ('INFO', 'No CodeCommit resources')
```

```python
# === Metrics overview ===

def render_metrics_overview():
    html = html_title('Metrics Overview')
    html += html_metric_range(include_period=True)

    # Accessible Lambda functions (cached for reuse)
    fns = [f['FunctionName'] for f in clients['lambda'].list_functions()['Functions']
           if _safe_check_lambda(f['FunctionName'])]
    apis = clients['apigateway'].get_rest_apis()['items']
    queue_urls = clients['sqs'].list_queues().get('QueueUrls', [])
    queue_names = [u.split('/')[-1] for u in queue_urls]
    topics = clients['sns'].list_topics()['Topics']

    # --- 1. Activity ---
    html += html_section('Activity')
    activity = []

    total_invocations = sum(get_metric('AWS/Lambda', 'Invocations',
        [{'Name': 'FunctionName', 'Value': fn}], 'Sum') for fn in fns)
    activity.append(('Lambda', 'Invocations', f'{int(total_invocations):,}'))

    total_requests = sum(get_metric('AWS/ApiGateway', 'Count',
        [{'Name': 'ApiName', 'Value': a['name']}], 'Sum') for a in apis)
    activity.append(('API Gateway', 'Requests', f'{int(total_requests):,}'))

    tables = clients['dynamodb'].list_tables()['TableNames']
    total_rcu = sum(get_metric('AWS/DynamoDB', 'ConsumedReadCapacityUnits',
        [{'Name': 'TableName', 'Value': t}], 'Sum') for t in tables)
    total_wcu = sum(get_metric('AWS/DynamoDB', 'ConsumedWriteCapacityUnits',
        [{'Name': 'TableName', 'Value': t}], 'Sum') for t in tables)
    activity.append(('DynamoDB', 'Read/Write capacity units', f'{int(total_rcu):,} RCU / {int(total_wcu):,} WCU'))

    total_sent = sum(get_metric('AWS/SQS', 'NumberOfMessagesSent',
        [{'Name': 'QueueName', 'Value': q}], 'Sum') for q in queue_names)
    activity.append(('SQS', 'Messages sent', f'{int(total_sent):,}'))

    total_published = sum(get_metric('AWS/SNS', 'NumberOfMessagesPublished',
        [{'Name': 'TopicName', 'Value': t['TopicArn'].split(':')[-1]}], 'Sum') for t in topics)
    activity.append(('SNS', 'Notifications published', f'{int(total_published):,}'))

    try:
        pools = clients['cognito'].list_user_pools(MaxResults=60).get('UserPools', [])
        total_signins = 0
        total_signups = 0
        for pool in pools:
            pool_id = pool['Id']
            app_clients = clients['cognito'].list_user_pool_clients(UserPoolId=pool_id, MaxResults=60).get('UserPoolClients', [])
            for app in app_clients:
                dims = [
                    {'Name': 'UserPool', 'Value': pool_id},
                    {'Name': 'UserPoolClient', 'Value': app['ClientId']}
                ]
                total_signins += get_metric('AWS/Cognito', 'SignInSuccesses', dims, 'Sum')
                total_signups += get_metric('AWS/Cognito', 'SignUpSuccesses', dims, 'Sum')
        activity.append(('Cognito', 'Sign-ins / sign-ups', f'{int(total_signins):,} sign-ins / {int(total_signups):,} sign-ups'))
    except Exception:
        pass

    try:
        rds_instances = clients['rds'].describe_db_instances().get('DBInstances', [])
        max_connections = max((get_metric('AWS/RDS', 'DatabaseConnections',
            [{'Name': 'DBInstanceIdentifier', 'Value': db['DBInstanceIdentifier']}], 'Maximum') for db in rds_instances), default=0)
        activity.append(('RDS', 'Max database connections', f'{int(max_connections):,}'))
    except Exception:
        pass

    try:
        clusters = clients['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True).get('CacheClusters', [])
        total_lookups = 0
        for cluster in clusters:
            for node in cluster.get('CacheNodes', []):
                node_id = f"{cluster['CacheClusterId']}/{node['CacheNodeId']}"
                dims = [{'Name': 'CacheClusterId', 'Value': cluster['CacheClusterId']}, {'Name': 'CacheNodeId', 'Value': node_id}]
                total_lookups += get_metric('AWS/ElastiCache', 'GetHits', dims, 'Sum') + get_metric('AWS/ElastiCache', 'GetMisses', dims, 'Sum')
        activity.append(('ElastiCache', 'Cache lookups', f'{int(total_lookups):,}'))
    except Exception:
        pass

    try:
        envs = clients['elasticbeanstalk'].describe_environments(IncludeDeleted=False).get('Environments', [])
        total_app_requests = sum(get_metric('AWS/ElasticBeanstalk', 'ApplicationRequestsTotal',
            [{'Name': 'EnvironmentName', 'Value': env['EnvironmentName']}], 'Sum') for env in envs)
        activity.append(('Elastic Beanstalk', 'Application requests', f'{int(total_app_requests):,}'))
    except Exception:
        pass

    try:
        sms = clients['stepfunctions'].list_state_machines()['stateMachines']
        total_exec = sum(get_metric('AWS/States', 'ExecutionsStarted',
            [{'Name': 'StateMachineArn', 'Value': sm['stateMachineArn']}], 'Sum') for sm in sms)
        activity.append(('Step Functions', 'Executions started', f'{int(total_exec):,}'))
    except Exception:
        pass

    try:
        acls = clients['wafv2'].list_web_acls(Scope='REGIONAL')['WebACLs']
        region = cloudwatch.meta.region_name or 'us-east-1'
        for acl in acls:
            try:
                detail = clients['wafv2'].get_web_acl(Name=acl['Name'], Scope='REGIONAL', Id=acl['Id'])['WebACL']
                mn = detail.get('VisibilityConfig', {}).get('MetricName', acl['Name'])
            except Exception:
                mn = acl['Name']
            dims = [{'Name': 'WebACL', 'Value': mn}, {'Name': 'Region', 'Value': region}, {'Name': 'Rule', 'Value': 'ALL'}]
            allowed = int(get_metric('AWS/WAFV2', 'AllowedRequests', dims, 'Sum'))
            blocked = int(get_metric('AWS/WAFV2', 'BlockedRequests', dims, 'Sum'))
            activity.append(('WAF', f'Requests ({acl["Name"]})', f'{allowed:,} allowed, {blocked:,} blocked'))
    except Exception:
        pass

    html += html_note('This table summarizes high-level request, message, execution and connection activity for services with comparable operational metrics.')
    html += html_table(['Service', 'Metric', 'Value'], activity)

    # --- 2. Errors ---
    html += html_section('Errors')
    errors = []
    error_values = []

    lambda_errors = sum(get_metric('AWS/Lambda', 'Errors',
        [{'Name': 'FunctionName', 'Value': fn}], 'Sum') for fn in fns)
    lambda_throttles = sum(get_metric('AWS/Lambda', 'Throttles',
        [{'Name': 'FunctionName', 'Value': fn}], 'Sum') for fn in fns)
    if lambda_errors > 0 or lambda_throttles > 0:
        errors.append(('Lambda', f'{int(lambda_errors)} errors, {int(lambda_throttles)} throttles'))
        error_values.append(('Lambda', int(lambda_errors) + int(lambda_throttles)))

    for api in apis:
        e4 = int(get_metric('AWS/ApiGateway', '4XXError', [{'Name': 'ApiName', 'Value': api['name']}], 'Sum'))
        e5 = int(get_metric('AWS/ApiGateway', '5XXError', [{'Name': 'ApiName', 'Value': api['name']}], 'Sum'))
        if e4 > 0 or e5 > 0:
            errors.append((f"API GW ({api['name']})", f'{e4} client (4XX), {e5} server (5XX)'))
            error_values.append((f"API GW ({api['name']})", e4 + e5))

    for q in queue_names:
        if 'dlq' in q.lower():
            try:
                attrs = clients['sqs'].get_queue_attributes(
                    QueueUrl=[u for u in queue_urls if q in u][0],
                    AttributeNames=['ApproximateNumberOfMessages']
                )['Attributes']
                dlq_count = int(attrs.get('ApproximateNumberOfMessages', 0))
                if dlq_count > 0:
                    errors.append((f'SQS DLQ ({q})', f'{dlq_count} failed messages'))
                    error_values.append((f'SQS DLQ ({q})', dlq_count))
            except Exception:
                pass

    sns_failed = sum(get_metric('AWS/SNS', 'NumberOfNotificationsFailed',
        [{'Name': 'TopicName', 'Value': t['TopicArn'].split(':')[-1]}], 'Sum') for t in topics)
    if sns_failed > 0:
        errors.append(('SNS', f'{int(sns_failed)} failed notifications'))
        error_values.append(('SNS', int(sns_failed)))

    try:
        cognito_throttles = 0
        pools = clients['cognito'].list_user_pools(MaxResults=60).get('UserPools', [])
        for pool in pools:
            pool_id = pool['Id']
            app_clients = clients['cognito'].list_user_pool_clients(UserPoolId=pool_id, MaxResults=60).get('UserPoolClients', [])
            for app in app_clients:
                dims = [
                    {'Name': 'UserPool', 'Value': pool_id},
                    {'Name': 'UserPoolClient', 'Value': app['ClientId']}
                ]
                cognito_throttles += get_metric('AWS/Cognito', 'SignInThrottles', dims, 'Sum')
                cognito_throttles += get_metric('AWS/Cognito', 'SignUpThrottles', dims, 'Sum')
        if cognito_throttles > 0:
            errors.append(('Cognito', f'{int(cognito_throttles)} authentication throttles'))
            error_values.append(('Cognito', int(cognito_throttles)))
    except Exception:
        pass

    if errors:
        html += html_table(['Service', 'Errors detected'], errors)
        fig = go.Figure(data=[go.Bar(
            x=[e[0] for e in error_values],
            y=[e[1] for e in error_values],
            marker_color='#c0392b'
        )])
        fig.update_layout(title='Errors by service', height=300, margin=dict(t=40, b=60, l=40, r=20),
                         yaxis_title='Count', template='plotly_white')
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_success('No errors detected in the selected time range.')

    # --- 3. Performance ---
    html += html_section('Performance')

    html += html_subsection('API Gateway latency breakdown')
    perf = []
    for api in apis:
        lat = get_metric('AWS/ApiGateway', 'Latency', [{'Name': 'ApiName', 'Value': api['name']}], 'Average')
        int_lat = get_metric('AWS/ApiGateway', 'IntegrationLatency', [{'Name': 'ApiName', 'Value': api['name']}], 'Average')
        overhead = round(lat - int_lat, 1) if lat > 0 and int_lat > 0 else '-'
        perf.append((api['name'], f'{lat} ms', f'{int_lat} ms', f'{overhead} ms'))
    if perf:
        html += html_table(['API', 'Total latency', 'Integration latency', 'API Gateway overhead'], perf)

    html += html_subsection('Lambda function duration')
    lambda_perf = []
    for fn in fns:
        avg = get_metric('AWS/Lambda', 'Duration', [{'Name': 'FunctionName', 'Value': fn}], 'Average')
        mx = get_metric('AWS/Lambda', 'Duration', [{'Name': 'FunctionName', 'Value': fn}], 'Maximum')
        if avg > 0:
            lambda_perf.append((fn, f'{avg} ms', f'{mx} ms'))
    if lambda_perf:
        html += html_table(['Function', 'Avg duration', 'Max duration'], lambda_perf)

    # --- 4. Alarms ---
    html += html_section('CloudWatch Alarms')
    try:
        alarms = cloudwatch.describe_alarms()
        metric_alarms = alarms.get('MetricAlarms', [])
        if metric_alarms:
            alarm_rows = []
            for a in metric_alarms:
                state = a.get('StateValue', 'N/A')
                alarm_rows.append((html_status(state, ''), a['AlarmName'],
                                  f"{a.get('Namespace', '')}/{a.get('MetricName', '')}"))
            html += alarm_table_note()
            html += html_table(['State', 'Alarm', 'Metric'], alarm_rows)
        else:
            html += html_empty('No alarms configured.')
    except Exception:
        html += html_empty('Unable to access alarms.')

    return html

def alarm_table_note():
    return html_note('OK = alarm condition not triggered. ALARM = condition triggered.')

def _safe_check_lambda(fn_name):
    try:
        clients['lambda'].get_function_configuration(FunctionName=fn_name)
        return True
    except Exception:
        return False
```

```python

def _sum_metric_series(series_list):
    totals = {}
    for series in series_list:
        for timestamp, value in series:
            totals[timestamp] = totals.get(timestamp, 0) + value
    return sorted(totals.items(), key=lambda item: item[0])


def _series_xy(series):
    return [point[0] for point in series], [point[1] for point in series]


def render_metric_line_chart(title, traces, yaxis_title):
    visible_traces = [trace for trace in traces if trace.get('x') and any(value != 0 for value in trace.get('y', []))]
    if not visible_traces:
        return html_empty(f'No time-series data detected for {title.lower()} in the selected range.')

    fig = go.Figure()
    for trace in visible_traces:
        fig.add_trace(go.Scatter(
            x=trace['x'],
            y=trace['y'],
            mode='lines+markers',
            name=trace['name'],
            line=dict(color=trace.get('color'), width=2),
            marker=dict(size=4),
        ))

    show_legend = len(visible_traces) > 1
    fig.update_layout(
        title=dict(text=title, font=dict(size=16, color='#1f2937'), x=0.01, xanchor='left'),
        height=300,
        margin=dict(t=52, b=72 if show_legend else 42, l=60, r=24),
        template='plotly_white',
        yaxis_title=yaxis_title,
        hovermode='x unified',
        showlegend=show_legend,
        legend=dict(
            orientation='h',
            yanchor='top', y=-0.18,
            xanchor='left', x=0,
            font=dict(size=11),
            itemwidth=30,
        ),
        paper_bgcolor='white',
        plot_bgcolor='white',
        font=dict(size=12, color='#0b2447'),
    )
    fig.update_xaxes(title_text='', tickformat='%b %d', showgrid=True, gridcolor='#e8eef7')
    fig.update_yaxes(showgrid=True, gridcolor='#e8eef7', zeroline=True, zerolinecolor='#d6deeb')
    return fig.to_html(full_html=False, include_plotlyjs='cdn')


# === Metrics service renderers ===

# --- Lambda + DynamoDB ---


def render_dual_axis_line_chart(title, left_trace, right_trace, left_title, right_title):
    left_has_data = left_trace.get('x') and any(value != 0 for value in left_trace.get('y', []))
    right_has_data = right_trace.get('x') and any(value != 0 for value in right_trace.get('y', []))
    if not left_has_data and not right_has_data:
        return html_empty(f'No time-series data detected for {title.lower()} in the selected range.')

    fig = go.Figure()
    if left_has_data:
        fig.add_trace(go.Scatter(
            x=left_trace['x'], y=left_trace['y'], mode='lines+markers', name=left_trace['name'],
            line=dict(color=left_trace.get('color', '#2E5090'), width=2), marker=dict(size=4), yaxis='y'
        ))
    if right_has_data:
        fig.add_trace(go.Scatter(
            x=right_trace['x'], y=right_trace['y'], mode='lines+markers', name=right_trace['name'],
            line=dict(color=right_trace.get('color', '#e67e22'), width=2), marker=dict(size=4), yaxis='y2'
        ))

    fig.update_layout(
        title=dict(text=title, font=dict(size=16, color='#1f2937'), x=0.01, xanchor='left'),
        height=310,
        margin=dict(t=52, b=72, l=60, r=70),
        template='plotly_white',
        hovermode='x unified',
        showlegend=True,
        legend=dict(orientation='h', yanchor='top', y=-0.18, xanchor='left', x=0, font=dict(size=11)),
        yaxis=dict(title=left_title, showgrid=True, gridcolor='#e8eef7', zeroline=True, zerolinecolor='#d6deeb'),
        yaxis2=dict(title=right_title, overlaying='y', side='right', showgrid=False, zeroline=False),
        paper_bgcolor='white', plot_bgcolor='white', font=dict(size=12, color='#0b2447'),
    )
    fig.update_xaxes(title_text='', tickformat='%b %d', showgrid=True, gridcolor='#e8eef7')
    return fig.to_html(full_html=False, include_plotlyjs='cdn')



def render_metrics_lambda():
    fns = [f['FunctionName'] for f in clients['lambda'].list_functions()['Functions']
           if _safe_check_lambda(f['FunctionName'])]
    metrics = [
        ('Invocations', 'Sum', 'Invocations'), ('Errors', 'Sum', 'Errors'),
        ('Throttles', 'Sum', 'Throttles'), ('DeadLetterErrors', 'Sum', 'DLQ Errors'),
        ('Duration', 'Average', 'Avg duration (ms)'), ('Duration', 'Maximum', 'Max duration (ms)'),
        ('ConcurrentExecutions', 'Maximum', 'Max concurrency'),
    ]
    rows = []
    function_data = []

    for fn in fns:
        dims = [{'Name': 'FunctionName', 'Value': fn}]
        values = [get_metric('AWS/Lambda', m, dims, s) for m, s, _ in metrics]
        invocations, errors, throttles, dlq_errors, avg_duration, max_duration, max_concurrency = values
        error_rate = round((errors / invocations) * 100, 2) if invocations > 0 else 0

        function_data.append({
            'name': fn,
            'invocations': invocations,
            'errors': errors,
            'throttles': throttles,
            'dlq_errors': dlq_errors,
            'avg_duration': avg_duration,
            'max_duration': max_duration,
            'max_concurrency': max_concurrency,
            'error_rate': error_rate,
        })

        rows.append([
            fn,
            f'{int(invocations):,}',
            f'{int(errors):,}',
            f'{int(throttles):,}',
            f'{int(dlq_errors):,}',
            f'{avg_duration}',
            f'{max_duration}',
            f'{int(max_concurrency):,}',
            f'{error_rate}%'
        ])

    html = html_title('Lambda - Metrics')
    html += html_metric_range()
    html += html_table(
        ['Function'] + [n for _, _, n in metrics] + ['Error rate'],
        rows
    )

    active_fns = [item for item in function_data if item['invocations'] > 0]
    if active_fns:
        active_fns = sorted(active_fns, key=lambda item: item['invocations'], reverse=True)
        fig = go.Figure(data=[go.Bar(
            x=[item['name'] for item in active_fns],
            y=[item['invocations'] for item in active_fns],
            marker_color='#2E5090'
        )])
        fig.update_layout(
            title='Traffic: invocations per function',
            height=320,
            margin=dict(t=45, b=90, l=50, r=20),
            template='plotly_white',
            yaxis_title='Invocations',
            xaxis_tickangle=-45
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No Lambda invocations detected in the selected range.')

    reliability_fns = [item for item in function_data if item['errors'] > 0 or item['throttles'] > 0 or item['dlq_errors'] > 0]
    if reliability_fns:
        reliability_fns = sorted(
            reliability_fns,
            key=lambda item: item['errors'] + item['throttles'] + item['dlq_errors'],
            reverse=True
        )
        fig = go.Figure()
        fig.add_trace(go.Bar(
            name='Errors',
            x=[item['name'] for item in reliability_fns],
            y=[item['errors'] for item in reliability_fns],
            marker_color='#c0392b'
        ))
        fig.add_trace(go.Bar(
            name='Throttles',
            x=[item['name'] for item in reliability_fns],
            y=[item['throttles'] for item in reliability_fns],
            marker_color='#e67e22'
        ))
        fig.add_trace(go.Bar(
            name='DLQ errors',
            x=[item['name'] for item in reliability_fns],
            y=[item['dlq_errors'] for item in reliability_fns],
            marker_color='#8e44ad'
        ))
        fig.update_layout(
            title='Reliability: errors, throttles and DLQ errors',
            barmode='stack',
            height=320,
            margin=dict(t=45, b=90, l=50, r=20),
            template='plotly_white',
            yaxis_title='Count',
            xaxis_tickangle=-45
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_success('No Lambda errors, throttles or DLQ errors detected in the selected range.')

    duration_fns = [item for item in function_data if item['avg_duration'] > 0 or item['max_duration'] > 0]
    if duration_fns:
        duration_fns = sorted(duration_fns, key=lambda item: item['avg_duration'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(
            name='Average duration',
            x=[item['name'] for item in duration_fns],
            y=[item['avg_duration'] for item in duration_fns],
            marker_color='#2E5090'
        ))
        fig.add_trace(go.Bar(
            name='Maximum duration',
            x=[item['name'] for item in duration_fns],
            y=[item['max_duration'] for item in duration_fns],
            marker_color='#95a5a6'
        ))
        fig.update_layout(
            title='Performance: duration by function',
            barmode='group',
            height=320,
            margin=dict(t=45, b=90, l=50, r=20),
            template='plotly_white',
            yaxis_title='Milliseconds',
            xaxis_tickangle=-45
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')

    html += html_section('Time series')
    invocation_series = []
    error_series = []
    for item in function_data:
        dims = [{'Name': 'FunctionName', 'Value': item['name']}]
        invocation_series.append(get_metric_series('AWS/Lambda', 'Invocations', dims, 'Sum'))
        error_series.append(get_metric_series('AWS/Lambda', 'Errors', dims, 'Sum'))

    invocations_total = _sum_metric_series(invocation_series)
    errors_total = _sum_metric_series(error_series)
    inv_x, inv_y = _series_xy(invocations_total)
    err_x, err_y = _series_xy(errors_total)
    html += render_metric_line_chart(
        'Lambda traffic and errors over time',
        [
            {'name': 'Invocations', 'x': inv_x, 'y': inv_y, 'color': '#2E5090'},
            {'name': 'Errors', 'x': err_x, 'y': err_y, 'color': '#c0392b'},
        ],
        'Events'
    )

    duration_traces_source = sorted(duration_fns, key=lambda item: item['invocations'], reverse=True)
    duration_traces = []
    for item in duration_traces_source:
        dims = [{'Name': 'FunctionName', 'Value': item['name']}]
        duration = get_metric_series('AWS/Lambda', 'Duration', dims, 'Average')
        x, y = _series_xy(duration)
        duration_traces.append({'name': item['name'], 'x': x, 'y': y})
    html += render_metric_line_chart('Lambda average duration over time', duration_traces, 'Milliseconds')

    notes = []
    for item in function_data:
        if item['invocations'] == 0:
            notes.append((html_status('INFO', ''), item['name'], 'No invocations detected in the selected range.'))
        if item['errors'] > 0:
            notes.append((html_status('WARN', ''), item['name'], f'Errors detected ({int(item["errors"]):,}; error rate {item["error_rate"]}%).'))
        if item['throttles'] > 0:
            notes.append((html_status('WARN', ''), item['name'], f'Throttles detected ({int(item["throttles"]):,}).'))
        if item['dlq_errors'] > 0:
            notes.append((html_status('WARN', ''), item['name'], f'DLQ delivery errors detected ({int(item["dlq_errors"]):,}).'))
        if item['avg_duration'] >= 1000:
            notes.append((html_status('WARN', ''), item['name'], f'Average duration is above 1s ({item["avg_duration"]} ms).'))
        if item['max_duration'] >= 5000:
            notes.append((html_status('WARN', ''), item['name'], f'Max duration is above 5s ({item["max_duration"]} ms).'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Function', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No Lambda operational warnings detected with the current thresholds.')

    return html

def render_metrics_dynamodb():
    tables = clients['dynamodb'].list_tables()['TableNames']
    metrics = [
        ('ConsumedReadCapacityUnits', 'Sum', 'Reads (RCU)'),
        ('ConsumedWriteCapacityUnits', 'Sum', 'Writes (WCU)'),
        ('UserErrors', 'Sum', 'User errors'),
        ('SystemErrors', 'Sum', 'System errors'),
        ('ThrottledRequests', 'Sum', 'Throttled'),
    ]
    ops = ['GetItem', 'PutItem', 'DeleteItem', 'Query', 'Scan']
    rows = []
    table_data = []
    latency_matrix = []
    latency_rows = []

    for t in tables:
        dims = [{'Name': 'TableName', 'Value': t}]
        vals = [int(get_metric('AWS/DynamoDB', m, dims, s)) for m, s, _ in metrics]
        reads, writes, user_errors, system_errors, throttled = vals
        rows.append([t] + [f'{v:,}' for v in vals])

        op_latencies = []
        latency_display = []
        for op in ops:
            latency = get_metric(
                'AWS/DynamoDB',
                'SuccessfulRequestLatency',
                [{'Name': 'TableName', 'Value': t}, {'Name': 'Operation', 'Value': op}],
                'Average'
            )
            op_latencies.append(latency)
            latency_display.append(f'{latency}' if latency > 0 else '-')

        latency_matrix.append(op_latencies)
        latency_rows.append([t] + latency_display)
        table_data.append({
            'table': t,
            'reads': reads,
            'writes': writes,
            'user_errors': user_errors,
            'system_errors': system_errors,
            'throttled': throttled,
            'latencies': dict(zip(ops, op_latencies)),
        })

    html = html_title('DynamoDB - Metrics')
    html += html_metric_range()
    html += html_section('Capacity and errors')
    html += html_table(['Table'] + [n for _, _, n in metrics], rows)

    active_capacity = [t for t in table_data if t['reads'] > 0 or t['writes'] > 0]
    if active_capacity:
        active_capacity = sorted(active_capacity, key=lambda t: t['reads'] + t['writes'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(
            name='Reads (RCU)',
            x=[t['table'] for t in active_capacity],
            y=[t['reads'] for t in active_capacity],
            marker_color='#2E5090'
        ))
        fig.add_trace(go.Bar(
            name='Writes (WCU)',
            x=[t['table'] for t in active_capacity],
            y=[t['writes'] for t in active_capacity],
            marker_color='#2FA36B'
        ))
        fig.update_layout(
            title='Capacity profile: reads vs writes by table',
            barmode='group',
            height=330,
            margin=dict(t=45, b=80, l=50, r=20),
            template='plotly_white',
            yaxis_title='Consumed capacity units',
            xaxis_tickangle=-35
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No DynamoDB read/write capacity consumption detected in the selected range.')

    html += html_section('Latency by operation (ms)')
    html += html_table(['Table'] + ops, latency_rows)

    has_latency = any(any(value > 0 for value in row) for row in latency_matrix)
    if has_latency:
        heatmap_values = [[value if value > 0 else None for value in row] for row in latency_matrix]
        fig = go.Figure(data=go.Heatmap(
            z=heatmap_values,
            x=ops,
            y=tables,
            colorscale='YlOrRd',
            colorbar=dict(title='ms'),
            hovertemplate='<b>%{y}</b><br>Operation: %{x}<br>Latency: %{z} ms<extra></extra>'
        ))
        fig.update_layout(
            title='Operation latency heatmap',
            height=max(320, 140 + len(tables) * 34),
            margin=dict(t=45, b=45, l=140, r=30),
            template='plotly_white',
            xaxis_title='Operation',
            yaxis_title='Table'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No DynamoDB operation latency detected in the selected range.')

    notes = []
    for table in table_data:
        if table['throttled'] > 0:
            notes.append((html_status('WARN', ''), table['table'], f'Throttled requests detected ({table["throttled"]:,}).'))
        if table['system_errors'] > 0:
            notes.append((html_status('WARN', ''), table['table'], f'System errors detected ({table["system_errors"]:,}).'))
        if table['user_errors'] > 0:
            notes.append((html_status('INFO', ''), table['table'], f'User errors detected ({table["user_errors"]:,}).'))
        if table['reads'] == 0 and table['writes'] == 0:
            notes.append((html_status('INFO', ''), table['table'], 'No read/write capacity consumption detected.'))
        slow_ops = [f'{op}: {latency} ms' for op, latency in table['latencies'].items() if latency >= 50]
        if slow_ops:
            notes.append((html_status('WARN', ''), table['table'], 'High operation latency detected: ' + ', '.join(slow_ops)))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Table', 'Observation'], notes, widths=['120px', '240px', 'auto'])
    else:
        html += html_success('No DynamoDB operational warnings detected with the current thresholds.')

    return html


# --- API Gateway + SQS + SNS + S3 ---
def render_metrics_apigateway():
    apis = clients['apigateway'].get_rest_apis()['items']
    metrics = [
        ('Count', 'Sum', 'Requests'), ('Latency', 'Average', 'Latency (ms)'),
        ('IntegrationLatency', 'Average', 'Integration latency (ms)'),
        ('4XXError', 'Sum', '4XX errors'), ('5XXError', 'Sum', '5XX errors'),
    ]
    rows = []
    api_data = []

    for api in apis:
        dims = [{'Name': 'ApiName', 'Value': api['name']}]
        vals = [get_metric('AWS/ApiGateway', m, dims, s) for m, s, _ in metrics]
        requests, latency, integration_latency, errors_4xx, errors_5xx = vals
        error_rate = round(((errors_4xx + errors_5xx) / requests) * 100, 2) if requests > 0 else 0

        api_data.append({
            'name': api['name'],
            'requests': requests,
            'latency': latency,
            'integration_latency': integration_latency,
            'errors_4xx': errors_4xx,
            'errors_5xx': errors_5xx,
            'error_rate': error_rate,
        })

        rows.append([
            api['name'],
            f'{int(requests):,}',
            f'{latency}',
            f'{integration_latency}',
            f'{int(errors_4xx):,}',
            f'{int(errors_5xx):,}',
            f'{error_rate}%'
        ])

    html = html_title('API Gateway - Metrics')
    html += html_metric_range()
    html += html_table(['API'] + [n for _, _, n in metrics] + ['Error rate'], rows)

    active_apis = [api for api in api_data if api['requests'] > 0]
    if active_apis:
        active_apis = sorted(active_apis, key=lambda api: api['requests'], reverse=True)
        fig = go.Figure(data=[go.Bar(
            x=[api['name'] for api in active_apis],
            y=[api['requests'] for api in active_apis],
            marker_color='#2E5090'
        )])
        fig.update_layout(
            title='Traffic: requests per API',
            height=320,
            margin=dict(t=45, b=70, l=50, r=20),
            template='plotly_white',
            yaxis_title='Requests',
            xaxis_tickangle=-35
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No API Gateway requests detected in the selected range.')

    error_apis = [api for api in api_data if api['errors_4xx'] > 0 or api['errors_5xx'] > 0]
    if error_apis:
        error_apis = sorted(error_apis, key=lambda api: api['errors_4xx'] + api['errors_5xx'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(
            name='4XX client errors',
            x=[api['name'] for api in error_apis],
            y=[api['errors_4xx'] for api in error_apis],
            marker_color='#e67e22'
        ))
        fig.add_trace(go.Bar(
            name='5XX server errors',
            x=[api['name'] for api in error_apis],
            y=[api['errors_5xx'] for api in error_apis],
            marker_color='#c0392b'
        ))
        fig.update_layout(
            title='Reliability: 4XX vs 5XX errors',
            barmode='stack',
            height=320,
            margin=dict(t=45, b=70, l=50, r=20),
            template='plotly_white',
            yaxis_title='Errors',
            xaxis_tickangle=-35
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_success('No API Gateway 4XX or 5XX errors detected in the selected range.')

    latency_apis = [api for api in api_data if api['latency'] > 0 or api['integration_latency'] > 0]
    if latency_apis:
        names = [api['name'] for api in latency_apis]
        integrations = [api['integration_latency'] for api in latency_apis]
        overheads = [round(api['latency'] - api['integration_latency'], 1) if api['latency'] > 0 else 0 for api in latency_apis]

        fig = go.Figure()
        fig.add_trace(go.Bar(name='Integration latency', x=names, y=integrations, marker_color='#2E5090'))
        fig.add_trace(go.Bar(name='API Gateway overhead', x=names, y=overheads, marker_color='#95a5a6'))
        fig.update_layout(
            barmode='stack',
            title='Performance: latency breakdown',
            height=320,
            margin=dict(t=45, b=70, l=50, r=20),
            template='plotly_white',
            yaxis_title='Milliseconds',
            xaxis_tickangle=-35
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')

    html += html_section('Latency trend')
    if latency_apis:
        for api in latency_apis:
            dims = [{'Name': 'ApiName', 'Value': api['name']}]
            latency = get_metric_series('AWS/ApiGateway', 'Latency', dims, 'Average')
            integration = get_metric_series('AWS/ApiGateway', 'IntegrationLatency', dims, 'Average')
            latency_x, latency_y = _series_xy(latency)
            integration_x, integration_y = _series_xy(integration)
            html += render_metric_line_chart(
                f'{api["name"]}: latency over time',
                [
                    {'name': 'Total latency', 'x': latency_x, 'y': latency_y, 'color': '#2E5090'},
                    {'name': 'Integration latency', 'x': integration_x, 'y': integration_y, 'color': '#27ae60'},
                ],
                'Milliseconds'
            )
    else:
        html += html_empty('No API Gateway latency datapoints detected in the selected range.')

    notes = []
    for api in api_data:
        if api['requests'] == 0:
            notes.append((html_status('INFO', ''), api['name'], 'No requests detected in the selected range.'))
        if api['errors_5xx'] > 0:
            notes.append((html_status('WARN', ''), api['name'], f'5XX server errors detected ({int(api["errors_5xx"]):,}).'))
        if api['errors_4xx'] > 0:
            notes.append((html_status('INFO', ''), api['name'], f'4XX client errors detected ({int(api["errors_4xx"]):,}).'))
        if api['latency'] >= 1000:
            notes.append((html_status('WARN', ''), api['name'], f'Average latency is above 1s ({api["latency"]} ms).'))
        if api['integration_latency'] > 0 and api['latency'] > 0 and api['latency'] - api['integration_latency'] >= 250:
            notes.append((html_status('INFO', ''), api['name'], f'API Gateway overhead is elevated ({round(api["latency"] - api["integration_latency"], 1)} ms).'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'API', 'Observation'], notes, widths=['120px', '240px', 'auto'])
    else:
        html += html_success('No API Gateway operational warnings detected with the current thresholds.')

    return html

def render_metrics_sqs():
    urls = clients['sqs'].list_queues().get('QueueUrls', [])
    names = [u.split('/')[-1] for u in urls]
    metrics = [
        ('NumberOfMessagesSent', 'Sum', 'Sent'), ('NumberOfMessagesReceived', 'Sum', 'Received'),
        ('NumberOfMessagesDeleted', 'Sum', 'Deleted'), ('NumberOfEmptyReceives', 'Sum', 'Empty receives'),
        ('SentMessageSize', 'Average', 'Avg size (B)'),
        ('ApproximateNumberOfMessagesVisible', 'Maximum', 'Visible (max)'),
        ('ApproximateNumberOfMessagesNotVisible', 'Maximum', 'In flight (max)'),
        ('ApproximateNumberOfMessagesDelayed', 'Maximum', 'Delayed (max)'),
        ('ApproximateAgeOfOldestMessage', 'Maximum', 'Oldest msg (s)'),
    ]
    rows = []
    queue_data = []

    for q in names:
        dims = [{'Name': 'QueueName', 'Value': q}]
        vals = [get_metric('AWS/SQS', m, dims, s) for m, s, _ in metrics]
        sent, received, deleted, empty_receives, avg_size, visible, in_flight, delayed, oldest_age = vals
        queue_data.append({
            'name': q,
            'sent': sent,
            'received': received,
            'deleted': deleted,
            'empty_receives': empty_receives,
            'avg_size': avg_size,
            'visible': visible,
            'in_flight': in_flight,
            'delayed': delayed,
            'oldest_age': oldest_age,
        })
        rows.append([q] + [f'{int(v):,}' if s in ('Sum', 'Maximum') else f'{v}' for v, (m, s, _) in zip(vals, metrics)])

    html = html_title('SQS - Metrics')
    html += html_metric_range()
    html += html_table(['Queue'] + [n for _, _, n in metrics], rows)

    flow_queues = [q for q in queue_data if q['sent'] > 0 or q['received'] > 0 or q['deleted'] > 0]
    if flow_queues:
        flow_queues = sorted(flow_queues, key=lambda q: q['sent'] + q['received'] + q['deleted'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Sent', x=[q['name'] for q in flow_queues], y=[q['sent'] for q in flow_queues], marker_color='#2E5090'))
        fig.add_trace(go.Bar(name='Received', x=[q['name'] for q in flow_queues], y=[q['received'] for q in flow_queues], marker_color='#2FA36B'))
        fig.add_trace(go.Bar(name='Deleted', x=[q['name'] for q in flow_queues], y=[q['deleted'] for q in flow_queues], marker_color='#95a5a6'))
        fig.update_layout(
            title='Message flow: sent, received and deleted',
            barmode='group',
            height=330,
            margin=dict(t=45, b=90, l=50, r=20),
            template='plotly_white',
            yaxis_title='Messages',
            xaxis_tickangle=-45
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No SQS message flow detected in the selected range.')

    backlog_queues = [q for q in queue_data if q['visible'] > 0 or q['in_flight'] > 0 or q['delayed'] > 0]
    if backlog_queues:
        backlog_queues = sorted(backlog_queues, key=lambda q: q['visible'] + q['in_flight'] + q['delayed'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Visible', x=[q['name'] for q in backlog_queues], y=[q['visible'] for q in backlog_queues], marker_color='#e67e22'))
        fig.add_trace(go.Bar(name='In flight', x=[q['name'] for q in backlog_queues], y=[q['in_flight'] for q in backlog_queues], marker_color='#2E5090'))
        fig.add_trace(go.Bar(name='Delayed', x=[q['name'] for q in backlog_queues], y=[q['delayed'] for q in backlog_queues], marker_color='#8e44ad'))
        fig.update_layout(
            title='Backlog: visible, in-flight and delayed messages',
            barmode='stack',
            height=330,
            margin=dict(t=45, b=90, l=50, r=20),
            template='plotly_white',
            yaxis_title='Messages',
            xaxis_tickangle=-45
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_success('No SQS backlog detected in the selected range.')

    delayed_queues = [q for q in queue_data if q['oldest_age'] > 0]
    if delayed_queues:
        delayed_queues = sorted(delayed_queues, key=lambda q: q['oldest_age'], reverse=True)
        fig = go.Figure(data=[go.Bar(
            x=[q['name'] for q in delayed_queues],
            y=[q['oldest_age'] for q in delayed_queues],
            marker_color='#c0392b'
        )])
        fig.update_layout(
            title='Processing delay: oldest message age',
            height=320,
            margin=dict(t=45, b=90, l=50, r=20),
            template='plotly_white',
            yaxis_title='Seconds',
            xaxis_tickangle=-45
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')

    html += html_section('Queue trends')
    for q in queue_data:
        dims = [{'Name': 'QueueName', 'Value': q['name']}]
        visible_x, visible_y = _series_xy(get_metric_series('AWS/SQS', 'ApproximateNumberOfMessagesVisible', dims, 'Maximum'))
        in_flight_x, in_flight_y = _series_xy(get_metric_series('AWS/SQS', 'ApproximateNumberOfMessagesNotVisible', dims, 'Maximum'))
        delayed_x, delayed_y = _series_xy(get_metric_series('AWS/SQS', 'ApproximateNumberOfMessagesDelayed', dims, 'Maximum'))
        html += render_metric_line_chart(
            f'{q["name"]}: queue depth over time',
            [
                {'name': 'Visible', 'x': visible_x, 'y': visible_y, 'color': '#e67e22'},
                {'name': 'In flight', 'x': in_flight_x, 'y': in_flight_y, 'color': '#2E5090'},
                {'name': 'Delayed', 'x': delayed_x, 'y': delayed_y, 'color': '#8e44ad'},
            ],
            'Messages'
        )

        age_x, age_y = _series_xy(get_metric_series('AWS/SQS', 'ApproximateAgeOfOldestMessage', dims, 'Maximum'))
        html += render_metric_line_chart(
            f'{q["name"]}: oldest message age over time',
            [{'name': 'Oldest message age', 'x': age_x, 'y': age_y, 'color': '#c0392b'}],
            'Seconds'
        )

    notes = []
    for q in queue_data:
        if q['sent'] == 0 and q['received'] == 0 and q['deleted'] == 0:
            notes.append((html_status('INFO', ''), q['name'], 'No message flow detected in the selected range.'))
        if q['visible'] > 0:
            notes.append((html_status('WARN', ''), q['name'], f'Visible backlog detected ({int(q["visible"]):,} message(s)).'))
        if q['in_flight'] > 0:
            notes.append((html_status('INFO', ''), q['name'], f'Messages in flight detected ({int(q["in_flight"]):,}).'))
        if q['delayed'] > 0:
            notes.append((html_status('INFO', ''), q['name'], f'Delayed messages detected ({int(q["delayed"]):,}).'))
        if q['oldest_age'] >= 300:
            notes.append((html_status('WARN', ''), q['name'], f'Oldest message age is above 5 minutes ({int(q["oldest_age"]):,} s).'))
        if q['empty_receives'] > q['received'] and q['empty_receives'] > 0:
            notes.append((html_status('INFO', ''), q['name'], f'Empty receives are higher than received messages ({int(q["empty_receives"]):,}).'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Queue', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No SQS operational warnings detected with the current thresholds.')

    return html

def render_metrics_sns():
    topics = clients['sns'].list_topics()['Topics']
    metrics = [
        ('NumberOfMessagesPublished', 'Sum', 'Published'),
        ('NumberOfNotificationsDelivered', 'Sum', 'Delivered'),
        ('NumberOfNotificationsFailed', 'Sum', 'Failed'),
        ('PublishSize', 'Average', 'Avg size (B)'),
    ]
    rows = []
    topic_data = []

    for t in topics:
        name = t['TopicArn'].split(':')[-1]
        dims = [{'Name': 'TopicName', 'Value': name}]
        vals = [get_metric('AWS/SNS', m, dims, s) for m, s, _ in metrics]
        published, delivered, failed, avg_size = vals
        failure_rate = round((failed / published) * 100, 2) if published > 0 else 0
        topic_data.append({
            'name': name,
            'published': published,
            'delivered': delivered,
            'failed': failed,
            'avg_size': avg_size,
            'failure_rate': failure_rate,
        })
        rows.append([
            name,
            f'{int(published):,}',
            f'{int(delivered):,}',
            f'{int(failed):,}',
            f'{avg_size}',
            f'{failure_rate}%'
        ])

    html = html_title('SNS - Metrics')
    html += html_metric_range()
    html += html_table(['Topic'] + [n for _, _, n in metrics] + ['Failure rate'], rows)

    active_topics = [topic for topic in topic_data if topic['published'] > 0 or topic['delivered'] > 0 or topic['failed'] > 0]
    if active_topics:
        active_topics = sorted(active_topics, key=lambda topic: topic['published'] + topic['delivered'] + topic['failed'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(
            name='Published',
            y=[topic['name'] for topic in active_topics],
            x=[topic['published'] for topic in active_topics],
            orientation='h',
            marker_color='#2E5090'
        ))
        fig.add_trace(go.Bar(
            name='Delivered',
            y=[topic['name'] for topic in active_topics],
            x=[topic['delivered'] for topic in active_topics],
            orientation='h',
            marker_color='#2FA36B'
        ))
        fig.add_trace(go.Bar(
            name='Failed',
            y=[topic['name'] for topic in active_topics],
            x=[topic['failed'] for topic in active_topics],
            orientation='h',
            marker_color='#c0392b'
        ))
        fig.update_layout(
            title='Delivery outcome: published, delivered and failed',
            barmode='group',
            height=max(320, 140 + len(active_topics) * 45),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Notifications',
            yaxis_title='Topic'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No SNS publish or delivery activity detected in the selected range.')

    notes = []
    for topic in topic_data:
        if topic['failed'] > 0:
            notes.append((html_status('WARN', ''), topic['name'], f'Failed notifications detected ({int(topic["failed"]):,}; failure rate {topic["failure_rate"]}%).'))
        if topic['published'] > 0 and topic['delivered'] == 0 and topic['failed'] == 0:
            notes.append((html_status('INFO', ''), topic['name'], 'Messages were published, but no delivery outcome metrics were reported.'))
        if topic['published'] == 0 and topic['delivered'] == 0 and topic['failed'] == 0:
            notes.append((html_status('INFO', ''), topic['name'], 'No SNS activity detected in the selected range.'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Topic', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No SNS delivery warnings detected in the selected range.')

    return html

def render_metrics_s3():
    buckets = [b['Name'] for b in clients['s3'].list_buckets()['Buckets']]
    rows = []
    for b in buckets:
        size = get_metric('AWS/S3', 'BucketSizeBytes',
            [{'Name': 'BucketName', 'Value': b}, {'Name': 'StorageType', 'Value': 'StandardStorage'}], 'Average')
        objs = get_metric('AWS/S3', 'NumberOfObjects',
            [{'Name': 'BucketName', 'Value': b}, {'Name': 'StorageType', 'Value': 'AllStorageTypes'}], 'Average')
        size_mb = round(size / (1024*1024), 2) if size > 0 else 0
        rows.append((b, f'{size_mb} MB', f'{int(objs):,}'))
    html = html_title('S3 — Metrics')
    html += html_metric_range()
    html += html_note('S3 metrics are published once per day with a 24-48h delay. Use Last month range for best results.')
    html += html_table(['Bucket', 'Size', 'Objects'], rows)

    notes = []
    for bucket, size_text, objects_text in rows:
        try:
            objects = int(objects_text.replace(',', ''))
        except Exception:
            objects = 0
        if objects == 0:
            notes.append((html_status('INFO', ''), bucket, 'No objects reported by CloudWatch daily storage metrics.'))
        elif objects >= 1000:
            notes.append((html_status('INFO', ''), bucket, f'Large object count reported ({objects_text} objects).'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Bucket', 'Observation'], notes, widths=['120px', '280px', 'auto'])
    else:
        html += html_success('No S3 storage warnings detected from daily CloudWatch metrics.')
    return html


# --- RDS + Step Functions + WAF + EB + ElastiCache + ECR + Cognito ---
def render_metrics_rds():
    try:
        instances = clients['rds'].describe_db_instances()['DBInstances']
    except Exception:
        return html_title('RDS - Metrics') + html_empty('No RDS instances found.')

    metrics = [
        ('CPUUtilization', 'Average', 'CPU avg (%)'), ('DatabaseConnections', 'Maximum', 'Max connections'),
        ('FreeableMemory', 'Average', 'Free memory (MB)'), ('FreeStorageSpace', 'Average', 'Free storage (MB)'),
        ('ReadLatency', 'Average', 'Read latency (s)'), ('WriteLatency', 'Average', 'Write latency (s)'),
        ('ReadIOPS', 'Average', 'Read IOPS'), ('WriteIOPS', 'Average', 'Write IOPS'),
    ]
    rows = []
    instance_data = []

    for db in instances:
        instance_id = db['DBInstanceIdentifier']
        dims = [{'Name': 'DBInstanceIdentifier', 'Value': instance_id}]
        vals = [get_metric('AWS/RDS', m, dims, s) for m, s, _ in metrics]
        cpu, connections, free_memory, free_storage, read_latency, write_latency, read_iops, write_iops = vals
        free_memory_mb = round(free_memory / (1024*1024), 1) if free_memory > 0 else 0
        free_storage_mb = round(free_storage / (1024*1024), 1) if free_storage > 0 else 0

        instance_data.append({
            'id': instance_id,
            'cpu': cpu,
            'connections': connections,
            'free_memory_mb': free_memory_mb,
            'free_storage_mb': free_storage_mb,
            'read_latency': read_latency,
            'write_latency': write_latency,
            'read_iops': read_iops,
            'write_iops': write_iops,
        })
        rows.append([
            instance_id,
            f'{cpu}',
            f'{int(connections):,}',
            f'{free_memory_mb}',
            f'{free_storage_mb}',
            f'{read_latency}',
            f'{write_latency}',
            f'{read_iops}',
            f'{write_iops}',
        ])

    html = html_title('RDS - Metrics')
    html += html_metric_range()
    html += html_table(['Instance'] + [n for _, _, n in metrics], rows)

    latency_instances = [db for db in instance_data if db['read_latency'] > 0 or db['write_latency'] > 0]
    if latency_instances:
        latency_instances = sorted(latency_instances, key=lambda db: max(db['read_latency'], db['write_latency']), reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Scatter(
            name='Read latency',
            x=[db['read_latency'] for db in latency_instances],
            y=[db['id'] for db in latency_instances],
            mode='markers',
            marker=dict(size=12, color='#2E5090')
        ))
        fig.add_trace(go.Scatter(
            name='Write latency',
            x=[db['write_latency'] for db in latency_instances],
            y=[db['id'] for db in latency_instances],
            mode='markers',
            marker=dict(size=12, color='#c0392b', symbol='diamond')
        ))
        fig.update_layout(
            title='Performance: read vs write latency',
            height=max(280, 120 + len(latency_instances) * 45),
            margin=dict(t=45, b=45, l=140, r=30),
            template='plotly_white',
            xaxis_title='Seconds',
            yaxis_title='Instance'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No RDS latency data detected in the selected range.')

    html += html_section('Resource trends')
    for db in instance_data:
        dims = [{'Name': 'DBInstanceIdentifier', 'Value': db['id']}]
        cpu_x, cpu_y = _series_xy(get_metric_series('AWS/RDS', 'CPUUtilization', dims, 'Average'))
        conn_x, conn_y = _series_xy(get_metric_series('AWS/RDS', 'DatabaseConnections', dims, 'Maximum'))
        html += render_dual_axis_line_chart(
            f'{db["id"]}: CPU and connections over time',
            {'name': 'CPU utilization', 'x': cpu_x, 'y': cpu_y, 'color': '#2E5090'},
            {'name': 'Database connections', 'x': conn_x, 'y': conn_y, 'color': '#e67e22'},
            'CPU (%)',
            'Connections'
        )

        memory_series = [(ts, round(value / (1024 * 1024), 1)) for ts, value in get_metric_series('AWS/RDS', 'FreeableMemory', dims, 'Average')]
        storage_series = [(ts, round(value / (1024 * 1024), 1)) for ts, value in get_metric_series('AWS/RDS', 'FreeStorageSpace', dims, 'Average')]
        mem_x, mem_y = _series_xy(memory_series)
        storage_x, storage_y = _series_xy(storage_series)
        html += render_metric_line_chart(
            f'{db["id"]}: free memory and storage over time',
            [
                {'name': 'Free memory', 'x': mem_x, 'y': mem_y, 'color': '#27ae60'},
                {'name': 'Free storage', 'x': storage_x, 'y': storage_y, 'color': '#8e44ad'},
            ],
            'MB'
        )

    notes = []
    for db in instance_data:
        if db['cpu'] >= 70:
            notes.append((html_status('WARN', ''), db['id'], f'High average CPU utilization ({db["cpu"]}%).'))
        if db['connections'] >= 80:
            notes.append((html_status('WARN', ''), db['id'], f'High number of database connections ({int(db["connections"]):,}).'))
        if 0 < db['free_memory_mb'] < 256:
            notes.append((html_status('WARN', ''), db['id'], f'Low free memory ({db["free_memory_mb"]} MB).'))
        if 0 < db['free_storage_mb'] < 1024:
            notes.append((html_status('WARN', ''), db['id'], f'Low free storage ({db["free_storage_mb"]} MB).'))
        if db['read_latency'] >= 0.1:
            notes.append((html_status('WARN', ''), db['id'], f'Read latency is above 100 ms ({db["read_latency"]} s).'))
        if db['write_latency'] >= 0.1:
            notes.append((html_status('WARN', ''), db['id'], f'Write latency is above 100 ms ({db["write_latency"]} s).'))
        if db['write_latency'] > 0 and db['read_latency'] > 0 and db['write_latency'] > db['read_latency'] * 2:
            notes.append((html_status('INFO', ''), db['id'], 'Write latency is significantly higher than read latency.'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Instance', 'Observation'], notes, widths=['120px', '240px', 'auto'])
    else:
        html += html_success('No RDS operational warnings detected with the current thresholds.')

    return html

def render_metrics_stepfunctions():
    try:
        sms = clients['stepfunctions'].list_state_machines()['stateMachines']
    except Exception:
        return html_title('Step Functions - Metrics') + html_empty('No state machines found.')

    metrics = [
        ('ExecutionsStarted', 'Sum', 'Started'), ('ExecutionsSucceeded', 'Sum', 'Succeeded'),
        ('ExecutionsFailed', 'Sum', 'Failed'), ('ExecutionsTimedOut', 'Sum', 'Timed out'),
        ('ExecutionsAborted', 'Sum', 'Aborted'),
        ('ExecutionTime', 'Average', 'Avg time (ms)'), ('ExecutionTime', 'Maximum', 'Max time (ms)'),
    ]
    rows = []
    sm_data = []

    for sm in sms:
        dims = [{'Name': 'StateMachineArn', 'Value': sm['stateMachineArn']}]
        vals = [get_metric('AWS/States', m, dims, s) for m, s, _ in metrics]
        started, succeeded, failed, timed_out, aborted, avg_time, max_time = vals
        success_rate = round((succeeded / started) * 100, 2) if started > 0 else 0
        sm_data.append({
            'name': sm['name'],
            'started': started,
            'succeeded': succeeded,
            'failed': failed,
            'timed_out': timed_out,
            'aborted': aborted,
            'avg_time': avg_time,
            'max_time': max_time,
            'success_rate': success_rate,
        })
        rows.append([
            sm['name'],
            f'{int(started):,}',
            f'{int(succeeded):,}',
            f'{int(failed):,}',
            f'{int(timed_out):,}',
            f'{int(aborted):,}',
            f'{avg_time}',
            f'{max_time}',
            f'{success_rate}%'
        ])

    html = html_title('Step Functions - Metrics')
    html += html_metric_range()
    html += html_table(['State machine'] + [n for _, _, n in metrics] + ['Success rate'], rows)

    active_sms = [sm for sm in sm_data if sm['started'] > 0 or sm['succeeded'] > 0 or sm['failed'] > 0 or sm['timed_out'] > 0 or sm['aborted'] > 0]
    if active_sms:
        active_sms = sorted(active_sms, key=lambda sm: sm['started'] + sm['failed'] + sm['timed_out'] + sm['aborted'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Succeeded', y=[sm['name'] for sm in active_sms], x=[sm['succeeded'] for sm in active_sms], orientation='h', marker_color='#2FA36B'))
        fig.add_trace(go.Bar(name='Failed', y=[sm['name'] for sm in active_sms], x=[sm['failed'] for sm in active_sms], orientation='h', marker_color='#c0392b'))
        fig.add_trace(go.Bar(name='Timed out', y=[sm['name'] for sm in active_sms], x=[sm['timed_out'] for sm in active_sms], orientation='h', marker_color='#e67e22'))
        fig.add_trace(go.Bar(name='Aborted', y=[sm['name'] for sm in active_sms], x=[sm['aborted'] for sm in active_sms], orientation='h', marker_color='#7f8c8d'))
        fig.update_layout(
            title='Execution outcome by state machine',
            barmode='stack',
            height=max(320, 140 + len(active_sms) * 45),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Executions',
            yaxis_title='State machine'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No Step Functions executions detected in the selected range.')

    duration_sms = [sm for sm in sm_data if sm['avg_time'] > 0 or sm['max_time'] > 0]
    if duration_sms:
        duration_sms = sorted(duration_sms, key=lambda sm: sm['max_time'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Scatter(
            name='Average time',
            x=[sm['avg_time'] for sm in duration_sms],
            y=[sm['name'] for sm in duration_sms],
            mode='markers',
            marker=dict(size=12, color='#2E5090')
        ))
        fig.add_trace(go.Scatter(
            name='Maximum time',
            x=[sm['max_time'] for sm in duration_sms],
            y=[sm['name'] for sm in duration_sms],
            mode='markers',
            marker=dict(size=12, color='#c0392b', symbol='diamond')
        ))
        fig.update_layout(
            title='Execution duration: average vs maximum',
            height=max(300, 120 + len(duration_sms) * 45),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Milliseconds',
            yaxis_title='State machine'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')

    notes = []
    for sm in sm_data:
        if sm['started'] == 0:
            notes.append((html_status('INFO', ''), sm['name'], 'No executions detected in the selected range.'))
        if sm['failed'] > 0:
            notes.append((html_status('WARN', ''), sm['name'], f'Failed executions detected ({int(sm["failed"]):,}).'))
        if sm['timed_out'] > 0:
            notes.append((html_status('WARN', ''), sm['name'], f'Timed out executions detected ({int(sm["timed_out"]):,}).'))
        if sm['aborted'] > 0:
            notes.append((html_status('INFO', ''), sm['name'], f'Aborted executions detected ({int(sm["aborted"]):,}).'))
        if sm['avg_time'] >= 30000:
            notes.append((html_status('WARN', ''), sm['name'], f'Average execution time is above 30s ({sm["avg_time"]} ms).'))
        if sm['max_time'] >= 60000:
            notes.append((html_status('WARN', ''), sm['name'], f'Max execution time is above 60s ({sm["max_time"]} ms).'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'State machine', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No Step Functions operational warnings detected with the current thresholds.')

    return html

def render_metrics_waf():
    try:
        acls = clients['wafv2'].list_web_acls(Scope='REGIONAL')['WebACLs']
    except Exception:
        return html_title('WAF - Metrics') + html_empty('No Web ACLs found.')

    region = cloudwatch.meta.region_name or 'us-east-1'
    metrics = [('AllowedRequests', 'Sum', 'Allowed'), ('BlockedRequests', 'Sum', 'Blocked'), ('CountedRequests', 'Sum', 'Counted')]
    rows = []
    acl_data = []

    for acl in acls:
        try:
            detail = clients['wafv2'].get_web_acl(Name=acl['Name'], Scope='REGIONAL', Id=acl['Id'])['WebACL']
            metric_name = detail.get('VisibilityConfig', {}).get('MetricName', acl['Name'])
        except Exception:
            metric_name = acl['Name']

        dims = [{'Name': 'WebACL', 'Value': metric_name}, {'Name': 'Region', 'Value': region}, {'Name': 'Rule', 'Value': 'ALL'}]
        vals = [int(get_metric('AWS/WAFV2', m, dims, s)) for m, s, _ in metrics]
        allowed, blocked, counted = vals
        total = allowed + blocked + counted
        blocked_rate = round((blocked / total) * 100, 2) if total > 0 else 0
        acl_data.append({
            'name': acl['Name'],
            'metric_name': metric_name,
            'allowed': allowed,
            'blocked': blocked,
            'counted': counted,
            'total': total,
            'blocked_rate': blocked_rate,
        })
        rows.append([acl['Name']] + [f'{v:,}' for v in vals] + [f'{blocked_rate}%'])

    html = html_title('WAF - Metrics')
    html += html_metric_range()
    html += html_table(['Web ACL'] + [n for _, _, n in metrics] + ['Blocked rate'], rows)

    active_acls = [acl for acl in acl_data if acl['total'] > 0]
    if active_acls:
        active_acls = sorted(active_acls, key=lambda acl: acl['total'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Allowed', y=[acl['name'] for acl in active_acls], x=[acl['allowed'] for acl in active_acls], orientation='h', marker_color='#2FA36B'))
        fig.add_trace(go.Bar(name='Blocked', y=[acl['name'] for acl in active_acls], x=[acl['blocked'] for acl in active_acls], orientation='h', marker_color='#c0392b'))
        fig.add_trace(go.Bar(name='Counted', y=[acl['name'] for acl in active_acls], x=[acl['counted'] for acl in active_acls], orientation='h', marker_color='#2E5090'))
        fig.update_layout(
            title='Traffic decision split: allowed, blocked and counted',
            barmode='stack',
            height=max(300, 130 + len(active_acls) * 45),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Requests',
            yaxis_title='Web ACL'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No WAF traffic metrics detected in the selected range.')

    html += html_section('Traffic trend')
    if active_acls:
        for acl in active_acls:
            dims = [
                {'Name': 'WebACL', 'Value': acl['metric_name']},
                {'Name': 'Region', 'Value': region},
                {'Name': 'Rule', 'Value': 'ALL'},
            ]
            allowed_x, allowed_y = _series_xy(get_metric_series('AWS/WAFV2', 'AllowedRequests', dims, 'Sum'))
            blocked_x, blocked_y = _series_xy(get_metric_series('AWS/WAFV2', 'BlockedRequests', dims, 'Sum'))
            html += render_metric_line_chart(
                f'{acl["name"]}: allowed and blocked requests over time',
                [
                    {'name': 'Allowed', 'x': allowed_x, 'y': allowed_y, 'color': '#27ae60'},
                    {'name': 'Blocked', 'x': blocked_x, 'y': blocked_y, 'color': '#c0392b'},
                ],
                'Requests'
            )
    else:
        html += html_empty('No WAF traffic datapoints detected for trend charts in the selected range.')

    notes = []
    for acl in acl_data:
        if acl['total'] == 0:
            notes.append((html_status('INFO', ''), acl['name'], 'No WAF traffic detected for this Web ACL.'))
        if acl['blocked'] > 0:
            level = 'WARN' if acl['blocked_rate'] >= 20 else 'INFO'
            notes.append((html_status(level, ''), acl['name'], f'Blocked requests detected ({acl["blocked"]:,}; blocked rate {acl["blocked_rate"]}%).'))
        if acl['counted'] > 0:
            notes.append((html_status('INFO', ''), acl['name'], f'Counted requests detected ({acl["counted"]:,}); at least one rule may be in count mode.'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Web ACL', 'Observation'], notes, widths=['120px', '240px', 'auto'])
    else:
        html += html_success('No WAF operational warnings detected in the selected range.')

    return html

def render_metrics_elasticbeanstalk():
    try:
        entornos = clients['elasticbeanstalk'].describe_environments(IncludeDeleted=False)['Environments']
    except Exception:
        return html_title('Elastic Beanstalk - Metrics') + html_empty('No environments found.')

    metrics = [
        ('ApplicationRequestsTotal', 'Sum', 'Requests'),
        ('ApplicationLatencyP50', 'Average', 'Latency p50 (ms)'),
        ('ApplicationLatencyP90', 'Average', 'Latency p90 (ms)'),
        ('ApplicationLatencyP99', 'Average', 'Latency p99 (ms)'),
        ('ApplicationRequests4xx', 'Sum', '4XX errors'),
        ('ApplicationRequests5xx', 'Sum', '5XX errors'),
        ('EnvironmentHealth', 'Average', 'Health avg'),
    ]
    rows = []
    env_data = []

    for env in entornos:
        env_name = env['EnvironmentName']
        dims = [{'Name': 'EnvironmentName', 'Value': env_name}]
        vals = [get_metric('AWS/ElasticBeanstalk', m, dims, s) for m, s, _ in metrics]
        requests, p50, p90, p99, e4xx, e5xx, health_avg = vals
        error_rate = round(((e4xx + e5xx) / requests) * 100, 2) if requests > 0 else 0
        env_data.append({
            'name': env_name,
            'requests': requests,
            'p50': p50,
            'p90': p90,
            'p99': p99,
            'e4xx': e4xx,
            'e5xx': e5xx,
            'health_avg': health_avg,
            'error_rate': error_rate,
        })
        rows.append([
            env_name,
            f'{int(requests):,}',
            f'{p50}',
            f'{p90}',
            f'{p99}',
            f'{int(e4xx):,}',
            f'{int(e5xx):,}',
            f'{health_avg}',
            f'{error_rate}%'
        ])

    html = html_title('Elastic Beanstalk - Metrics')
    html += html_metric_range()
    html += html_note('Enhanced Health metrics require Enhanced Health Reporting to be enabled. Values may be 0 if not configured.')
    html += html_table(['Environment'] + [n for _, _, n in metrics] + ['Error rate'], rows)

    active_envs = [env for env in env_data if env['requests'] > 0 or env['e4xx'] > 0 or env['e5xx'] > 0]
    if active_envs:
        active_envs = sorted(active_envs, key=lambda env: env['requests'] + env['e4xx'] + env['e5xx'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Requests', x=[env['name'] for env in active_envs], y=[env['requests'] for env in active_envs], marker_color='#2E5090'))
        fig.add_trace(go.Bar(name='4XX errors', x=[env['name'] for env in active_envs], y=[env['e4xx'] for env in active_envs], marker_color='#e67e22'))
        fig.add_trace(go.Bar(name='5XX errors', x=[env['name'] for env in active_envs], y=[env['e5xx'] for env in active_envs], marker_color='#c0392b'))
        fig.update_layout(
            title='Application traffic and errors',
            barmode='group',
            height=330,
            margin=dict(t=45, b=80, l=50, r=20),
            template='plotly_white',
            yaxis_title='Count',
            xaxis_tickangle=-35
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No Elastic Beanstalk request/error metrics detected in the selected range.')

    latency_envs = [env for env in env_data if env['p50'] > 0 or env['p90'] > 0 or env['p99'] > 0]
    if latency_envs:
        latency_envs = sorted(latency_envs, key=lambda env: env['p99'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Scatter(name='p50', x=[env['p50'] for env in latency_envs], y=[env['name'] for env in latency_envs], mode='markers', marker=dict(size=11, color='#2FA36B')))
        fig.add_trace(go.Scatter(name='p90', x=[env['p90'] for env in latency_envs], y=[env['name'] for env in latency_envs], mode='markers', marker=dict(size=11, color='#e67e22', symbol='diamond')))
        fig.add_trace(go.Scatter(name='p99', x=[env['p99'] for env in latency_envs], y=[env['name'] for env in latency_envs], mode='markers', marker=dict(size=12, color='#c0392b', symbol='x')))
        fig.update_layout(
            title='Latency percentiles: p50 vs p90 vs p99',
            height=max(300, 120 + len(latency_envs) * 45),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Milliseconds',
            yaxis_title='Environment'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')

    notes = []
    for env in env_data:
        if env['requests'] == 0 and env['p50'] == 0 and env['p90'] == 0 and env['p99'] == 0:
            notes.append((html_status('INFO', ''), env['name'], 'No request or latency metrics detected; Enhanced Health may be disabled or there may be no traffic.'))
        if env['e5xx'] > 0:
            notes.append((html_status('WARN', ''), env['name'], f'5XX application errors detected ({int(env["e5xx"]):,}).'))
        if env['e4xx'] > 0:
            notes.append((html_status('INFO', ''), env['name'], f'4XX client errors detected ({int(env["e4xx"]):,}).'))
        if env['p99'] >= 1000:
            notes.append((html_status('WARN', ''), env['name'], f'p99 latency is above 1s ({env["p99"]} ms).'))
        if env['p50'] > 0 and env['p99'] > env['p50'] * 4:
            notes.append((html_status('INFO', ''), env['name'], 'p99 latency is much higher than p50; latency tail may be significant.'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Environment', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No Elastic Beanstalk operational warnings detected with the current thresholds.')

    return html

def render_metrics_elasticache():
    try:
        clusters = clients['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True)['CacheClusters']
    except Exception:
        return html_title('ElastiCache - Metrics') + html_empty('No clusters found.')

    metrics = [
        ('CPUUtilization', 'Average', 'CPU (%)'), ('FreeableMemory', 'Average', 'Free mem (MB)'),
        ('CurrConnections', 'Maximum', 'Connections'), ('Evictions', 'Sum', 'Evictions'),
        ('GetHits', 'Sum', 'Hits'), ('GetMisses', 'Sum', 'Misses'),
        ('BytesUsedForCacheItems', 'Average', 'Cache bytes used'),
        ('NetworkBytesIn', 'Sum', 'Bytes in'), ('NetworkBytesOut', 'Sum', 'Bytes out'),
    ]
    rows = []
    node_data = []

    for cl in clusters:
        nodes = cl.get('CacheNodes', []) or [{'CacheNodeId': '0001'}]
        for n in nodes:
            node_id = f"{cl['CacheClusterId']}/{n.get('CacheNodeId', '0001')}"
            dims = [
                {'Name': 'CacheClusterId', 'Value': cl['CacheClusterId']},
                {'Name': 'CacheNodeId', 'Value': n.get('CacheNodeId', '0001')}
            ]
            vals = [get_metric('AWS/ElastiCache', m, dims, s) for m, s, _ in metrics]
            cpu, free_mem, connections, evictions, hits, misses, used_bytes, bytes_in, bytes_out = vals
            free_mem_mb = round(free_mem / (1024*1024), 1) if free_mem > 0 else 0
            used_mb = round(used_bytes / (1024*1024), 1) if used_bytes > 0 else 0
            lookups = hits + misses
            hit_rate = round((hits / lookups) * 100, 2) if lookups > 0 else None
            eviction_rate = round((evictions / lookups) * 1000, 2) if lookups > 0 else 0

            node_data.append({
                'node': node_id,
                'cpu': cpu,
                'free_mem_mb': free_mem_mb,
                'connections': connections,
                'evictions': evictions,
                'hits': hits,
                'misses': misses,
                'used_mb': used_mb,
                'bytes_in': bytes_in,
                'bytes_out': bytes_out,
                'lookups': lookups,
                'hit_rate': hit_rate,
                'eviction_rate': eviction_rate,
            })
            rows.append([
                node_id,
                f'{cpu}',
                f'{free_mem_mb}',
                f'{int(connections):,}',
                f'{int(evictions):,}',
                f'{int(hits):,}',
                f'{int(misses):,}',
                f'{used_bytes}',
                f'{int(bytes_in):,}',
                f'{int(bytes_out):,}',
            ])

    html = html_title('ElastiCache - Metrics')
    html += html_metric_range()
    html += html_table(['Node'] + [n for _, _, n in metrics], rows)

    active_cache_nodes = [node for node in node_data if node['lookups'] > 0]
    if active_cache_nodes:
        active_cache_nodes = sorted(active_cache_nodes, key=lambda node: node['lookups'], reverse=True)
        fig = go.Figure(data=[go.Scatter(
            x=[node['lookups'] for node in active_cache_nodes],
            y=[node['hit_rate'] for node in active_cache_nodes],
            mode='markers+text',
            text=[node['node'] for node in active_cache_nodes],
            textposition='top center',
            marker=dict(
                size=[max(10, min(42, node['evictions'] + 10)) for node in active_cache_nodes],
                color=[node['evictions'] for node in active_cache_nodes],
                colorscale='Reds',
                showscale=True,
                colorbar=dict(title='Evictions')
            ),
            hovertemplate='<b>%{text}</b><br>Lookups: %{x}<br>Hit rate: %{y}%<extra></extra>'
        )])
        fig.update_layout(
            title='Cache effectiveness: hit rate vs lookup volume',
            height=340,
            margin=dict(t=45, b=55, l=60, r=40),
            template='plotly_white',
            xaxis_title='Cache lookups (hits + misses)',
            yaxis_title='Hit rate (%)',
            yaxis=dict(range=[0, 105])
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No cache lookup activity detected in the selected range. Hit rate cannot be calculated.')

    memory_nodes = [node for node in node_data if node['used_mb'] > 0]
    if memory_nodes:
        memory_nodes = sorted(memory_nodes, key=lambda node: node['used_mb'] + node['free_mem_mb'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(
            name='Cache items used',
            y=[node['node'] for node in memory_nodes],
            x=[node['used_mb'] for node in memory_nodes],
            orientation='h',
            marker_color='#2E5090'
        ))
        fig.add_trace(go.Bar(
            name='Freeable memory',
            y=[node['node'] for node in memory_nodes],
            x=[node['free_mem_mb'] for node in memory_nodes],
            orientation='h',
            marker_color='#2FA36B'
        ))
        fig.update_layout(
            title='Memory composition: cache items vs freeable memory',
            barmode='stack',
            height=max(280, 120 + len(memory_nodes) * 45),
            margin=dict(t=45, b=45, l=160, r=30),
            template='plotly_white',
            xaxis_title='MB',
            yaxis_title='Node'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No cache item memory reported in the selected range; memory composition chart is skipped.')

    html += html_section('Cache trends')
    for node in node_data:
        cluster_id, node_id = node['node'].split('/', 1)
        dims = [
            {'Name': 'CacheClusterId', 'Value': cluster_id},
            {'Name': 'CacheNodeId', 'Value': node_id},
        ]
        conn_x, conn_y = _series_xy(get_metric_series('AWS/ElastiCache', 'CurrConnections', dims, 'Maximum'))
        evict_x, evict_y = _series_xy(get_metric_series('AWS/ElastiCache', 'Evictions', dims, 'Sum'))
        html += render_dual_axis_line_chart(
            f'{node["node"]}: connections and evictions over time',
            {'name': 'Connections', 'x': conn_x, 'y': conn_y, 'color': '#2E5090'},
            {'name': 'Evictions', 'x': evict_x, 'y': evict_y, 'color': '#c0392b'},
            'Connections',
            'Evictions'
        )

        hits_x, hits_y = _series_xy(get_metric_series('AWS/ElastiCache', 'GetHits', dims, 'Sum'))
        misses_x, misses_y = _series_xy(get_metric_series('AWS/ElastiCache', 'GetMisses', dims, 'Sum'))
        html += render_metric_line_chart(
            f'{node["node"]}: hits and misses over time',
            [
                {'name': 'Hits', 'x': hits_x, 'y': hits_y, 'color': '#27ae60'},
                {'name': 'Misses', 'x': misses_x, 'y': misses_y, 'color': '#e67e22'},
            ],
            'Lookups'
        )

    notes = []
    for node in node_data:
        if node['lookups'] == 0:
            notes.append((html_status('INFO', ''), node['node'], 'No hit/miss activity detected; cache effectiveness cannot be evaluated.'))
        elif node['hit_rate'] is not None and node['hit_rate'] < 70:
            notes.append((html_status('WARN', ''), node['node'], f'Low cache hit rate ({node["hit_rate"]}%).'))
        if node['evictions'] > 0:
            notes.append((html_status('WARN', ''), node['node'], f'Evictions detected ({int(node["evictions"]):,}); cache may be under memory pressure.'))
        if 0 < node['free_mem_mb'] < 64:
            notes.append((html_status('WARN', ''), node['node'], f'Low free memory ({node["free_mem_mb"]} MB).'))
        if node['cpu'] >= 70:
            notes.append((html_status('WARN', ''), node['node'], f'High CPU utilization ({node["cpu"]}%).'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Node', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No ElastiCache operational warnings detected with the current thresholds.')

    return html

def render_metrics_ecr():
    try:
        repositorios = clients['ecr'].describe_repositories()['repositories']
    except Exception:
        return html_title('ECR - Operational Metrics') + html_empty('No repositories found.')

    rows = []
    repo_data = []
    for repo in repositorios:
        name = repo['repositoryName']
        try:
            imagenes = clients['ecr'].describe_images(repositoryName=name).get('imageDetails', [])
        except Exception:
            imagenes = []

        total = len(imagenes)
        size_mb = round(sum(img.get('imageSizeInBytes', 0) for img in imagenes) / (1024*1024), 2)
        ultima = max((img.get('imagePushedAt') for img in imagenes if img.get('imagePushedAt')), default=None)
        if ultima:
            try:
                dias = ((datetime.now(ultima.tzinfo) if ultima.tzinfo else datetime.now()) - ultima).days
            except Exception:
                dias = 'N/A'
        else:
            dias = 'N/A'

        findings = {}
        imgs_with_date = [i for i in imagenes if i.get('imagePushedAt')]
        if imgs_with_date:
            latest = max(imgs_with_date, key=lambda i: i['imagePushedAt'])
            findings = latest.get('imageScanFindingsSummary', {}).get('findingSeverityCounts', {})

        critical = int(findings.get('CRITICAL', 0))
        high = int(findings.get('HIGH', 0))
        medium = int(findings.get('MEDIUM', 0))
        pulls = int(get_metric('AWS/ECR', 'RepositoryPullCount',
            [{'Name': 'RepositoryName', 'Value': name}], 'Sum'))
        repo_data.append({
            'name': name,
            'images': total,
            'size_mb': size_mb,
            'days_since_last_push': dias,
            'pulls': pulls,
            'critical': critical,
            'high': high,
            'medium': medium,
        })
        rows.append((name, str(total), f'{size_mb} MB', str(dias), f'{pulls:,}', str(critical), str(high), str(medium)))

    html = html_title('ECR - Operational Metrics')
    html += html_metric_range(include_period=True)
    html += html_note('ECR exposes limited CloudWatch metrics. Pull count comes from AWS/ECR; image inventory and vulnerability indicators come from the ECR API.')
    html += html_table(['Repository', 'Images', 'Total size', 'Days since last push', 'Pulls',
                        'Critical vulns', 'High vulns', 'Medium vulns'], rows)

    pulled_repos = [repo for repo in repo_data if repo['pulls'] > 0]
    if pulled_repos:
        pulled_repos = sorted(pulled_repos, key=lambda repo: repo['pulls'], reverse=True)
        fig = go.Figure(data=[go.Bar(
            y=[repo['name'] for repo in pulled_repos],
            x=[repo['pulls'] for repo in pulled_repos],
            orientation='h',
            marker_color='#2F5597'
        )])
        fig.update_layout(
            title='Repository consumption: image pulls',
            height=max(300, 120 + len(pulled_repos) * 42),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Pulls',
            yaxis_title='Repository'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No ECR image pulls detected in the selected range.')

    freshness_repos = [repo for repo in repo_data if isinstance(repo['days_since_last_push'], int)]
    if freshness_repos:
        freshness_repos = sorted(freshness_repos, key=lambda repo: repo['days_since_last_push'], reverse=True)
        fig = go.Figure(data=[go.Bar(
            y=[repo['name'] for repo in freshness_repos],
            x=[repo['days_since_last_push'] for repo in freshness_repos],
            orientation='h',
            marker_color=['#c0392b' if repo['days_since_last_push'] > 60 else '#e67e22' if repo['days_since_last_push'] > 30 else '#2FA36B' for repo in freshness_repos]
        )])
        fig.update_layout(
            title='Repository freshness: days since last push',
            height=max(300, 120 + len(freshness_repos) * 42),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Days',
            yaxis_title='Repository'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')

    vulnerable_repos = [repo for repo in repo_data if repo['critical'] > 0 or repo['high'] > 0 or repo['medium'] > 0]
    if vulnerable_repos:
        vulnerable_repos = sorted(vulnerable_repos, key=lambda repo: repo['critical'] * 100 + repo['high'] * 10 + repo['medium'], reverse=True)
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Critical', y=[repo['name'] for repo in vulnerable_repos], x=[repo['critical'] for repo in vulnerable_repos], orientation='h', marker_color='#8e0000'))
        fig.add_trace(go.Bar(name='High', y=[repo['name'] for repo in vulnerable_repos], x=[repo['high'] for repo in vulnerable_repos], orientation='h', marker_color='#c0392b'))
        fig.add_trace(go.Bar(name='Medium', y=[repo['name'] for repo in vulnerable_repos], x=[repo['medium'] for repo in vulnerable_repos], orientation='h', marker_color='#e67e22'))
        fig.update_layout(
            title='Image risk profile: vulnerability findings by repository',
            barmode='stack',
            height=max(300, 120 + len(vulnerable_repos) * 42),
            margin=dict(t=45, b=45, l=180, r=30),
            template='plotly_white',
            xaxis_title='Findings',
            yaxis_title='Repository'
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_success('No critical, high or medium ECR scan findings detected on latest images.')

    notes = []
    for repo in repo_data:
        if repo['images'] == 0:
            notes.append((html_status('INFO', ''), repo['name'], 'Repository has no images.'))
        elif repo['pulls'] == 0:
            notes.append((html_status('INFO', ''), repo['name'], 'Repository has images but no pulls were detected in the selected range.'))
        if repo['critical'] > 0:
            notes.append((html_status('CRIT', ''), repo['name'], f'Critical vulnerability findings detected ({repo["critical"]}).'))
        if repo['high'] > 0:
            notes.append((html_status('WARN', ''), repo['name'], f'High vulnerability findings detected ({repo["high"]}).'))
        if isinstance(repo['days_since_last_push'], int) and repo['days_since_last_push'] > 60:
            notes.append((html_status('WARN', ''), repo['name'], f'Last image push was more than 60 days ago ({repo["days_since_last_push"]} days).'))
        elif isinstance(repo['days_since_last_push'], int) and repo['days_since_last_push'] > 30:
            notes.append((html_status('INFO', ''), repo['name'], f'Last image push was more than 30 days ago ({repo["days_since_last_push"]} days).'))

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Repository', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No ECR operational warnings detected with the current thresholds.')

    return html


def render_metrics_cognito():
    html = html_title('Cognito - Metrics')
    html += html_metric_range(include_period=True)
    html += html_note('Cognito metrics are emitted when there is authentication activity in a user pool and app client during the selected range.')

    try:
        pools = clients['cognito'].list_user_pools(MaxResults=60).get('UserPools', [])
    except Exception as e:
        return html + html_empty(f'Unable to list Cognito user pools: {e}')

    if not pools:
        return html + html_empty('No Cognito user pools found.')

    metric_defs = [
        ('SignUpSuccesses', 'Sign-up successes'),
        ('SignUpThrottles', 'Sign-up throttles'),
        ('SignInSuccesses', 'Sign-in successes'),
        ('SignInThrottles', 'Sign-in throttles'),
        ('TokenRefreshSuccesses', 'Token refresh successes'),
    ]

    rows = []
    chart_rows = []
    notes = []

    for pool in pools:
        pool_id = pool['Id']
        pool_name = pool.get('Name', pool_id)
        try:
            app_clients = clients['cognito'].list_user_pool_clients(UserPoolId=pool_id, MaxResults=60).get('UserPoolClients', [])
        except Exception as e:
            rows.append((pool_name, '-', '-', '-', '-', '-', f'Could not list app clients: {e}'))
            notes.append((html_status('INFO', ''), pool_name, 'Could not inspect app clients for this user pool.'))
            continue

        if not app_clients:
            rows.append((pool_name, '-', '0', '0', '0', '0', '0'))
            notes.append((html_status('INFO', ''), pool_name, 'No app clients found for this user pool.'))
            continue

        for app in app_clients:
            client_id = app['ClientId']
            client_name = app.get('ClientName', client_id)
            dims = [
                {'Name': 'UserPool', 'Value': pool_id},
                {'Name': 'UserPoolClient', 'Value': client_id}
            ]
            values = {metric: int(get_metric('AWS/Cognito', metric, dims, 'Sum')) for metric, _ in metric_defs}
            rows.append((
                pool_name,
                client_name,
                f"{values['SignUpSuccesses']:,}",
                f"{values['SignUpThrottles']:,}",
                f"{values['SignInSuccesses']:,}",
                f"{values['SignInThrottles']:,}",
                f"{values['TokenRefreshSuccesses']:,}",
            ))
            chart_rows.append({
                'label': f'{pool_name}/{client_name}',
                'signups': values['SignUpSuccesses'],
                'signin': values['SignInSuccesses'],
                'refresh': values['TokenRefreshSuccesses'],
                'signup_throttles': values['SignUpThrottles'],
                'signin_throttles': values['SignInThrottles'],
            })

            if values['SignUpThrottles'] > 0 or values['SignInThrottles'] > 0:
                notes.append((html_status('WARN', ''), f'{pool_name}/{client_name}', f"Authentication throttles detected: {values['SignUpThrottles']} sign-up, {values['SignInThrottles']} sign-in."))
            if values['SignUpSuccesses'] == 0 and values['SignInSuccesses'] == 0 and values['TokenRefreshSuccesses'] == 0:
                notes.append((html_status('INFO', ''), f'{pool_name}/{client_name}', 'No authentication activity detected in the selected range.'))

    html += html_table([
        'User pool', 'App client', 'Sign-up successes', 'Sign-up throttles',
        'Sign-in successes', 'Sign-in throttles', 'Token refresh successes'
    ], rows, widths=['180px', '180px', '120px', '120px', '120px', '120px', '140px'])

    active_rows = [row for row in chart_rows if row['signups'] > 0 or row['signin'] > 0 or row['refresh'] > 0]
    if active_rows:
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Sign-ins', x=[row['label'] for row in active_rows], y=[row['signin'] for row in active_rows], marker_color='#2F5597'))
        fig.add_trace(go.Bar(name='Sign-ups', x=[row['label'] for row in active_rows], y=[row['signups'] for row in active_rows], marker_color='#2FA36B'))
        fig.add_trace(go.Bar(name='Token refreshes', x=[row['label'] for row in active_rows], y=[row['refresh'] for row in active_rows], marker_color='#7F8C8D'))
        fig.update_layout(
            title='Authentication activity by app client',
            barmode='group',
            height=max(320, 180 + len(active_rows) * 28),
            margin=dict(t=45, b=95, l=55, r=25),
            template='plotly_white',
            xaxis_title='User pool / app client',
            yaxis_title='Events',
            xaxis_tickangle=-25
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')
    else:
        html += html_empty('No Cognito authentication activity detected in the selected range.')

    throttled_rows = [row for row in chart_rows if row['signup_throttles'] > 0 or row['signin_throttles'] > 0]
    if throttled_rows:
        fig = go.Figure()
        fig.add_trace(go.Bar(name='Sign-up throttles', x=[row['label'] for row in throttled_rows], y=[row['signup_throttles'] for row in throttled_rows], marker_color='#e67e22'))
        fig.add_trace(go.Bar(name='Sign-in throttles', x=[row['label'] for row in throttled_rows], y=[row['signin_throttles'] for row in throttled_rows], marker_color='#c0392b'))
        fig.update_layout(
            title='Authentication throttles by app client',
            barmode='group',
            height=max(300, 180 + len(throttled_rows) * 28),
            margin=dict(t=45, b=95, l=55, r=25),
            template='plotly_white',
            xaxis_title='User pool / app client',
            yaxis_title='Throttled requests',
            xaxis_tickangle=-25
        )
        html += fig.to_html(full_html=False, include_plotlyjs='cdn')

    html += html_section('Operational notes')
    if notes:
        html += html_table(['Level', 'Resource', 'Observation'], notes, widths=['120px', '260px', 'auto'])
    else:
        html += html_success('No Cognito operational warnings detected with the current thresholds.')

    return html

def render_metrics_no_data(service):
    return html_title(f'{service} — Metrics') + html_empty('This service does not expose CloudWatch metrics. See Discovery, Logs, or Security tabs.')

METRICS_RENDERERS = {
    'Lambda': render_metrics_lambda, 'DynamoDB': render_metrics_dynamodb,
    'API Gateway': render_metrics_apigateway, 'SQS': render_metrics_sqs,
    'SNS': render_metrics_sns, 'S3': render_metrics_s3,
    'RDS': render_metrics_rds, 'Step Functions': render_metrics_stepfunctions,
    'WAF': render_metrics_waf, 'Elastic Beanstalk': render_metrics_elasticbeanstalk,
    'ElastiCache': render_metrics_elasticache, 'ECR': render_metrics_ecr,
    'Cognito': render_metrics_cognito,
    'CodeCommit': lambda: render_metrics_no_data('CodeCommit'),
}
```

```python
# === Security collectors ===

SECURITY_SERVICES = [svc for svc in SERVICES if svc != 'Overview']
SECURITY_ORDER = {'CRIT': 0, 'WARN': 1, 'INFO': 2, 'OK': 3}


def _error_code(exc):
    return getattr(exc, 'response', {}).get('Error', {}).get('Code', '')


def _is_missing(exc):
    return _error_code(exc) in {
        'NoSuchBucketPolicy',
        'NoSuchBucket',
        'NoSuchLifecyclePolicyException',
        'LifecyclePolicyNotFoundException',
        'PolicyNotFoundException',
        'ResourceNotFoundException',
        'WAFNonexistentItemException',
    }


def add_security_check(checks, service, resource, status, check, message):
    checks.append({
        'service': service,
        'resource': str(resource or 'N/A'),
        'status': status,
        'check': check,
        'message': message,
    })


def as_list(value):
    if value is None:
        return []
    return value if isinstance(value, list) else [value]


def public_principal(principal):
    if principal == '*':
        return True
    if isinstance(principal, dict):
        for value in principal.values():
            if value == '*' or (isinstance(value, list) and '*' in value):
                return True
    return False


def public_policy_doc(doc):
    for stmt in as_list(doc.get('Statement', [])):
        if stmt.get('Effect') == 'Allow' and public_principal(stmt.get('Principal', {})) and not stmt.get('Condition'):
            return True
    return False


def broad_action(actions):
    return any(action == '*' or str(action).endswith(':*') for action in as_list(actions))


def broad_resource(resources):
    return any(resource == '*' for resource in as_list(resources))


def summarize_actions(actions, limit=3):
    action_list = as_list(actions)
    labels = [str(action).split(':')[-1] for action in action_list[:limit]]
    text = ', '.join(labels) if labels else 'No actions reported'
    if len(action_list) > limit:
        text += f' (+{len(action_list) - limit} more)'
    return text



def add_iam_role_security_checks(checks, service, resource, role_arn):
    if not role_arn:
        add_security_check(checks, service, resource, 'INFO', 'IAM role', 'No IAM role detected')
        return

    role_name = role_arn.split('/')[-1]
    add_security_check(checks, service, resource, 'INFO', 'IAM role', role_name)

    try:
        attached = clients['iam'].list_attached_role_policies(RoleName=role_name).get('AttachedPolicies', [])
        if not attached:
            add_security_check(checks, service, resource, 'INFO', 'Managed policies', 'No managed policies attached')
        else:
            managed_rows = []
            broad_count = 0
            for policy in attached:
                policy_name = policy.get('PolicyName', 'N/A')
                if 'FullAccess' in policy_name or 'Administrator' in policy_name:
                    broad_count += 1
                    managed_rows.append(f'[WARN] Broad managed policy: {policy_name}')
                else:
                    managed_rows.append(f'[OK] {policy_name}')
            status = 'WARN' if broad_count else 'OK'
            summary = f'{len(attached)} attached managed policy(ies)'
            if broad_count:
                summary += f'; {broad_count} broad policy(ies) detected'
            add_security_check(checks, service, resource, status, 'Managed policies', details_list(summary, managed_rows))
    except Exception as exc:
        add_security_check(checks, service, resource, 'INFO', 'Managed policies', f'Could not inspect managed policies: {_error_code(exc) or exc}')

    try:
        inline_names = clients['iam'].list_role_policies(RoleName=role_name).get('PolicyNames', [])
        if not inline_names:
            return

        inline_rows = []
        worst_status = 'OK'
        for policy_name in inline_names:
            try:
                doc = clients['iam'].get_role_policy(RoleName=role_name, PolicyName=policy_name).get('PolicyDocument', {})
                statements = as_list(doc.get('Statement', []))
                if not statements:
                    inline_rows.append(f'[INFO] {policy_name}: no statements')
                    if worst_status == 'OK':
                        worst_status = 'INFO'
                for stmt in statements:
                    if stmt.get('Effect') == 'Allow' and broad_action(stmt.get('Action')) and broad_resource(stmt.get('Resource')):
                        inline_rows.append(f'[CRIT] {policy_name}: broad actions over broad resources')
                        worst_status = 'CRIT'
                    elif stmt.get('Effect') == 'Allow' and (broad_action(stmt.get('Action')) or broad_resource(stmt.get('Resource'))):
                        inline_rows.append(f'[WARN] {policy_name}: broad permissions')
                        if worst_status not in ('CRIT',):
                            worst_status = 'WARN'
                    else:
                        inline_rows.append(f'[OK] {policy_name}: {summarize_actions(stmt.get("Action", []))}')
            except Exception as exc:
                inline_rows.append(f'[INFO] {policy_name}: could not inspect content ({_error_code(exc) or exc})')
                if worst_status == 'OK':
                    worst_status = 'INFO'

        add_security_check(checks, service, resource, worst_status, 'Inline policies', details_list(f'{len(inline_names)} inline policy(ies)', inline_rows))
    except Exception:
        pass

def check_security_groups(checks, service, resource, security_group_ids):
    sg_ids = [sg for sg in security_group_ids if sg]
    if not sg_ids:
        add_security_check(checks, service, resource, 'INFO', 'Security groups', 'No security groups detected')
        return

    try:
        response = clients['ec2'].describe_security_groups(GroupIds=sg_ids)
    except Exception as exc:
        add_security_check(checks, service, resource, 'INFO', 'Security groups', f'Could not inspect security groups: {_error_code(exc) or exc}')
        return

    public_rules = []
    for sg in response.get('SecurityGroups', []):
        sg_id = sg.get('GroupId', 'N/A')
        for rule in sg.get('IpPermissions', []):
            from_port = rule.get('FromPort', 'all')
            to_port = rule.get('ToPort', 'all')
            port_text = f'{from_port}-{to_port}'
            public_ipv4 = any(ip.get('CidrIp') == '0.0.0.0/0' for ip in rule.get('IpRanges', []))
            public_ipv6 = any(ip.get('CidrIpv6') == '::/0' for ip in rule.get('Ipv6Ranges', []))
            if public_ipv4 or public_ipv6:
                public_rules.append(f'{sg_id} allows public inbound traffic on ports {port_text}')

    if public_rules:
        add_security_check(checks, service, resource, 'WARN', 'Security groups', '; '.join(public_rules))
    else:
        add_security_check(checks, service, resource, 'OK', 'Security groups', f'{len(sg_ids)} security group(s) without public inbound rules')


def collect_lambda_security(checks):
    service = 'Lambda'
    try:
        functions = clients['lambda'].list_functions().get('Functions', [])
    except Exception as exc:
        add_security_check(checks, service, 'Lambda', 'WARN', 'Discovery', f'Could not list functions: {_error_code(exc) or exc}')
        return

    old_runtimes = ('python2', 'python3.6', 'python3.7', 'nodejs10', 'nodejs12', 'nodejs14')
    secret_words = ('password', 'secret', 'key', 'token', 'credential', 'api_key', 'apikey', 'passwd', 'pwd')

    for fn in functions:
        name = fn.get('FunctionName', 'N/A')

        try:
            cfg = clients['lambda'].get_function_configuration(FunctionName=name)
        except Exception:
            # Same criterion as notebook 18: functions without configuration access are skipped.
            continue

        try:
            policy = clients['lambda'].get_policy(FunctionName=name)
            policy_doc = json.loads(policy.get('Policy', '{}'))
            statements = as_list(policy_doc.get('Statement', []))
            if public_policy_doc(policy_doc):
                add_security_check(checks, service, name, 'CRIT', 'Invoke policy', 'Public invocation without condition')
            else:
                permissions = []
                for stmt in statements:
                    principal = stmt.get('Principal', {})
                    if isinstance(principal, dict):
                        principal = principal.get('Service', principal.get('AWS', str(principal)))
                    condition = stmt.get('Condition', {})
                    source = ''
                    if condition:
                        for condition_value in condition.values():
                            if isinstance(condition_value, dict):
                                source_arn = condition_value.get('AWS:SourceArn', '')
                                if source_arn:
                                    source = f' <- {source_arn.split(":")[-1]}'
                    permissions.append(f'[OK] {principal}{source}')
                add_security_check(checks, service, name, 'INFO', 'Invoke policy', details_list(f'{len(statements)} rule(s)', permissions))
        except Exception as exc:
            if _is_missing(exc) or _error_code(exc) == 'ResourceNotFoundException':
                add_security_check(checks, service, name, 'OK', 'Invoke policy', 'No external invoke policy')
            else:
                add_security_check(checks, service, name, 'INFO', 'Invoke policy', f'Could not inspect invoke policy: {_error_code(exc) or exc}')

        try:
            url_configs = clients['lambda'].list_function_url_configs(FunctionName=name).get('FunctionUrlConfigs', [])
            if url_configs:
                for url in url_configs:
                    auth_type = url.get('AuthType', 'N/A')
                    if auth_type == 'NONE':
                        add_security_check(checks, service, name, 'CRIT', 'Function URL', f'Public URL without authentication: {url.get("FunctionUrl", "N/A")}')
                    else:
                        add_security_check(checks, service, name, 'OK', 'Function URL', f'Auth type: {auth_type}')
            else:
                add_security_check(checks, service, name, 'OK', 'Function URL', 'Not configured')
        except Exception:
            add_security_check(checks, service, name, 'OK', 'Function URL', 'Not configured')

        runtime = cfg.get('Runtime', 'N/A')
        if any(old in runtime for old in old_runtimes):
            add_security_check(checks, service, name, 'WARN', 'Runtime', f'Old runtime: {runtime}')
        else:
            add_security_check(checks, service, name, 'OK', 'Runtime', f'Runtime: {runtime}')

        env_vars = cfg.get('Environment', {}).get('Variables', {})
        suspicious = [key for key in env_vars if any(word in key.lower() for word in secret_words)]
        if suspicious:
            add_security_check(checks, service, name, 'WARN', 'Environment variables', f'Potential secret-like variables: {", ".join(suspicious)}')
        else:
            add_security_check(checks, service, name, 'OK', 'Environment variables', f'{len(env_vars)} variable(s), no secret-like names detected')

        vpc_config = cfg.get('VpcConfig', {})
        subnet_ids = vpc_config.get('SubnetIds', [])
        sg_ids = vpc_config.get('SecurityGroupIds', [])
        if subnet_ids:
            add_security_check(checks, service, name, 'OK', 'VPC configuration', f'{len(subnet_ids)} subnet(s), {len(sg_ids)} security group(s)')
            check_security_groups(checks, service, name, sg_ids)
        else:
            add_security_check(checks, service, name, 'INFO', 'VPC configuration', 'Not configured; runs in AWS public network')

        add_iam_role_security_checks(checks, service, name, cfg.get('Role'))

def collect_dynamodb_security(checks):
    service = 'DynamoDB'
    try:
        table_names = clients['dynamodb'].list_tables().get('TableNames', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list tables: {_error_code(exc) or exc}')
        return

    for table in table_names:
        try:
            desc = clients['dynamodb'].describe_table(TableName=table).get('Table', {})

            encryption = desc.get('SSEDescription', {})
            if encryption.get('Status') == 'ENABLED':
                add_security_check(checks, service, table, 'OK', 'Encryption', f"SSE enabled ({encryption.get('SSEType', 'KMS')})")
            else:
                add_security_check(checks, service, table, 'OK', 'Encryption', 'AES256 encryption enabled by default with AWS owned key')

            deletion_protection = desc.get('DeletionProtectionEnabled')
            if deletion_protection is True:
                add_security_check(checks, service, table, 'OK', 'Deletion protection', 'Enabled')
            elif deletion_protection is False:
                add_security_check(checks, service, table, 'WARN', 'Deletion protection', 'Disabled')
            else:
                add_security_check(checks, service, table, 'INFO', 'Deletion protection', 'Not available for this table')

            try:
                pitr = clients['dynamodb'].describe_continuous_backups(TableName=table)
                status = pitr.get('ContinuousBackupsDescription', {}).get('PointInTimeRecoveryDescription', {}).get('PointInTimeRecoveryStatus', 'DISABLED')
                if status == 'ENABLED':
                    add_security_check(checks, service, table, 'OK', 'Point-in-time recovery', 'Enabled')
                else:
                    add_security_check(checks, service, table, 'WARN', 'Point-in-time recovery', 'Disabled; no point-in-time recovery after accidental writes/deletes')
            except Exception as exc:
                add_security_check(checks, service, table, 'INFO', 'Point-in-time recovery', f'Could not inspect PITR: {_error_code(exc) or exc}')

            try:
                policy = clients['dynamodb'].get_resource_policy(ResourceArn=desc.get('TableArn', ''))
                policy_doc = json.loads(policy.get('Policy', '{}'))
                if public_policy_doc(policy_doc):
                    add_security_check(checks, service, table, 'CRIT', 'Resource policy', 'Public policy without condition')
                else:
                    add_security_check(checks, service, table, 'OK', 'Resource policy', 'Policy configured without public access')
            except Exception as exc:
                if _is_missing(exc):
                    add_security_check(checks, service, table, 'INFO', 'Resource policy', 'Not configured; access managed by IAM roles/policies')
                else:
                    add_security_check(checks, service, table, 'INFO', 'Resource policy', f'Could not inspect policy: {_error_code(exc) or exc}')
        except Exception as exc:
            add_security_check(checks, service, table, 'WARN', 'Table inspection', f'Could not inspect table: {_error_code(exc) or exc}')

def collect_apigateway_security(checks):
    service = 'API Gateway'
    try:
        apis = clients['apigateway'].get_rest_apis(limit=500).get('items', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list APIs: {_error_code(exc) or exc}')
        return

    for api in apis:
        api_id = api.get('id')
        name = api.get('name', api_id)

        endpoint_types = api.get('endpointConfiguration', {}).get('types', [])
        add_security_check(checks, service, name, 'INFO', 'Endpoint type', ', '.join(endpoint_types) or 'Not reported')

        try:
            authorizers = clients['apigateway'].get_authorizers(restApiId=api_id, limit=500).get('items', [])
            if authorizers:
                details = [f"{auth.get('name', 'N/A')} ({auth.get('type', 'N/A')})" for auth in authorizers]
                add_security_check(checks, service, name, 'OK', 'Authorizers', details_list(f'{len(authorizers)} authorizer(s)', details))
            else:
                add_security_check(checks, service, name, 'WARN', 'Authorizers', 'No authorizers configured; API is public unless protected elsewhere')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Authorizers', f'Could not inspect authorizers: {_error_code(exc) or exc}')

        try:
            api_keys = clients['apigateway'].get_api_keys(limit=500, includeValues=False).get('items', [])
            if api_keys:
                active = sum(1 for key in api_keys if key.get('enabled', False))
                add_security_check(checks, service, name, 'OK', 'API keys', f'{active} active key(s) of {len(api_keys)} total')
            else:
                add_security_check(checks, service, name, 'WARN', 'API keys', 'No API keys configured')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'API keys', f'Could not inspect API keys: {_error_code(exc) or exc}')

        try:
            validators = clients['apigateway'].get_request_validators(restApiId=api_id, limit=500).get('items', [])
            if validators:
                details = [validator.get('name', validator.get('id', 'N/A')) for validator in validators]
                add_security_check(checks, service, name, 'OK', 'Request validators', details_list(f'{len(validators)} validator(s)', details))
            else:
                add_security_check(checks, service, name, 'WARN', 'Request validators', 'No request validators configured; API accepts any request shape')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Request validators', f'Could not inspect validators: {_error_code(exc) or exc}')

        try:
            stages = clients['apigateway'].get_stages(restApiId=api_id).get('item', [])
            if not stages:
                add_security_check(checks, service, name, 'WARN', 'Stages', 'No stages detected')
            for stage in stages:
                stage_name = stage.get('stageName', 'N/A')
                stage_resource = f'{name}/{stage_name}'

                if stage.get('accessLogSettings'):
                    add_security_check(checks, service, stage_resource, 'OK', 'Access logging', f'Configured on {stage_name}')
                else:
                    add_security_check(checks, service, stage_resource, 'WARN', 'Access logging', f'Not configured on {stage_name}')

                method_settings = stage.get('methodSettings', {}).get('*/*', {})
                rate_limit = method_settings.get('throttlingRateLimit')
                if rate_limit:
                    add_security_check(checks, service, stage_resource, 'INFO', 'Throttling', f'{stage_name}: {rate_limit} req/s')
                else:
                    add_security_check(checks, service, stage_resource, 'INFO', 'Throttling', f'{stage_name}: AWS default throttling values')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Stages', f'Could not inspect stages: {_error_code(exc) or exc}')

        try:
            unauthenticated = []
            resources = clients['apigateway'].get_resources(restApiId=api_id, limit=500).get('items', [])
            for resource in resources:
                for method_name in resource.get('resourceMethods', {}):
                    if method_name == 'OPTIONS':
                        continue
                    try:
                        method = clients['apigateway'].get_method(
                            restApiId=api_id,
                            resourceId=resource.get('id'),
                            httpMethod=method_name,
                        )
                        if method.get('authorizationType') == 'NONE':
                            unauthenticated.append(f'{method_name} {resource.get("path", "/")}')
                    except Exception:
                        pass

            if unauthenticated:
                add_security_check(
                    checks,
                    service,
                    name,
                    'WARN',
                    'Method authorization',
                    details_list(f'{len(unauthenticated)} unauthenticated functional method(s)', unauthenticated),
                )
            else:
                add_security_check(checks, service, name, 'OK', 'Method authorization', 'All functional methods require authentication')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Method authorization', f'Could not inspect methods: {_error_code(exc) or exc}')

def collect_sqs_security(checks):
    service = 'SQS'
    try:
        queue_urls = clients['sqs'].list_queues().get('QueueUrls', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list queues: {_error_code(exc) or exc}')
        return

    for url in queue_urls:
        name = url.rsplit('/', 1)[-1]
        try:
            attrs = clients['sqs'].get_queue_attributes(QueueUrl=url, AttributeNames=['All']).get('Attributes', {})

            kms_key = attrs.get('KmsMasterKeyId', '')
            sqs_managed = attrs.get('SqsManagedSseEnabled', 'false')
            if kms_key:
                add_security_check(checks, service, name, 'OK', 'Encryption', f'KMS encryption enabled: {kms_key}')
            elif sqs_managed == 'true':
                add_security_check(checks, service, name, 'OK', 'Encryption', 'SSE-SQS enabled; managed by AWS')
            else:
                add_security_check(checks, service, name, 'WARN', 'Encryption', 'Encryption not configured')

            if attrs.get('RedrivePolicy'):
                add_security_check(checks, service, name, 'OK', 'Dead-letter queue', 'DLQ configured')
            else:
                is_dlq = 'dlq' in name.lower()
                if is_dlq:
                    add_security_check(checks, service, name, 'INFO', 'Dead-letter queue', 'This queue is a DLQ')
                else:
                    add_security_check(checks, service, name, 'WARN', 'Dead-letter queue', 'DLQ not configured')

            policy_str = attrs.get('Policy', '')
            if policy_str:
                policy_doc = json.loads(policy_str)
                if public_policy_doc(policy_doc):
                    add_security_check(checks, service, name, 'CRIT', 'Queue policy', 'Public policy without condition')
                else:
                    add_security_check(checks, service, name, 'OK', 'Queue policy', 'Policy configured without direct public access')
            else:
                add_security_check(checks, service, name, 'INFO', 'Queue policy', 'Not configured; access managed by IAM roles/policies')
        except Exception as exc:
            add_security_check(checks, service, name, 'WARN', 'Queue inspection', f'Could not inspect queue: {_error_code(exc) or exc}')

def collect_sns_security(checks):
    service = 'SNS'
    try:
        topics = clients['sns'].list_topics().get('Topics', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list topics: {_error_code(exc) or exc}')
        return

    for topic in topics:
        arn = topic.get('TopicArn', '')
        name = arn.split(':')[-1]
        try:
            attrs = clients['sns'].get_topic_attributes(TopicArn=arn).get('Attributes', {})
        except Exception:
            # Same criterion as notebook 18: inaccessible system topics are skipped.
            continue

        kms_key = attrs.get('KmsMasterKeyId', '')
        if kms_key:
            add_security_check(checks, service, name, 'OK', 'Encryption', f'KMS encryption configured: {kms_key}')
        else:
            add_security_check(checks, service, name, 'WARN', 'Encryption', 'KMS encryption not configured')

        policy_str = attrs.get('Policy', '')
        if policy_str:
            try:
                policy_doc = json.loads(policy_str)
                if public_policy_doc(policy_doc):
                    add_security_check(checks, service, name, 'CRIT', 'Topic policy', 'Public policy without condition')
                else:
                    add_security_check(checks, service, name, 'OK', 'Topic policy', 'Policy configured without direct public access')
            except Exception as exc:
                add_security_check(checks, service, name, 'INFO', 'Topic policy', f'Could not parse policy: {_error_code(exc) or exc}')
        else:
            add_security_check(checks, service, name, 'INFO', 'Topic policy', 'Not configured; access managed by IAM roles/policies')

        try:
            data_policy = clients['sns'].get_data_protection_policy(ResourceArn=arn)
            if data_policy.get('DataProtectionPolicy'):
                add_security_check(checks, service, name, 'OK', 'Data protection policy', 'Configured')
            else:
                add_security_check(checks, service, name, 'INFO', 'Data protection policy', 'Not configured')
        except Exception:
            add_security_check(checks, service, name, 'INFO', 'Data protection policy', 'Not configured')

def collect_s3_security(checks):
    service = 'S3'
    try:
        buckets = clients['s3'].list_buckets().get('Buckets', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list buckets: {_error_code(exc) or exc}')
        return

    bpa_keys = ['BlockPublicAcls', 'IgnorePublicAcls', 'BlockPublicPolicy', 'RestrictPublicBuckets']

    for bucket in buckets:
        name = bucket.get('Name')

        try:
            policy_status = clients['s3'].get_bucket_policy_status(Bucket=name)
            is_public = policy_status.get('PolicyStatus', {}).get('IsPublic', False)
            if is_public:
                add_security_check(checks, service, name, 'CRIT', 'Public access', 'Bucket is public; bucket policy allows external access')
            else:
                add_security_check(checks, service, name, 'OK', 'Public access', 'Bucket is private')
        except Exception as exc:
            if _is_missing(exc):
                add_security_check(checks, service, name, 'OK', 'Public access', 'Bucket is private; no bucket policy detected')
            else:
                add_security_check(checks, service, name, 'INFO', 'Public access', f'Could not inspect public status: {_error_code(exc) or exc}')

        try:
            bpa = clients['s3'].get_public_access_block(Bucket=name).get('PublicAccessBlockConfiguration', {})
            enabled = sum(1 for key in bpa_keys if bpa.get(key, False))
            details = [f'{key}: {"enabled" if bpa.get(key, False) else "disabled"}' for key in bpa_keys]
            if enabled == 4:
                add_security_check(checks, service, name, 'OK', 'Block public access', details_list('Enabled (4/4)', details))
            else:
                add_security_check(checks, service, name, 'WARN', 'Block public access', details_list(f'Partial block ({enabled}/4)', details))
        except Exception as exc:
            if _is_missing(exc):
                add_security_check(checks, service, name, 'WARN', 'Block public access', 'Not configured')
            else:
                add_security_check(checks, service, name, 'INFO', 'Block public access', f'Could not inspect: {_error_code(exc) or exc}')

        try:
            policy = clients['s3'].get_bucket_policy(Bucket=name)
            policy_doc = json.loads(policy.get('Policy', '{}'))
            statements = as_list(policy_doc.get('Statement', []))
            details = []
            for stmt in statements:
                effect = stmt.get('Effect', 'N/A')
                principal = stmt.get('Principal', 'N/A')
                action = stmt.get('Action', 'N/A')
                details.append(f'{effect} | Principal: {principal} | Action: {action}')
            add_security_check(checks, service, name, 'INFO', 'Bucket policy', details_list(f'{len(statements)} statement(s)', details))
        except Exception as exc:
            if _is_missing(exc):
                add_security_check(checks, service, name, 'INFO', 'Bucket policy', 'No bucket policy configured')
            else:
                add_security_check(checks, service, name, 'INFO', 'Bucket policy', f'Could not inspect: {_error_code(exc) or exc}')

        try:
            ownership = clients['s3'].get_bucket_ownership_controls(Bucket=name)
            rules = ownership.get('OwnershipControls', {}).get('Rules', [])
            rule = rules[0].get('ObjectOwnership', 'N/A') if rules else 'N/A'
            if rule == 'BucketOwnerEnforced':
                add_security_check(checks, service, name, 'OK', 'ACL / object ownership', 'ACLs disabled (BucketOwnerEnforced)')
            else:
                try:
                    acl = clients['s3'].get_bucket_acl(Bucket=name)
                    grants = acl.get('Grants', [])
                    public_acl = any(
                        'AllUsers' in grant.get('Grantee', {}).get('URI', '') or
                        'AuthenticatedUsers' in grant.get('Grantee', {}).get('URI', '')
                        for grant in grants
                    )
                    if public_acl:
                        add_security_check(checks, service, name, 'CRIT', 'ACL / object ownership', f'ACLs active with public access ({rule})')
                    else:
                        add_security_check(checks, service, name, 'WARN', 'ACL / object ownership', f'ACLs active ({rule}); AWS recommends BucketOwnerEnforced')
                except Exception:
                    add_security_check(checks, service, name, 'WARN', 'ACL / object ownership', f'ACLs active ({rule})')
        except Exception as exc:
            if _is_missing(exc):
                try:
                    acl = clients['s3'].get_bucket_acl(Bucket=name)
                    grants = acl.get('Grants', [])
                    public_acl = any(
                        'AllUsers' in grant.get('Grantee', {}).get('URI', '') or
                        'AuthenticatedUsers' in grant.get('Grantee', {}).get('URI', '')
                        for grant in grants
                    )
                    if public_acl:
                        add_security_check(checks, service, name, 'CRIT', 'ACL / object ownership', 'ACLs active with public access')
                    else:
                        add_security_check(checks, service, name, 'WARN', 'ACL / object ownership', 'ACLs active; AWS recommends BucketOwnerEnforced')
                except Exception:
                    add_security_check(checks, service, name, 'INFO', 'ACL / object ownership', 'Could not inspect ACLs')
            else:
                add_security_check(checks, service, name, 'INFO', 'ACL / object ownership', f'Could not inspect: {_error_code(exc) or exc}')

        try:
            encryption = clients['s3'].get_bucket_encryption(Bucket=name)
            rules = encryption.get('ServerSideEncryptionConfiguration', {}).get('Rules', [])
            algo = rules[0].get('ApplyServerSideEncryptionByDefault', {}).get('SSEAlgorithm', 'N/A') if rules else 'N/A'
            add_security_check(checks, service, name, 'OK', 'Encryption', f'At-rest encryption: {algo}')
        except Exception as exc:
            if _is_missing(exc):
                add_security_check(checks, service, name, 'WARN', 'Encryption', 'Encryption not configured')
            else:
                add_security_check(checks, service, name, 'INFO', 'Encryption', f'Could not inspect: {_error_code(exc) or exc}')

        try:
            versioning = clients['s3'].get_bucket_versioning(Bucket=name).get('Status')
            if versioning == 'Enabled':
                add_security_check(checks, service, name, 'OK', 'Versioning', 'Enabled')
            else:
                add_security_check(checks, service, name, 'WARN', 'Versioning', 'Not enabled')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Versioning', f'Could not inspect: {_error_code(exc) or exc}')

        try:
            logging_config = clients['s3'].get_bucket_logging(Bucket=name)
            if 'LoggingEnabled' in logging_config:
                target = logging_config.get('LoggingEnabled', {}).get('TargetBucket', 'N/A')
                add_security_check(checks, service, name, 'OK', 'Access logging', f'Enabled; target bucket: {target}')
            else:
                add_security_check(checks, service, name, 'WARN', 'Access logging', 'Not enabled')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Access logging', f'Could not inspect: {_error_code(exc) or exc}')

def collect_rds_security(checks):
    service = 'RDS'
    try:
        instances = clients['rds'].describe_db_instances().get('DBInstances', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list instances: {_error_code(exc) or exc}')
        return

    for db in instances:
        name = db.get('DBInstanceIdentifier', 'N/A')
        add_security_check(checks, service, name, 'OK' if db.get('StorageEncrypted') else 'CRIT', 'Encryption', 'Storage encrypted' if db.get('StorageEncrypted') else 'Storage encryption disabled')
        add_security_check(checks, service, name, 'CRIT' if db.get('PubliclyAccessible') else 'OK', 'Public access', f"PubliclyAccessible = {db.get('PubliclyAccessible')}")
        backup_days = db.get('BackupRetentionPeriod', 0)
        add_security_check(checks, service, name, 'OK' if backup_days else 'WARN', 'Backups', f'Backup retention: {backup_days} day(s)')
        add_security_check(checks, service, name, 'OK' if db.get('DeletionProtection') else 'WARN', 'Deletion protection', 'Enabled' if db.get('DeletionProtection') else 'Disabled')
        exports = db.get('EnabledCloudwatchLogsExports', [])
        add_security_check(checks, service, name, 'OK' if exports else 'WARN', 'CloudWatch logs', f'Exports: {", ".join(exports)}' if exports else 'No log exports configured')
        add_security_check(checks, service, name, 'INFO', 'Multi-AZ', str(db.get('MultiAZ')))
        add_security_check(checks, service, name, 'INFO', 'IAM DB auth', str(db.get('IAMDatabaseAuthenticationEnabled')))
        add_security_check(checks, service, name, 'INFO', 'Auto minor upgrade', str(db.get('AutoMinorVersionUpgrade')))
        check_security_groups(checks, service, name, [sg.get('VpcSecurityGroupId') for sg in db.get('VpcSecurityGroups', [])])


def collect_cognito_security(checks):
    service = 'Cognito'
    try:
        pools = clients['cognito'].list_user_pools(MaxResults=60).get('UserPools', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list user pools: {_error_code(exc) or exc}')
        return

    for pool in pools:
        pool_id = pool.get('Id')
        name = pool.get('Name', pool_id)
        try:
            desc = clients['cognito'].describe_user_pool(UserPoolId=pool_id).get('UserPool', {})

            mfa = desc.get('MfaConfiguration', 'OFF')
            if mfa in ('ON', 'OPTIONAL'):
                add_security_check(checks, service, name, 'OK', 'MFA', mfa)
            else:
                add_security_check(checks, service, name, 'WARN', 'MFA', 'Disabled')

            policy = desc.get('Policies', {}).get('PasswordPolicy', {})
            min_len = policy.get('MinimumLength', 0)
            requirements = [
                policy.get('RequireUppercase'),
                policy.get('RequireLowercase'),
                policy.get('RequireNumbers'),
                policy.get('RequireSymbols'),
            ]
            strong = min_len >= 8 and all(requirements)
            detail = (
                f'Min length {min_len}; uppercase={policy.get("RequireUppercase")}; '
                f'lowercase={policy.get("RequireLowercase")}; numbers={policy.get("RequireNumbers")}; '
                f'symbols={policy.get("RequireSymbols")}'
            )
            add_security_check(checks, service, name, 'OK' if strong else 'WARN', 'Password policy', detail)

            advanced = desc.get('UserPoolAddOns', {}).get('AdvancedSecurityMode', 'OFF')
            add_security_check(checks, service, name, 'INFO', 'Advanced security', advanced)
        except Exception as exc:
            add_security_check(checks, service, name, 'WARN', 'User pool inspection', f'Could not inspect pool: {_error_code(exc) or exc}')

        try:
            clients_list = clients['cognito'].list_user_pool_clients(UserPoolId=pool_id, MaxResults=60).get('UserPoolClients', [])
            app_rows = []
            implicit_clients = []
            http_callback_clients = []

            for app in clients_list:
                app_name = app.get('ClientName', app.get('ClientId'))
                details = clients['cognito'].describe_user_pool_client(
                    UserPoolId=pool_id,
                    ClientId=app.get('ClientId'),
                ).get('UserPoolClient', {})

                auth_flows = details.get('ExplicitAuthFlows', [])
                app_rows.append(f'{app_name}: {", ".join(auth_flows) or "No explicit auth flows reported"}')

                oauth_flows = details.get('AllowedOAuthFlows', [])
                if 'implicit' in oauth_flows:
                    implicit_clients.append(app_name)

                callbacks = details.get('CallbackURLs', [])
                public_http_callbacks = [url for url in callbacks if url.startswith('http://') and 'localhost' not in url]
                if public_http_callbacks:
                    http_callback_clients.append(f'{app_name}: {", ".join(public_http_callbacks)}')

            add_security_check(checks, service, name, 'INFO', 'App clients', details_list(f'{len(clients_list)} app client(s)', app_rows))

            if implicit_clients:
                add_security_check(checks, service, name, 'WARN', 'OAuth flow', details_list(f'{len(implicit_clients)} client(s) with implicit flow enabled', implicit_clients))
            else:
                add_security_check(checks, service, name, 'OK', 'OAuth flow', 'Implicit flow not enabled')

            if http_callback_clients:
                add_security_check(checks, service, name, 'WARN', 'Callback URLs', details_list(f'{len(http_callback_clients)} client(s) with public HTTP callback URLs', http_callback_clients))
            else:
                add_security_check(checks, service, name, 'OK', 'Callback URLs', 'No public HTTP callback URLs detected')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'App clients', f'Could not inspect app clients: {_error_code(exc) or exc}')

def collect_stepfunctions_security(checks):
    service = 'Step Functions'
    try:
        machines = clients['stepfunctions'].list_state_machines(maxResults=100).get('stateMachines', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list state machines: {_error_code(exc) or exc}')
        return

    for sm in machines:
        name = sm.get('name', 'N/A')
        try:
            desc = clients['stepfunctions'].describe_state_machine(stateMachineArn=sm.get('stateMachineArn'))
            add_security_check(checks, service, name, 'INFO', 'Type', desc.get('type', 'N/A'))

            logging_cfg = desc.get('loggingConfiguration', {})
            log_level = logging_cfg.get('level', 'OFF')
            if log_level != 'OFF':
                add_security_check(checks, service, name, 'OK', 'Logging', f'Level: {log_level}')
            else:
                add_security_check(checks, service, name, 'WARN', 'Logging', 'Disabled')

            tracing_enabled = desc.get('tracingConfiguration', {}).get('enabled', False)
            if tracing_enabled:
                add_security_check(checks, service, name, 'OK', 'X-Ray tracing', 'Enabled')
            else:
                add_security_check(checks, service, name, 'INFO', 'X-Ray tracing', 'Not enabled')

            add_iam_role_security_checks(checks, service, name, desc.get('roleArn'))
        except Exception as exc:
            add_security_check(checks, service, name, 'WARN', 'State machine inspection', f'Could not inspect state machine: {_error_code(exc) or exc}')

def collect_waf_security(checks):
    service = 'WAF'
    try:
        acls = clients['wafv2'].list_web_acls(Scope='REGIONAL', Limit=100).get('WebACLs', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list web ACLs: {_error_code(exc) or exc}')
        return

    for acl in acls:
        name = acl.get('Name', 'N/A')
        arn = acl.get('ARN')
        try:
            full = clients['wafv2'].get_web_acl(Name=name, Scope='REGIONAL', Id=acl.get('Id')).get('WebACL', {})

            rules = full.get('Rules', [])
            if rules:
                add_security_check(checks, service, name, 'OK', 'Rules', f'{len(rules)} rule(s) configured')
            else:
                add_security_check(checks, service, name, 'CRIT', 'Rules', 'Web ACL has no rules')

            managed = [rule for rule in rules if 'ManagedRuleGroupStatement' in rule.get('Statement', {})]
            if managed:
                add_security_check(checks, service, name, 'OK', 'Managed rules', f'{len(managed)} managed rule group(s)')
            else:
                add_security_check(checks, service, name, 'WARN', 'Managed rules', 'No managed rule groups configured')

            visibility = full.get('VisibilityConfig', {})
            if visibility.get('CloudWatchMetricsEnabled'):
                add_security_check(checks, service, name, 'OK', 'CloudWatch metrics', 'Enabled')
            else:
                add_security_check(checks, service, name, 'WARN', 'CloudWatch metrics', 'Disabled')

            try:
                clients['wafv2'].get_logging_configuration(ResourceArn=arn)
                add_security_check(checks, service, name, 'OK', 'Logging', 'Configured')
            except Exception as exc:
                add_security_check(checks, service, name, 'WARN' if _is_missing(exc) else 'INFO', 'Logging', 'Not configured' if _is_missing(exc) else f'Could not inspect: {_error_code(exc) or exc}')

            associated = []
            for resource_type in ['APPLICATION_LOAD_BALANCER', 'API_GATEWAY', 'APPSYNC', 'COGNITO_USER_POOL']:
                try:
                    resources = clients['wafv2'].list_resources_for_web_acl(
                        WebACLArn=arn,
                        ResourceType=resource_type,
                    ).get('ResourceArns', [])
                    for resource_arn in resources:
                        associated.append(f'{resource_type}: {resource_arn.split("/")[-1]}')
                except Exception:
                    pass

            if associated:
                add_security_check(checks, service, name, 'OK', 'Associations', details_list(f'{len(associated)} associated resource(s)', associated))
            else:
                add_security_check(checks, service, name, 'WARN', 'Associations', 'No associated resources')
        except Exception as exc:
            add_security_check(checks, service, name, 'WARN', 'Web ACL inspection', f'Could not inspect web ACL: {_error_code(exc) or exc}')

def collect_beanstalk_security(checks):
    service = 'Elastic Beanstalk'
    try:
        envs = clients['elasticbeanstalk'].describe_environments(IncludeDeleted=False).get('Environments', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list environments: {_error_code(exc) or exc}')
        return

    secret_words = ('password', 'secret', 'token', 'credential', 'api_key', 'apikey', 'passwd', 'pwd')
    for env in envs:
        name = env.get('EnvironmentName', 'N/A')
        health = env.get('Health', 'N/A')
        add_security_check(checks, service, name, 'OK' if health in ('Green', 'Ok') else 'WARN', 'Environment health', f'Health: {health}; environment status: {env.get("Status", "N/A")}')
        try:
            settings = clients['elasticbeanstalk'].describe_configuration_settings(
                ApplicationName=env.get('ApplicationName'),
                EnvironmentName=name,
            ).get('ConfigurationSettings', [])
            options = settings[0].get('OptionSettings', []) if settings else []
            env_vars = [o.get('OptionName', '') for o in options if o.get('Namespace') == 'aws:elasticbeanstalk:application:environment']
            suspicious = [key for key in env_vars if any(word in key.lower() for word in secret_words)]
            add_security_check(checks, service, name, 'WARN' if suspicious else 'OK', 'Environment variables', f'Potential secret-like variables: {", ".join(suspicious)}' if suspicious else f'{len(env_vars)} variable(s), no secret-like names detected')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Configuration settings', f'Could not inspect settings: {_error_code(exc) or exc}')


def collect_elasticache_security(checks):
    service = 'ElastiCache'
    try:
        clusters = clients['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True).get('CacheClusters', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list clusters: {_error_code(exc) or exc}')
        return

    for cluster in clusters:
        name = cluster.get('CacheClusterId', 'N/A')
        engine = cluster.get('Engine', 'N/A')
        at_rest = cluster.get('AtRestEncryptionEnabled')
        transit = cluster.get('TransitEncryptionEnabled')
        if engine == 'memcached':
            add_security_check(checks, service, name, 'WARN', 'At-rest encryption', 'At-rest encryption not detected')
            add_security_check(checks, service, name, 'INFO', 'Transit encryption', 'Not available for Memcached in this configuration')
            add_security_check(checks, service, name, 'INFO', 'Snapshots', 'Snapshots are not available for Memcached')
        else:
            add_security_check(checks, service, name, 'OK' if at_rest else 'WARN', 'At-rest encryption', f'Enabled: {bool(at_rest)}')
            add_security_check(checks, service, name, 'OK' if transit else 'WARN', 'Transit encryption', f'Enabled: {bool(transit)}')
            retention = cluster.get('SnapshotRetentionLimit')
            add_security_check(checks, service, name, 'OK' if retention else 'WARN', 'Snapshots', f'Retention: {retention or 0} day(s)')
        check_security_groups(checks, service, name, [sg.get('SecurityGroupId') for sg in cluster.get('SecurityGroups', [])])


def collect_ecr_security(checks):
    service = 'ECR'
    try:
        repos = clients['ecr'].describe_repositories().get('repositories', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list repositories: {_error_code(exc) or exc}')
        return

    for repo in repos:
        name = repo.get('repositoryName', 'N/A')
        scan_on_push = repo.get('imageScanningConfiguration', {}).get('scanOnPush', False)
        immutable = repo.get('imageTagMutability') == 'IMMUTABLE'

        add_security_check(checks, service, name, 'OK' if scan_on_push else 'WARN', 'Image scanning', 'Scan on push enabled' if scan_on_push else 'Scan on push disabled')
        add_security_check(checks, service, name, 'OK' if immutable else 'WARN', 'Tag immutability', 'Tags are immutable' if immutable else 'Tags are mutable and can be overwritten')

        try:
            clients['ecr'].get_lifecycle_policy(repositoryName=name)
            add_security_check(checks, service, name, 'OK', 'Lifecycle policy', 'Configured')
        except Exception as exc:
            add_security_check(checks, service, name, 'WARN' if _is_missing(exc) else 'INFO', 'Lifecycle policy', 'Not configured' if _is_missing(exc) else f'Could not inspect: {_error_code(exc) or exc}')

        try:
            images = clients['ecr'].describe_images(repositoryName=name).get('imageDetails', [])
            untagged = [img for img in images if not img.get('imageTags')]
            add_security_check(checks, service, name, 'WARN' if untagged else 'OK', 'Untagged images', f'{len(untagged)} untagged image(s)' if untagged else 'No untagged images')

            summary = {}
            for img in images:
                sev = img.get('imageScanFindingsSummary', {}).get('findingSeverityCounts', {})
                for key, value in sev.items():
                    summary[key] = summary.get(key, 0) + value

            if summary.get('CRITICAL', 0):
                add_security_check(checks, service, name, 'CRIT', 'Vulnerabilities', f"Critical findings: {summary.get('CRITICAL', 0)}")
            elif summary.get('HIGH', 0):
                add_security_check(checks, service, name, 'WARN', 'Vulnerabilities', f"High findings: {summary.get('HIGH', 0)}")
            elif summary:
                add_security_check(checks, service, name, 'OK', 'Vulnerabilities', f'No critical/high findings; summary: {summary}')
            else:
                add_security_check(checks, service, name, 'INFO', 'Vulnerabilities', 'No scan findings summary available')
        except Exception as exc:
            add_security_check(checks, service, name, 'INFO', 'Images', f'Could not inspect images: {_error_code(exc) or exc}')

def collect_codecommit_security(checks):
    service = 'CodeCommit'
    try:
        repos = clients['codecommit'].list_repositories().get('repositories', [])
    except Exception as exc:
        add_security_check(checks, service, service, 'WARN', 'Discovery', f'Could not list repositories: {_error_code(exc) or exc}')
        return

    for repo in repos:
        name = repo.get('repositoryName', 'N/A')
        try:
            metadata = clients['codecommit'].get_repository(repositoryName=name).get('repositoryMetadata', {})
            add_security_check(checks, service, name, 'INFO' if metadata.get('defaultBranch') else 'WARN', 'Default branch', metadata.get('defaultBranch') or 'No default branch detected')
            triggers = clients['codecommit'].get_repository_triggers(repositoryName=name).get('triggers', [])
            add_security_check(checks, service, name, 'INFO', 'Triggers', f'{len(triggers)} trigger(s) configured')
            branches = clients['codecommit'].list_branches(repositoryName=name).get('branches', [])
            add_security_check(checks, service, name, 'INFO', 'Branches', f'{len(branches)} branch(es)')
        except Exception as exc:
            add_security_check(checks, service, name, 'WARN', 'Repository inspection', f'Could not inspect repository: {_error_code(exc) or exc}')


SECURITY_COLLECTORS = {
    'Lambda': collect_lambda_security,
    'DynamoDB': collect_dynamodb_security,
    'API Gateway': collect_apigateway_security,
    'SQS': collect_sqs_security,
    'SNS': collect_sns_security,
    'S3': collect_s3_security,
    'RDS': collect_rds_security,
    'Cognito': collect_cognito_security,
    'Step Functions': collect_stepfunctions_security,
    'WAF': collect_waf_security,
    'Elastic Beanstalk': collect_beanstalk_security,
    'ElastiCache': collect_elasticache_security,
    'ECR': collect_ecr_security,
    'CodeCommit': collect_codecommit_security,
}


def collect_security_checks(service=None):
    checks = []
    selected = SECURITY_SERVICES if service in (None, 'Overview') else [service]
    for svc in selected:
        collector = SECURITY_COLLECTORS.get(svc)
        if collector:
            collector(checks)
    return checks
```

```python
# === Security renderers ===

def security_counts(checks):
    return {status: sum(1 for item in checks if item['status'] == status) for status in ['CRIT', 'WARN', 'INFO', 'OK']}


def render_security_donut(counts, title='Security status distribution'):
    total = sum(counts.values())
    if total == 0:
        return ''

    colors = {
        'CRIT': '#C94A44',
        'WARN': '#E39A2D',
        'INFO': '#7F8C8D',
        'OK': '#2FA36B',
    }
    label_names = {
        'CRIT': 'Critical',
        'WARN': 'Warning',
        'INFO': 'Info',
        'OK': 'OK',
    }
    labels = ['OK', 'WARN', 'CRIT', 'INFO']
    radius = 58
    circumference = 2 * 3.141592653589793 * radius
    offset = 0
    segments = []

    for label in labels:
        value = counts.get(label, 0)
        if value <= 0:
            continue
        length = circumference * value / total
        segments.append(
            f'<circle cx="80" cy="80" r="{radius}" fill="none" '
            f'stroke="{colors[label]}" stroke-width="24" '
            f'stroke-dasharray="{length:.2f} {circumference - length:.2f}" '
            f'stroke-dashoffset="{-offset:.2f}" stroke-linecap="butt" '
            f'transform="rotate(-90 80 80)" />'
        )
        offset += length

    legend_items = ''.join(
        f'<div style="display:flex; align-items:center; gap:8px; margin:4px 0; font-size:13px;">'
        f'<span style="width:10px; height:10px; background:{colors[label]}; display:inline-block; border-radius:2px;"></span>'
        f'<span style="min-width:70px;">{label_names[label]}</span><strong>{counts.get(label, 0)}</strong>'
        f'</div>'
        for label in labels
    )

    return (
        '<div style="display:flex; align-items:center; gap:28px; margin: 8px 0 18px 0; '
        'padding: 14px 16px; border:1px solid #e6e6e6; background:#fafafa;">'
        '<svg width="160" height="160" viewBox="0 0 160 160" role="img" aria-label="Security status distribution">'
        '<circle cx="80" cy="80" r="58" fill="none" stroke="#eeeeee" stroke-width="24" />'
        + ''.join(segments) +
        f'<text x="80" y="76" text-anchor="middle" style="font-size:24px; font-weight:700; fill:#2E5090;">{total}</text>'
        '<text x="80" y="96" text-anchor="middle" style="font-size:11px; fill:#777;">checks</text>'
        '</svg>'
        '<div>'
        f'<div style="font-weight:700; color:#2E5090; margin-bottom:8px;">{title}</div>'
        + legend_items +
        '</div>'
        '</div>'
    )

def render_security_table(checks, include_service=False):
    rows = []
    ordered = sorted(checks, key=lambda item: (SECURITY_ORDER.get(item['status'], 9), item['service'], item['resource'], item['check']))
    for item in ordered:
        row = [html_status(item['status'], '')]
        if include_service:
            row.append(item['service'])
        row.extend([f"<strong>{item['resource']}</strong>", item['check'], item['message']])
        rows.append(row)
    headers = ['Status']
    if include_service:
        headers.append('Service')
    headers.extend(['Resource', 'Check', 'Detail'])
    return html_table(headers, rows)


def render_security_findings(checks, include_service=False):
    findings = [item for item in checks if item['status'] in ('CRIT', 'WARN')]
    if not findings:
        return '<p class="muted">No critical or warning security findings detected.</p>'
    return render_security_table(findings, include_service=include_service)


def render_security_by_service(checks):
    rows = []
    for svc in SECURITY_SERVICES:
        svc_checks = [item for item in checks if item['service'] == svc]
        if not svc_checks:
            continue
        counts = security_counts(svc_checks)
        resources = len(set(item['resource'] for item in svc_checks))
        rows.append([
            svc,
            resources,
            counts['CRIT'],
            counts['WARN'],
            counts['INFO'],
            counts['OK'],
        ])
    return html_table(['Service', 'Resources', 'CRIT', 'WARN', 'INFO', 'OK'], rows)


def render_security_grouped_by_resource(checks):
    if not checks:
        return '<p class="muted">No security checks available for this service.</p>'

    html = ''
    resources = sorted(set(item['resource'] for item in checks))
    for resource in resources:
        resource_checks = [item for item in checks if item['resource'] == resource]
        counts = security_counts(resource_checks)
        html += (
            f'<h3 style="margin:18px 0 8px 0; color:#333; font-size:18px;">{resource}</h3>'
            f'<div style="color:#7f8c8d; font-size:12px; margin-bottom:8px;">'
            f'CRIT: {counts["CRIT"]} | WARN: {counts["WARN"]} | INFO: {counts["INFO"]} | OK: {counts["OK"]}'
            f'</div>'
        )
        rows = []
        ordered = sorted(resource_checks, key=lambda item: (SECURITY_ORDER.get(item['status'], 9), item['check']))
        for item in ordered:
            rows.append([html_status(item['status'], ''), item['check'], item['message']])
        html += html_table(['Status', 'Check', 'Detail'], rows)
    return html

def render_security_overview():
    checks = collect_security_checks()
    counts = security_counts(checks)
    html = html_title('Security Overview')
    html += render_security_donut(counts, 'Security status distribution')
    html += html_section('Findings by service')
    html += render_security_by_service(checks)
    html += html_section('Priority findings')
    html += render_security_findings(checks, include_service=True)
    return html

def render_security_service(service):
    checks = collect_security_checks(service)
    counts = security_counts(checks)
    html = html_title(f'{service} - Security')
    html += render_security_donut(counts, f'{service} security status distribution')
    html += html_section('Checks by resource')
    html += render_security_grouped_by_resource(checks)
    return html
```

```python
# === Logs collectors ===

LOG_ERROR_FILTER_PATTERNS = [
    '"ERROR"', '"Error"', '"error"',
    '"Traceback"', '"traceback"',
    '"Exception"', '"exception"',
    '"Task timed out"', '"task timed out"',
    '"FAILED"', '"Failed"', '"failed"',
    '"AccessDenied"', '"ValidationException"',
]
LOG_MONITORED_SERVICES = ['Lambda', 'API Gateway', 'Step Functions', 'Elastic Beanstalk', 'RDS', 'WAF']
LOG_COVERAGE_SERVICES = [service for service in SERVICES if service != 'Overview']
LOG_ERROR_REGEXES = [
    re.compile(pattern, re.IGNORECASE) for pattern in [
        r'\bERROR\b', r'traceback', r'exception', r'task timed out',
        r'\bFAILED\b', r'invoke error', r'unhandled', r'runtime\.exiterror',
        r'process exited', r'cannot find module', r'accessdenied', r'validationexception',
    ]
]
LOG_CLOUDTRAIL_SOURCES = {
    'Lambda': 'lambda.amazonaws.com',
    'DynamoDB': 'dynamodb.amazonaws.com',
    'API Gateway': 'apigateway.amazonaws.com',
    'SQS': 'sqs.amazonaws.com',
    'SNS': 'sns.amazonaws.com',
    'S3': 's3.amazonaws.com',
    'RDS': 'rds.amazonaws.com',
    'Cognito': 'cognito-idp.amazonaws.com',
    'Step Functions': 'states.amazonaws.com',
    'WAF': 'wafv2.amazonaws.com',
    'Elastic Beanstalk': 'elasticbeanstalk.amazonaws.com',
    'ElastiCache': 'elasticache.amazonaws.com',
    'ECR': 'ecr.amazonaws.com',
    'CodeCommit': 'codecommit.amazonaws.com',
}
LOG_EXCLUDED_LAMBDA_NAMES = {
    'MainMonitoringFunction',
    'ModLabRole',
    'RedshiftEventSubscription',
    'RedshiftOverwatch',
    'RoleCreationFunction',
}


def is_excluded_lambda_name(function_name):
    return function_name in LOG_EXCLUDED_LAMBDA_NAMES or function_name.startswith('Redshift')


def logs_time_window_ms():
    return int(start_time.timestamp() * 1000), int(end_time.timestamp() * 1000)


def classify_log_group(name):
    n = name.lower()
    if name.startswith('/aws/lambda/'):
        return 'Lambda'
    if name.startswith('API-Gateway-Execution-Logs') or '/aws/apigateway' in n or 'api-gateway' in n:
        return 'API Gateway'
    if name.startswith('/aws/vendedlogs/states') or '/states/' in n or 'stepfunctions' in n:
        return 'Step Functions'
    if name.startswith('/aws/elasticbeanstalk/'):
        return 'Elastic Beanstalk'
    if name.startswith('/aws/rds/'):
        return 'RDS'
    if name.startswith('aws-waf-logs-') or '/aws/waf' in n or 'waf' in n:
        return 'WAF'
    if name.startswith('/aws/codebuild/'):
        return 'CodeBuild'
    return 'Other'


def list_log_groups():
    try:
        paginator = clients['logs'].get_paginator('describe_log_groups')
        groups = []
        for page in paginator.paginate():
            groups.extend(page.get('logGroups', []))
        return groups
    except Exception:
        try:
            return clients['logs'].describe_log_groups().get('logGroups', [])
        except Exception:
            return []


def accessible_lambda_names():
    try:
        return {fn['FunctionName'] for fn in clients['lambda'].list_functions().get('Functions', [])}
    except Exception:
        return set()


def collect_log_groups():
    lambda_names = accessible_lambda_names()
    groups = []
    for group in list_log_groups():
        name = group.get('logGroupName', '')
        service = classify_log_group(name)
        lambda_function = ''
        lambda_status = ''
        excluded = False
        if name.startswith('/aws/lambda/'):
            lambda_function = name.replace('/aws/lambda/', '')
            excluded = is_excluded_lambda_name(lambda_function)
            lambda_status = 'excluded' if excluded else 'active' if lambda_function in lambda_names else 'orphan'
        groups.append({
            'name': name,
            'service': service,
            'size_mb': round(group.get('storedBytes', 0) / (1024 * 1024), 2),
            'retention': group.get('retentionInDays'),
            'lambda_function': lambda_function,
            'lambda_status': lambda_status,
            'excluded': excluded,
        })
    return groups


def filter_log_events(log_group, filter_pattern=None, limit=100):
    try:
        start_ms, end_ms = logs_time_window_ms()
        kwargs = {
            'logGroupName': log_group,
            'startTime': start_ms,
            'endTime': end_ms,
            'limit': limit,
            'interleaved': True,
        }
        if filter_pattern:
            kwargs['filterPattern'] = filter_pattern
        response = clients['logs'].filter_log_events(**kwargs)
        return sorted(response.get('events', []), key=lambda e: e.get('timestamp', 0), reverse=True)
    except Exception:
        return []


def compact_log_message(message, max_chars=180):
    msg = str(message or '').strip().replace('\r', ' ')
    msg = ' '.join(line.strip() for line in msg.split('\n') if line.strip())
    msg = re.sub(r'^\d{4}-\d{2}-\d{2}T[\d:.]+Z\s+[a-f0-9-]+\s+(INFO|WARN|ERROR|DEBUG)\s*', '', msg, flags=re.IGNORECASE)
    return msg[:max_chars] + ('...' if len(msg) > max_chars else '')




def detect_log_error_event(message):
    text = str(message or '')
    return any(regex.search(text) for regex in LOG_ERROR_REGEXES)


def classify_log_event_type(message):
    text = str(message or '').strip()
    upper = text.upper()
    if detect_log_error_event(text):
        return 'ERROR'
    if upper.startswith('REPORT'):
        return 'REPORT'
    if upper.startswith('START'):
        return 'START'
    if upper.startswith('END'):
        return 'END'
    if upper.startswith('INIT_START'):
        return 'INIT_START'
    if ' WARN ' in upper or upper.startswith('WARN'):
        return 'WARN'
    if ' INFO ' in upper or upper.startswith('INFO'):
        return 'INFO'
    return 'APPLICATION'


def log_enablement_status(service):
    try:
        if service == 'Lambda':
            return 'Lambda log groups are created after each function writes logs for the first time.'
        if service == 'API Gateway':
            apis = clients['apigateway'].get_rest_apis(limit=500).get('items', [])
            stages = []
            logging_enabled = 0
            for api in apis:
                for stage in clients['apigateway'].get_stages(restApiId=api['id']).get('item', []):
                    stages.append(f"{api['name']}/{stage['stageName']}")
                    if stage.get('accessLogSettings', {}).get('destinationArn'):
                        logging_enabled += 1
            if not stages:
                return 'No API Gateway stages detected.'
            return f'Access logging enabled on {logging_enabled}/{len(stages)} stage(s). Enable stage access logging to create API Gateway log groups.'
        if service == 'Step Functions':
            sms = clients['stepfunctions'].list_state_machines().get('stateMachines', [])
            enabled = 0
            for sm in sms:
                desc = clients['stepfunctions'].describe_state_machine(stateMachineArn=sm['stateMachineArn'])
                if desc.get('loggingConfiguration', {}).get('level', 'OFF') != 'OFF':
                    enabled += 1
            return f'Logging enabled on {enabled}/{len(sms)} state machine(s). Enable loggingConfiguration to send executions to CloudWatch Logs.'
        if service == 'RDS':
            dbs = clients['rds'].describe_db_instances().get('DBInstances', [])
            enabled = sum(1 for db in dbs if db.get('EnabledCloudwatchLogsExports'))
            return f'CloudWatch log exports configured on {enabled}/{len(dbs)} DB instance(s). Configure EnabledCloudwatchLogsExports to publish database logs.'
        if service == 'WAF':
            acls = clients['wafv2'].list_web_acls(Scope='REGIONAL').get('WebACLs', [])
            enabled = 0
            for acl in acls:
                try:
                    clients['wafv2'].get_logging_configuration(ResourceArn=acl['ARN'])
                    enabled += 1
                except Exception:
                    pass
            return f'Logging configured on {enabled}/{len(acls)} Web ACL(s). Configure WAF logging to publish request logs.'
        if service == 'Elastic Beanstalk':
            envs = clients['elasticbeanstalk'].describe_environments(IncludeDeleted=False).get('Environments', [])
            enabled = 0
            for env in envs:
                try:
                    settings = clients['elasticbeanstalk'].describe_configuration_settings(
                        ApplicationName=env['ApplicationName'], EnvironmentName=env['EnvironmentName']
                    ).get('ConfigurationSettings', [])
                    options = settings[0].get('OptionSettings', []) if settings else []
                    if any(opt.get('Namespace') == 'aws:elasticbeanstalk:cloudwatch:logs' and opt.get('OptionName') == 'StreamLogs' and str(opt.get('Value')).lower() == 'true' for opt in options):
                        enabled += 1
                except Exception:
                    pass
            return f'CloudWatch log streaming enabled on {enabled}/{len(envs)} environment(s). Enable StreamLogs to publish application logs.'
        if service == 'Cognito':
            return 'No log groups detected. Cognito can publish authentication logs if User Pool logging or advanced security logging is configured.'
        if service in ('DynamoDB', 'SQS', 'SNS', 'S3', 'ElastiCache', 'ECR', 'CodeCommit'):
            return 'This service usually does not create application log groups directly in this architecture; use CloudTrail activity for audit events.'
        return 'No logging diagnostic available for this service.'
    except Exception as exc:
        return f'Could not inspect logging configuration: {exc}'


def event_timestamp(event):
    return datetime.fromtimestamp(event.get('timestamp', 0) / 1000).strftime('%Y-%m-%d %H:%M:%S')


def collect_log_errors(log_groups=None, service=None, limit_per_group=200):
    groups = log_groups if log_groups is not None else collect_log_groups()
    findings = []
    for group in groups:
        if service and group['service'] != service:
            continue
        matched_events = {}
        for pattern in LOG_ERROR_FILTER_PATTERNS:
            events = filter_log_events(group['name'], filter_pattern=pattern, limit=limit_per_group)
            for event in events:
                msg = event.get('message', '')
                if detect_log_error_event(msg):
                    key = event.get('eventId') or f"{event.get('timestamp', 0)}:{msg}"
                    matched_events[key] = event
        error_events = sorted(matched_events.values(), key=lambda event: event.get('timestamp', 0), reverse=True)
        if error_events:
            latest = error_events[0]
            findings.append({
                'service': group['service'],
                'log_group': group['name'],
                'count': len(error_events),
                'latest_event': compact_log_message(latest.get('message', '')),
                'latest_raw_event': latest.get('message', ''),
                'latest_time': event_timestamp(latest),
            })
    return sorted(findings, key=lambda item: (item['service'], -item['count'], item['log_group']))
def collect_recent_log_activity(log_groups=None, service=None, limit_per_group=5, overall_limit=20):
    groups = log_groups if log_groups is not None else collect_log_groups()
    rows = []
    for group in groups:
        if service and group['service'] != service:
            continue
        if group.get('excluded'):
            continue
        events = filter_log_events(group['name'], limit=limit_per_group)
        for event in events:
            event_type = classify_log_event_type(event.get('message', ''))
            rows.append({
                'timestamp_raw': event.get('timestamp', 0),
                'timestamp': event_timestamp(event),
                'service': group['service'],
                'log_group': group['name'],
                'type': event_type,
                'message': str(event.get('message', '')).strip(),
            })
    return sorted(rows, key=lambda item: item['timestamp_raw'], reverse=True)[:overall_limit]


def collect_lambda_cold_starts(log_groups=None):
    groups = log_groups if log_groups is not None else collect_log_groups()
    rows = []
    for group in groups:
        if group['service'] != 'Lambda' or group.get('lambda_status') in ('orphan', 'excluded'):
            continue
        events = filter_log_events(group['name'], 'REPORT', limit=100)
        reports = 0
        cold_starts = 0
        init_durations = []
        for event in events:
            msg = event.get('message', '')
            if 'REPORT' not in msg:
                continue
            reports += 1
            match = re.search(r'Init Duration: ([\d.]+) ms', msg)
            if match:
                cold_starts += 1
                init_durations.append(float(match.group(1)))
        if reports:
            rows.append({
                'function': group.get('lambda_function') or group['name'].split('/')[-1],
                'reports': reports,
                'cold_starts': cold_starts,
                'cold_start_pct': round(cold_starts / reports * 100, 1),
                'avg_init_ms': round(sum(init_durations) / len(init_durations), 1) if init_durations else 0,
                'max_init_ms': round(max(init_durations), 1) if init_durations else 0,
            })
    return sorted(rows, key=lambda item: item['cold_start_pct'], reverse=True)


def collect_lambda_memory_usage(log_groups=None):
    groups = log_groups if log_groups is not None else collect_log_groups()
    rows = []
    for group in groups:
        if group['service'] != 'Lambda' or group.get('lambda_status') in ('orphan', 'excluded'):
            continue
        events = filter_log_events(group['name'], 'REPORT', limit=100)
        used_values = []
        assigned_mb = None
        for event in events:
            msg = event.get('message', '')
            used_match = re.search(r'Max Memory Used: (\d+) MB', msg)
            assigned_match = re.search(r'Memory Size: (\d+) MB', msg)
            if used_match:
                used_values.append(int(used_match.group(1)))
            if assigned_match:
                assigned_mb = int(assigned_match.group(1))
        if used_values and assigned_mb:
            max_used = max(used_values)
            rows.append({
                'function': group.get('lambda_function') or group['name'].split('/')[-1],
                'assigned_mb': assigned_mb,
                'avg_used_mb': round(sum(used_values) / len(used_values)),
                'max_used_mb': max_used,
                'max_usage_pct': round(max_used / assigned_mb * 100, 1),
                'free_margin_mb': assigned_mb - max_used,
            })
    return sorted(rows, key=lambda item: item['max_usage_pct'], reverse=True)


def collect_lambda_latest_executions(log_groups=None, limit_per_function=20):
    groups = log_groups if log_groups is not None else collect_log_groups()
    rows = []
    for group in groups:
        if group['service'] != 'Lambda' or group.get('lambda_status') in ('orphan', 'excluded'):
            continue
        function = group.get('lambda_function') or group['name'].split('/')[-1]
        events = sorted(filter_log_events(group['name'], limit=limit_per_function), key=lambda e: e.get('timestamp', 0), reverse=True)
        for event in events[:limit_per_function]:
            msg = event.get('message', '').strip()
            duration = re.search(r'Duration: ([\d.]+) ms', msg)
            memory = re.search(r'Max Memory Used: (\d+) MB', msg)
            init = re.search(r'Init Duration: ([\d.]+) ms', msg)
            runtime = ''
            if 'Runtime Version:' in msg:
                runtime = msg.split('Runtime Version:', 1)[1].split('Runtime', 1)[0].strip()
            rows.append({
                'timestamp_raw': event.get('timestamp', 0),
                'timestamp': event_timestamp(event),
                'function': function,
                'log_group': group['name'],
                'type': classify_log_event_type(msg),
                'duration_ms': duration.group(1) if duration else 'N/A',
                'memory_mb': memory.group(1) if memory else 'N/A',
                'cold_start_ms': init.group(1) if init else '',
                'runtime': runtime,
                'message': msg,
            })
    return sorted(rows, key=lambda item: (item['function'].lower(), -item['timestamp_raw']))

def _nested_value(data, path):
    current = data
    for part in path:
        if not isinstance(current, dict):
            return None
        current = current.get(part)
    return current


def event_resource_from_cloudtrail(event):
    resources = event.get('Resources') or []
    names = [res.get('ResourceName') for res in resources if res.get('ResourceName')]
    if names:
        return ', '.join(names[:3])

    try:
        detail = json.loads(event.get('CloudTrailEvent', '{}'))
    except Exception:
        detail = {}

    paths = [
        ('requestParameters', 'functionName'),
        ('requestParameters', 'bucketName'),
        ('requestParameters', 'tableName'),
        ('requestParameters', 'queueName'),
        ('requestParameters', 'topicArn'),
        ('requestParameters', 'repositoryName'),
        ('requestParameters', 'stateMachineArn'),
        ('requestParameters', 'userPoolId'),
        ('requestParameters', 'restApiId'),
        ('requestParameters', 'webACLName'),
        ('requestParameters', 'dBInstanceIdentifier'),
        ('requestParameters', 'cacheClusterId'),
        ('requestParameters', 'environmentName'),
        ('responseElements', 'functionName'),
        ('responseElements', 'tableDescription', 'tableName'),
        ('responseElements', 'queueUrl'),
        ('responseElements', 'repositoryMetadata', 'repositoryName'),
    ]
    found = []
    for path in paths:
        value = _nested_value(detail, path)
        if value:
            if isinstance(value, list):
                found.extend(str(item) for item in value[:3])
            else:
                found.append(str(value))
    return ', '.join(dict.fromkeys(found[:3]))

def collect_cloudtrail_events(service=None, max_results_per_service=200):
    selected = {service: LOG_CLOUDTRAIL_SOURCES.get(service)} if service else LOG_CLOUDTRAIL_SOURCES
    rows = []
    for svc, source in selected.items():
        if not source:
            continue
        collected_for_service = 0
        next_token = None
        truncated = False
        while collected_for_service < max_results_per_service:
            try:
                page_size = min(50, max_results_per_service - collected_for_service)
                kwargs = {
                    'LookupAttributes': [{'AttributeKey': 'EventSource', 'AttributeValue': source}],
                    'StartTime': start_time,
                    'EndTime': end_time,
                    'MaxResults': page_size,
                }
                if next_token:
                    kwargs['NextToken'] = next_token
                response = clients['cloudtrail'].lookup_events(**kwargs)
                events = response.get('Events', [])
                next_token = response.get('NextToken')
                if next_token and collected_for_service + len(events) >= max_results_per_service:
                    truncated = True
                for event in events:
                    rows.append({
                        'timestamp_raw': event.get('EventTime'),
                        'date': event.get('EventTime').strftime('%Y-%m-%d %H:%M:%S') if event.get('EventTime') else '',
                        'service': svc,
                        'event': event.get('EventName', ''),
                        'user': event.get('Username', ''),
                        'resource': event_resource_from_cloudtrail(event),
                        'sample_limited': truncated,
                    })
                collected_for_service += len(events)
                if not next_token or not events:
                    break
            except Exception:
                break
    return sorted(rows, key=lambda item: item['timestamp_raw'] or datetime.min, reverse=True)
```

```python
# === Logs renderers ===

def render_log_coverage(log_groups):
    services = LOG_COVERAGE_SERVICES
    rows = []
    for service in services:
        service_groups = [group for group in log_groups if group['service'] == service]
        total_size = round(sum(group['size_mb'] for group in service_groups), 2)
        no_retention = sum(1 for group in service_groups if not group.get('retention'))
        if not service_groups:
            retention = 'No log groups detected'
        elif no_retention:
            retention = f'{no_retention}/{len(service_groups)} without retention policy'
        else:
            retention = 'Retention configured'
        diagnostic = log_enablement_status(service)
        rows.append((service, str(len(service_groups)), f'{total_size} MB', retention, diagnostic))
    return html_table(['Service', 'Log groups', 'Total size', 'Retention', 'Logging diagnostic'], rows, widths=['160px', '90px', '100px', '170px', 'auto'])


def render_log_groups_table(log_groups, service=None):
    groups = [group for group in log_groups if not service or group['service'] == service]
    if not groups:
        return html_empty('No CloudWatch Logs groups detected for this service.')
    rows = []
    for group in sorted(groups, key=lambda item: item['name']):
        retention = f"{group['retention']} days" if group.get('retention') else 'No retention policy'
        status = group.get('lambda_status') or group['service']
        rows.append((group['service'], group['name'], f"{group['size_mb']} MB", retention, status))
    return html_table(['Service', 'Log group', 'Size', 'Retention', 'Status'], rows, widths=['150px', 'auto', '100px', '150px', '120px'])


def render_log_errors_table(errors):
    if not errors:
        return html_success('No matching error log patterns detected in CloudWatch Logs.')
    rows = [
        (
            item['service'],
            item['log_group'],
            str(item['count']),
            item['latest_time'],
            log_type_details('ERROR', item.get('latest_raw_event') or item.get('latest_event', ''))
        )
        for item in errors
    ]
    return html_table(['Service', 'Log group', 'Errors found', 'Latest time', 'Latest event'], rows, widths=['130px', '260px', '110px', '150px', 'auto'])


def render_log_errors_chart(errors, title='Error patterns by log group'):
    if not errors:
        return ''
    chart_rows = sorted(errors, key=lambda item: item['count'], reverse=True)[:12]
    labels = [item['log_group'] for item in chart_rows]
    counts = [item['count'] for item in chart_rows]
    fig = go.Figure(go.Bar(
        x=counts,
        y=labels,
        orientation='h',
        marker_color='#C94A44',
        hovertemplate='%{y}<br>Errors: %{x}<extra></extra>'
    ))
    fig.update_layout(
        title=title,
        height=max(280, 120 + len(chart_rows) * 38),
        margin=dict(t=45, b=30, l=260, r=30),
        template='plotly_white',
        xaxis_title='Detected error events',
        yaxis_title='',
        yaxis=dict(autorange='reversed')
    )
    return fig.to_html(full_html=False, include_plotlyjs='cdn')


def log_type_details(event_type, raw_message):
    raw = html_escape(raw_message or '')
    return (
        '<details style="margin:0;">'
        f'<summary style="cursor:pointer; color:#2E5090; font-weight:600;">{html_escape(event_type)}</summary>'
        f'<pre style="white-space:pre-wrap; margin:8px 0 0 0; font-family:Consolas, monospace; font-size:12px; color:#333;">{raw}</pre>'
        '</details>'
    )


def render_recent_log_activity_table(events):
    if not events:
        return html_empty('No recent CloudWatch Logs events detected in the selected range.')
    rows = [
        (item['timestamp'], item['service'], item['log_group'], log_type_details(item.get('type', 'APPLICATION'), item.get('message', '')))
        for item in events
    ]
    return html_table(['Timestamp', 'Service', 'Log group', 'Type'], rows, widths=['150px', '120px', '300px', 'auto'])


def render_cloudtrail_table(events, include_service=True, limit=60):
    if not events:
        return html_empty('No CloudTrail events detected in the selected range.')
    selected = events[:limit]
    if include_service:
        rows = [(item['date'], item['service'], item['event'], item['user'], item['resource'] or '-') for item in selected]
        return html_table(['Date', 'Service', 'Event', 'User', 'Resource'], rows, widths=['150px', '130px', '180px', '180px', 'auto'])
    rows = [(item['date'], item['event'], item['user'], item['resource'] or '-') for item in selected]
    return html_table(['Date', 'Event', 'User', 'Resource'], rows, widths=['150px', '200px', '180px', 'auto'])

def render_lambda_cold_start_analysis(rows):
    if not rows:
        return html_empty('Cold start analysis is not available because no REPORT events were detected.')
    high = [row for row in rows if row.get('cold_start_pct', 0) > 50]
    if not high:
        return html_success('No Lambda functions above the cold start threshold.')
    data = [
        (html_status('WARN', ''), row['function'], f"High cold start ratio: {row['cold_start_pct']}% ({row['cold_starts']}/{row['reports']} REPORT events).")
        for row in high
    ]
    return html_table(['Level', 'Function', 'Observation'], data, widths=['110px', '220px', 'auto'])


def render_lambda_memory_analysis(rows):
    if not rows:
        return html_empty('Memory usage analysis is not available because no REPORT memory data was detected.')
    notes = []
    for row in rows:
        usage = row.get('max_usage_pct', 0)
        if usage < 50:
            notes.append((html_status('INFO', ''), row['function'], f"Low max memory usage: {usage}%; function may be over-provisioned."))
        elif usage > 85:
            notes.append((html_status('WARN', ''), row['function'], f"High max memory usage: {usage}%; memory may be tight."))
    if not notes:
        return html_success('All functions have balanced memory usage based on the current thresholds.')
    return html_table(['Level', 'Function', 'Observation'], notes, widths=['110px', '220px', 'auto'])


def render_lambda_cold_starts_table(rows):
    if not rows:
        return html_empty('No Lambda REPORT events with cold start data detected in the selected range.')
    data = [(r['function'], str(r['reports']), str(r['cold_starts']), f"{r['cold_start_pct']}%", f"{r['avg_init_ms']} ms", f"{r['max_init_ms']} ms") for r in rows]
    return html_table(['Function', 'Reports', 'Cold starts', 'Cold start %', 'Avg init', 'Max init'], data)


def render_lambda_memory_table(rows):
    if not rows:
        return html_empty('No Lambda REPORT memory data detected in the selected range.')
    data = [(r['function'], f"{r['assigned_mb']} MB", f"{r['avg_used_mb']} MB", f"{r['max_used_mb']} MB", f"{r['max_usage_pct']}%", f"{r['free_margin_mb']} MB") for r in rows]
    return html_table(['Function', 'Assigned', 'Avg used', 'Max used', 'Max usage', 'Free margin'], data)


def render_lambda_executions_table(rows):
    if not rows:
        return html_empty('No Lambda execution events detected in the selected range.')
    html = ''
    for function in sorted({row['function'] for row in rows}, key=str.lower):
        function_rows = sorted([row for row in rows if row['function'] == function], key=lambda row: row['timestamp_raw'], reverse=True)
        html += html_subsection(f'{function} ({len(function_rows)} events)')
        data = [
            (r['timestamp'], r.get('duration_ms', 'N/A'), r.get('memory_mb', 'N/A'), r.get('cold_start_ms') or '-', log_type_details(r.get('type', 'APPLICATION'), r.get('message', '')))
            for r in function_rows
        ]
        html += html_table(['Timestamp', 'Duration ms', 'Memory MB', 'Cold start ms', 'Type'], data, widths=['150px', '100px', '100px', '110px', 'auto'])
    return html


def render_recent_log_activity_by_service(rows):
    if not rows:
        return html_empty('No recent CloudWatch Logs events detected in the selected range.')
    html = ''
    for service in sorted({row['service'] for row in rows}, key=str.lower):
        service_rows = sorted([row for row in rows if row['service'] == service], key=lambda row: row['timestamp_raw'], reverse=True)
        html += html_subsection(service)
        data = [
            (row['timestamp'], row['log_group'], log_type_details(row.get('type', 'APPLICATION'), row.get('message', '')))
            for row in service_rows
        ]
        html += html_table(['Timestamp', 'Log group', 'Type'], data, widths=['150px', '300px', 'auto'])
    return html

def render_logs_overview():
    log_groups = collect_log_groups()
    recent = collect_recent_log_activity(log_groups, limit_per_group=20, overall_limit=40)
    trail = collect_cloudtrail_events(max_results_per_service=200)

    html = html_title('Logs Overview')
    html += html_metric_range(include_period=True)

    html += html_section('Log coverage')
    html += render_log_coverage(log_groups)

    orphan_lambda = [group for group in log_groups if group.get('lambda_status') == 'orphan']
    if orphan_lambda:
        html += html_subsection('Orphan Lambda log groups')
        rows = [(group['lambda_function'], group['name'], f"{group['size_mb']} MB") for group in orphan_lambda]
        html += html_table(['Function', 'Log group', 'Size'], rows, widths=['220px', 'auto', '100px'])

    errors = collect_log_errors(log_groups)
    html += html_section('Error log patterns')
    html += render_log_errors_chart(errors)
    html += render_log_errors_table(errors)

    html += html_section('Recent CloudWatch Logs activity')
    html += html_note('Shows recent events sampled from detected log groups. Expand the Type field to inspect the raw log message.')
    html += render_recent_log_activity_by_service(recent)

    html += html_section('Latest CloudTrail events')
    html += html_note('CloudTrail shows management and audit API calls. Many entries may be generated by this dashboard; application-level activity is better represented by CloudWatch Metrics and CloudWatch Logs.')
    html += render_cloudtrail_table(trail, include_service=True, limit=40)

    html += html_section('Analysis summary')
    lambda_cs = collect_lambda_cold_starts(log_groups)
    lambda_mem = collect_lambda_memory_usage(log_groups)
    summary_rows = [
        ('CloudWatch log groups', str(len(log_groups))),
        ('Services with log groups', str(len({group['service'] for group in log_groups}))),
        ('Log groups with error patterns', str(len(errors))),
        ('Lambda functions with cold start data', str(len(lambda_cs))),
        ('Lambda functions with memory data', str(len(lambda_mem))),
    ]
    html += html_table(['Indicator', 'Value'], summary_rows, widths=['300px', 'auto'])
    return html


def render_logs_lambda(log_groups):
    lambda_groups = [group for group in log_groups if group['service'] == 'Lambda']
    html = html_title('Lambda - Logs')
    html += html_metric_range(include_period=True)

    html += html_section('Log groups')
    html += render_log_groups_table(log_groups, 'Lambda')

    html += html_section('Error log patterns')
    lambda_errors = collect_log_errors(lambda_groups, service='Lambda')
    html += render_log_errors_chart(lambda_errors, title='Lambda error patterns by log group')
    html += render_log_errors_table(lambda_errors)

    html += html_section('Cold starts')
    html += html_note('Based on up to 100 REPORT log events per function.')
    cold_start_rows = collect_lambda_cold_starts(lambda_groups)
    html += render_lambda_cold_starts_table(cold_start_rows)
    html += html_subsection('Cold start analysis')
    html += render_lambda_cold_start_analysis(cold_start_rows)

    html += html_section('Memory usage')
    html += html_note('Based on up to 100 REPORT log events per function.')
    memory_rows = collect_lambda_memory_usage(lambda_groups)
    html += render_lambda_memory_table(memory_rows)
    html += html_subsection('Memory usage analysis')
    html += render_lambda_memory_analysis(memory_rows)

    html += html_section('Latest executions')
    html += html_note('Shows up to 20 recent events per Lambda log group with events in the selected range. Active log groups without recent events are listed above but do not appear here. Expand the Type field to inspect the raw log message.')
    html += render_lambda_executions_table(collect_lambda_latest_executions(lambda_groups, limit_per_function=20))

    html += html_section('CloudTrail activity')
    html += render_cloudtrail_table(collect_cloudtrail_events('Lambda'), include_service=False, limit=30)
    return html


def render_logs_service(service):
    log_groups = collect_log_groups()
    if service == 'Lambda':
        return render_logs_lambda(log_groups)

    service_groups = [group for group in log_groups if group['service'] == service]
    html = html_title(f'{service} - Logs')
    html += html_metric_range(include_period=True)

    html += html_section('CloudWatch log groups')
    html += render_log_groups_table(log_groups, service)

    html += html_section('Error log patterns')
    service_errors = collect_log_errors(service_groups, service=service)
    html += render_log_errors_chart(service_errors, title=f'{service} error patterns by log group')
    html += render_log_errors_table(service_errors)

    html += html_section('Recent CloudWatch Logs activity')
    html += render_recent_log_activity_table(collect_recent_log_activity(service_groups, service=service, overall_limit=20))

    html += html_section('CloudTrail activity')
    html += render_cloudtrail_table(collect_cloudtrail_events(service), include_service=False, limit=30)
    return html
```

```python
# === Discovery collectors ===

DISCOVERY_SERVICES = [svc for svc in SERVICES if svc != 'Overview']


def _safe_aws(default, fn):
    try:
        return fn()
    except Exception:
        return default


def _resource_name_from_arn(arn):
    if not arn:
        return ''
    return str(arn).split(':')[-1].split('/')[-1]


def _lambda_name_from_uri(uri):
    if not uri:
        return ''
    match = re.search(r':function:([^:/]+)', uri)
    return match.group(1) if match else ''



def _ecr_repo_from_image_uri(image_uri):
    if not image_uri or '.dkr.ecr.' not in str(image_uri):
        return ''
    path = str(image_uri).split('/', 1)[-1]
    path = path.split('@', 1)[0]
    path = path.rsplit(':', 1)[0]
    return path

def _api_id_from_stage_arn(arn):
    match = re.search(r'/restapis/([^/]+)/stages/([^/]+)', arn or '')
    if not match:
        return '', ''
    return match.group(1), match.group(2)


def _walk_values(value):
    if isinstance(value, dict):
        for item in value.values():
            yield from _walk_values(item)
    elif isinstance(value, list):
        for item in value:
            yield from _walk_values(item)
    else:
        yield value


def _discovery_details(items):
    return sorted(str(item) for item in items if item not in (None, ''))


def _add_relation(relations, source_service, source, target_service, target, relation, evidence='', confidence='Direct'):
    """Adds a relationship, merging repeated edges and keeping all evidence."""
    if not source or not target:
        return

    source = str(source)
    target = str(target)
    evidence = str(evidence or '').strip()
    key = (source_service, source, target_service, target, relation, confidence)

    for existing in relations:
        existing_key = (
            existing['source_service'], existing['source'], existing['target_service'],
            existing['target'], existing['relation'], existing['confidence']
        )
        if existing_key == key:
            if evidence:
                evidence_items = [item.strip() for item in existing.get('evidence', '').split(';') if item.strip()]
                if evidence not in evidence_items:
                    evidence_items.append(evidence)
                existing['evidence'] = '; '.join(evidence_items)
            return

    relations.append({
        'source_service': source_service,
        'source': source,
        'relation': relation,
        'target_service': target_service,
        'target': target,
        'evidence': evidence,
        'confidence': confidence,
    })



def _value_mentions(value, candidates):
    text = str(value or '')
    if not text:
        return []
    matches = []
    for label, identifiers in candidates:
        for identifier in identifiers:
            identifier = str(identifier or '')
            if identifier and identifier in text:
                matches.append((label, identifier))
                break
    return matches


def _environment_variables_from_lambda(function_name):
    config = _safe_aws({}, lambda function_name=function_name: clients['lambda'].get_function_configuration(FunctionName=function_name))
    return (config.get('Environment') or {}).get('Variables', {}) or {}


def _elastic_beanstalk_environment_variables(env_name, app_name):
    settings = _safe_aws([], lambda env_name=env_name, app_name=app_name: clients['elasticbeanstalk'].describe_configuration_settings(
        ApplicationName=app_name,
        EnvironmentName=env_name
    ).get('ConfigurationSettings', []))
    variables = {}
    for setting_group in settings:
        for option in setting_group.get('OptionSettings', []) or []:
            if option.get('Namespace') == 'aws:elasticbeanstalk:application:environment':
                variables[option.get('OptionName')] = option.get('Value')
    return variables


def _discovery_resource_candidates():
    tables = _safe_aws([], lambda: clients['dynamodb'].list_tables().get('TableNames', []))
    dynamodb = [(table, [table]) for table in tables]

    buckets = _safe_aws([], lambda: clients['s3'].list_buckets().get('Buckets', []))
    s3 = [(bucket.get('Name'), [bucket.get('Name')]) for bucket in buckets]

    queues = _safe_aws([], lambda: clients['sqs'].list_queues().get('QueueUrls', []))
    sqs = []
    for url in queues:
        name = url.rstrip('/').split('/')[-1]
        sqs.append((name, [name, url]))

    topics = _safe_aws([], lambda: clients['sns'].list_topics().get('Topics', []))
    sns = []
    for topic in topics:
        arn = topic.get('TopicArn')
        name = _resource_name_from_arn(arn)
        sns.append((name, [name, arn]))

    rds_instances = _safe_aws([], lambda: clients['rds'].describe_db_instances().get('DBInstances', []))
    rds = []
    for instance in rds_instances:
        instance_id = instance.get('DBInstanceIdentifier')
        endpoint = ((instance.get('Endpoint') or {}).get('Address'))
        rds.append((instance_id, [instance_id, endpoint]))

    cache_clusters = _safe_aws([], lambda: clients['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True).get('CacheClusters', []))
    elasticache = []
    for cluster in cache_clusters:
        cluster_id = cluster.get('CacheClusterId')
        identifiers = [cluster_id]
        config_endpoint = cluster.get('ConfigurationEndpoint') or {}
        if config_endpoint.get('Address'):
            identifiers.append(config_endpoint.get('Address'))
        for node in cluster.get('CacheNodes', []) or []:
            endpoint = node.get('Endpoint') or {}
            if endpoint.get('Address'):
                identifiers.append(endpoint.get('Address'))
        elasticache.append((cluster_id, identifiers))

    return {
        'DynamoDB': dynamodb,
        'S3': s3,
        'SQS': sqs,
        'SNS': sns,
        'RDS': rds,
        'ElastiCache': elasticache,
    }


def _add_inferred_environment_relations(relations, source_service, source_label, variables, candidates):
    for var_name, var_value in (variables or {}).items():
        evidence_value = str(var_value or '')
        for target_service, target_candidates in candidates.items():
            for resource_name, matched_identifier in _value_mentions(evidence_value, target_candidates):
                _add_relation(
                    relations,
                    source_service,
                    source_label,
                    target_service,
                    f'{target_service}: {resource_name}',
                    'Environment reference',
                    f'{var_name} contains {matched_identifier}',
                    confidence='Inferred'
                )


def _collect_s3_notification_relations(relations):
    """Adds relationships created by S3 bucket notification configuration."""
    buckets = _safe_aws([], lambda: clients['s3'].list_buckets().get('Buckets', []))
    for bucket in buckets:
        bucket_name = bucket.get('Name')
        notification_config = _safe_aws({}, lambda bucket_name=bucket_name: clients['s3'].get_bucket_notification_configuration(Bucket=bucket_name))
        for config in notification_config.get('LambdaFunctionConfigurations', []) or []:
            lambda_arn = config.get('LambdaFunctionArn', '')
            lambda_name = _resource_name_from_arn(lambda_arn)
            events = ', '.join(config.get('Events', []) or [])
            _add_relation(
                relations, 'S3', f"S3: {bucket_name}",
                'Lambda', f"Lambda: {lambda_name}",
                'Bucket notification', events
            )
        for config in notification_config.get('QueueConfigurations', []) or []:
            queue_arn = config.get('QueueArn', '')
            queue_name = _resource_name_from_arn(queue_arn)
            events = ', '.join(config.get('Events', []) or [])
            _add_relation(
                relations, 'S3', f"S3: {bucket_name}",
                'SQS', f"SQS: {queue_name}",
                'Bucket notification', events
            )
        for config in notification_config.get('TopicConfigurations', []) or []:
            topic_arn = config.get('TopicArn', '')
            topic_name = _resource_name_from_arn(topic_arn)
            events = ', '.join(config.get('Events', []) or [])
            _add_relation(
                relations, 'S3', f"S3: {bucket_name}",
                'SNS', f"SNS: {topic_name}",
                'Bucket notification', events
            )

def collect_discovery_inventory():
    """Collects primary discovered resources for the dashboard inventory."""
    inventory = []

    functions = _safe_aws([], lambda: clients['lambda'].list_functions().get('Functions', []))
    inventory.append(('Lambda', 'Functions', len(functions), _discovery_details(f.get('FunctionName') for f in functions)))

    tables = _safe_aws([], lambda: clients['dynamodb'].list_tables().get('TableNames', []))
    inventory.append(('DynamoDB', 'Tables', len(tables), _discovery_details(tables)))

    apis = _safe_aws([], lambda: clients['apigateway'].get_rest_apis().get('items', []))
    api_details = []
    for api in apis:
        api_id = api.get('id')
        stages = _safe_aws([], lambda api_id=api_id: clients['apigateway'].get_stages(restApiId=api_id).get('item', []))
        resources = _safe_aws([], lambda api_id=api_id: clients['apigateway'].get_resources(restApiId=api_id, embed=['methods']).get('items', []))
        method_count = sum(len(res.get('resourceMethods', {})) for res in resources)
        api_details.append(f"{api.get('name', api_id)} ({len(stages)} stage(s), {method_count} method(s))")
    inventory.append(('API Gateway', 'REST APIs', len(apis), _discovery_details(api_details)))

    queues = _safe_aws([], lambda: clients['sqs'].list_queues().get('QueueUrls', []))
    queue_names = [url.rstrip('/').split('/')[-1] for url in queues]
    inventory.append(('SQS', 'Queues', len(queues), _discovery_details(queue_names)))

    topics = _safe_aws([], lambda: clients['sns'].list_topics().get('Topics', []))
    topic_names = [_resource_name_from_arn(topic.get('TopicArn')) for topic in topics]
    inventory.append(('SNS', 'Topics', len(topics), _discovery_details(topic_names)))

    buckets = _safe_aws([], lambda: clients['s3'].list_buckets().get('Buckets', []))
    bucket_names = [bucket.get('Name') for bucket in buckets]
    inventory.append(('S3', 'Buckets', len(buckets), _discovery_details(bucket_names)))

    rds_clusters = _safe_aws([], lambda: clients['rds'].describe_db_clusters().get('DBClusters', []))
    rds_instances = _safe_aws([], lambda: clients['rds'].describe_db_instances().get('DBInstances', []))
    rds_details = [f"Cluster: {cluster.get('DBClusterIdentifier')}" for cluster in rds_clusters]
    rds_details += [f"Instance: {instance.get('DBInstanceIdentifier')}" for instance in rds_instances]
    inventory.append(('RDS', 'Clusters and instances', len(rds_clusters) + len(rds_instances), _discovery_details(rds_details)))

    pools = _safe_aws([], lambda: clients['cognito'].list_user_pools(MaxResults=60).get('UserPools', []))
    pool_names = [pool.get('Name') for pool in pools]
    inventory.append(('Cognito', 'User pools', len(pools), _discovery_details(pool_names)))

    state_machines = _safe_aws([], lambda: clients['stepfunctions'].list_state_machines().get('stateMachines', []))
    sm_names = [sm.get('name') for sm in state_machines]
    inventory.append(('Step Functions', 'State machines', len(state_machines), _discovery_details(sm_names)))

    web_acls = []

    for scope in ('REGIONAL', 'CLOUDFRONT'):
        web_acls.extend(_safe_aws([], lambda scope=scope: clients['wafv2'].list_web_acls(Scope=scope).get('WebACLs', [])))
    acl_names = [acl.get('Name') for acl in web_acls]
    inventory.append(('WAF', 'Web ACLs', len(web_acls), _discovery_details(acl_names)))

    environments = _safe_aws([], lambda: clients['elasticbeanstalk'].describe_environments().get('Environments', []))
    env_names = [env.get('EnvironmentName') for env in environments]
    inventory.append(('Elastic Beanstalk', 'Environments', len(environments), _discovery_details(env_names)))

    cache_clusters = _safe_aws([], lambda: clients['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True).get('CacheClusters', []))
    cache_names = [cluster.get('CacheClusterId') for cluster in cache_clusters]
    inventory.append(('ElastiCache', 'Clusters', len(cache_clusters), _discovery_details(cache_names)))

    ecr_repos = _safe_aws([], lambda: clients['ecr'].describe_repositories().get('repositories', []))
    repo_names = [repo.get('repositoryName') for repo in ecr_repos]
    inventory.append(('ECR', 'Repositories', len(ecr_repos), _discovery_details(repo_names)))

    code_repos = _safe_aws([], lambda: clients['codecommit'].list_repositories().get('repositories', []))
    code_names = [repo.get('repositoryName') for repo in code_repos]
    inventory.append(('CodeCommit', 'Repositories', len(code_repos), _discovery_details(code_names)))

    return inventory


def collect_architecture_relations():
    """Detects direct service-to-service relationships from AWS configuration."""
    relations = []


    lambda_functions = _safe_aws([], lambda: clients['lambda'].list_functions().get('Functions', []))
    for function in lambda_functions:
        function_name = function.get('FunctionName')
        if function.get('PackageType') == 'Image':
            function_data = _safe_aws({}, lambda function_name=function_name: clients['lambda'].get_function(FunctionName=function_name))
            image_uri = (function_data.get('Code') or {}).get('ImageUri', '')
            repo_name = _ecr_repo_from_image_uri(image_uri)
            if repo_name:
                _add_relation(
                    relations, 'ECR', f"ECR: {repo_name}",
                    'Lambda', f"Lambda: {function_name}",
                    'Container image', image_uri
                )
    api_names = {}
    apis = _safe_aws([], lambda: clients['apigateway'].get_rest_apis().get('items', []))
    for api in apis:
        api_id = api.get('id')
        api_label = f"API Gateway: {api.get('name', api_id)}"
        api_names[api_id] = api_label

        resources = _safe_aws([], lambda api_id=api_id: clients['apigateway'].get_resources(restApiId=api_id, embed=['methods']).get('items', []))
        for resource in resources:
            path = resource.get('path', '/')
            for method, method_def in (resource.get('resourceMethods') or {}).items():
                integration = _safe_aws({}, lambda api_id=api_id, resource=resource, method=method: clients['apigateway'].get_integration(restApiId=api_id, resourceId=resource['id'], httpMethod=method))
                lambda_name = _lambda_name_from_uri(integration.get('uri'))
                if lambda_name:
                    _add_relation(
                        relations, 'API Gateway', api_label,
                        'Lambda', f"Lambda: {lambda_name}",
                        'API integration', f"{method} {path}"
                    )

                authorizer_id = method_def.get('authorizerId')
                if authorizer_id:
                    authorizer = _safe_aws({}, lambda api_id=api_id, authorizer_id=authorizer_id: clients['apigateway'].get_authorizer(restApiId=api_id, authorizerId=authorizer_id))
                    for provider_arn in authorizer.get('providerARNs', []) or []:
                        pool_id = _resource_name_from_arn(provider_arn)
                        pool = _safe_aws({}, lambda pool_id=pool_id: clients['cognito'].describe_user_pool(UserPoolId=pool_id).get('UserPool', {}))
                        pool_name = pool.get('Name') or pool_id
                        _add_relation(
                            relations, 'Cognito', f"Cognito: {pool_name}",
                            'API Gateway', api_label,
                            'Authorizer', authorizer.get('name', authorizer_id)
                        )


    pools = _safe_aws([], lambda: clients['cognito'].list_user_pools(MaxResults=60).get('UserPools', []))
    for pool in pools:
        pool_id = pool.get('Id')
        pool_name = pool.get('Name') or pool_id
        pool_detail = _safe_aws({}, lambda pool_id=pool_id: clients['cognito'].describe_user_pool(UserPoolId=pool_id).get('UserPool', {}))
        lambda_config = pool_detail.get('LambdaConfig') or {}
        for trigger_name, trigger_arn in lambda_config.items():
            if isinstance(trigger_arn, str) and ':lambda:' in trigger_arn and ':function:' in trigger_arn:
                lambda_name = _resource_name_from_arn(trigger_arn)
                _add_relation(
                    relations, 'Cognito', f"Cognito: {pool_name}",
                    'Lambda', f"Lambda: {lambda_name}",
                    'User pool trigger', trigger_name
                )
    event_source_mappings = _safe_aws([], lambda: clients['lambda'].list_event_source_mappings().get('EventSourceMappings', []))
    for mapping in event_source_mappings:
        function_name = _resource_name_from_arn(mapping.get('FunctionArn') or mapping.get('FunctionName'))
        source_arn = mapping.get('EventSourceArn', '')
        source_label = _resource_name_from_arn(source_arn)
        if ':sqs:' in source_arn:
            _add_relation(relations, 'SQS', f"SQS: {source_label}", 'Lambda', f"Lambda: {function_name}", 'Event source mapping', mapping.get('State', ''))
        elif ':dynamodb:' in source_arn:
            table_name = source_arn.split(':table/')[-1].split('/')[0]
            _add_relation(relations, 'DynamoDB', f"DynamoDB: {table_name}", 'Lambda', f"Lambda: {function_name}", 'Stream event source', mapping.get('State', ''))

    topics = _safe_aws([], lambda: clients['sns'].list_topics().get('Topics', []))
    for topic in topics:
        topic_arn = topic.get('TopicArn')
        topic_name = _resource_name_from_arn(topic_arn)
        subs = _safe_aws([], lambda topic_arn=topic_arn: clients['sns'].list_subscriptions_by_topic(TopicArn=topic_arn).get('Subscriptions', []))
        for sub in subs:
            protocol = sub.get('Protocol', '')
            endpoint = sub.get('Endpoint', '')
            if protocol == 'lambda':
                target = f"Lambda: {_resource_name_from_arn(endpoint)}"
                target_service = 'Lambda'
            elif protocol == 'sqs':
                target = f"SQS: {_resource_name_from_arn(endpoint)}"
                target_service = 'SQS'
            else:
                target = f"{protocol}: {endpoint}"
                target_service = protocol or 'Subscription'
            _add_relation(relations, 'SNS', f"SNS: {topic_name}", target_service, target, f"Subscription: {protocol}", sub.get('SubscriptionArn', ''))


    _collect_s3_notification_relations(relations)
    for scope in ('REGIONAL', 'CLOUDFRONT'):
        web_acls = _safe_aws([], lambda scope=scope: clients['wafv2'].list_web_acls(Scope=scope).get('WebACLs', []))
        for acl in web_acls:
            acl_name = acl.get('Name')
            acl_arn = acl.get('ARN')
            associated = _safe_aws([], lambda acl_arn=acl_arn: clients['wafv2'].list_resources_for_web_acl(WebACLArn=acl_arn).get('ResourceArns', []))
            for arn in associated:
                api_id, stage = _api_id_from_stage_arn(arn)
                if api_id and api_id in api_names:
                    target = api_names[api_id]
                    evidence = f"stage {stage}" if stage else arn
                    _add_relation(relations, 'WAF', f"WAF: {acl_name}", 'API Gateway', target, 'Web ACL association', evidence)
                else:
                    _add_relation(relations, 'WAF', f"WAF: {acl_name}", 'AWS resource', arn, 'Web ACL association', arn)

    state_machines = _safe_aws([], lambda: clients['stepfunctions'].list_state_machines().get('stateMachines', []))
    for sm in state_machines:
        sm_name = sm.get('name')
        sm_arn = sm.get('stateMachineArn')
        definition_text = _safe_aws('', lambda sm_arn=sm_arn: clients['stepfunctions'].describe_state_machine(stateMachineArn=sm_arn).get('definition', ''))
        try:
            definition = json.loads(definition_text) if definition_text else {}
        except Exception:
            definition = {}
        for value in _walk_values(definition):
            if isinstance(value, str) and ':lambda:' in value and ':function:' in value:
                lambda_name = _resource_name_from_arn(value)
                _add_relation(relations, 'Step Functions', f"Step Functions: {sm_name}", 'Lambda', f"Lambda: {lambda_name}", 'Task resource', 'state machine definition')


    candidates = _discovery_resource_candidates()

    lambda_functions_for_env = _safe_aws([], lambda: clients['lambda'].list_functions().get('Functions', []))
    for function in lambda_functions_for_env:
        function_name = function.get('FunctionName')
        variables = _environment_variables_from_lambda(function_name)
        _add_inferred_environment_relations(
            relations,
            'Lambda',
            f"Lambda: {function_name}",
            variables,
            candidates
        )

    environments_for_env = _safe_aws([], lambda: clients['elasticbeanstalk'].describe_environments().get('Environments', []))
    for env in environments_for_env:
        env_name = env.get('EnvironmentName')
        app_name = env.get('ApplicationName')
        variables = _elastic_beanstalk_environment_variables(env_name, app_name) if env_name and app_name else {}
        _add_inferred_environment_relations(
            relations,
            'Elastic Beanstalk',
            f"Elastic Beanstalk: {env_name}",
            variables,
            candidates
        )
    return relations
```

```python
# === Discovery renderers: common helpers + overview ===

DISCOVERY_COLORS = {
    'Lambda': '#F58536',
    'API Gateway': '#2E5090',
    'DynamoDB': '#3B6FB6',
    'SQS': '#B48B2A',
    'SNS': '#C65D7B',
    'S3': '#2FA36B',
    'RDS': '#4E79A7',
    'Cognito': '#7B61A8',
    'Step Functions': '#9C755F',
    'WAF': '#C94A44',
    'Elastic Beanstalk': '#59A14F',
    'ElastiCache': '#E15759',
    'ECR': '#76B7B2',
    'CodeCommit': '#6B6B6B',
}

DISCOVERY_MERMAID_STYLES = {
    'Lambda': ('#FFF1E5', '#F58536', '#1F2937'),
    'API Gateway': ('#EAF0FB', '#2E5090', '#1F2937'),
    'DynamoDB': ('#EAF1FC', '#3B6FB6', '#1F2937'),
    'SQS': ('#FFF6D8', '#B48B2A', '#1F2937'),
    'SNS': ('#FBEAF0', '#C65D7B', '#1F2937'),
    'S3': ('#E8F6EE', '#2FA36B', '#1F2937'),
    'RDS': ('#E9F1FA', '#4E79A7', '#1F2937'),
    'Cognito': ('#F0ECFA', '#7B61A8', '#1F2937'),
    'Step Functions': ('#F4EFEA', '#9C755F', '#1F2937'),
    'WAF': ('#FBEAEA', '#C94A44', '#1F2937'),
    'Elastic Beanstalk': ('#EAF6EA', '#59A14F', '#1F2937'),
    'ElastiCache': ('#FCECEC', '#E15759', '#1F2937'),
    'ECR': ('#EAF6F5', '#76B7B2', '#1F2937'),
    'CodeCommit': ('#F0F0F0', '#6B6B6B', '#1F2937'),
    'email': ('#F0F0F0', '#7F8C8D', '#1F2937'),
    'AWS resource': ('#F0F0F0', '#7F8C8D', '#1F2937'),
    'Subscription': ('#F0F0F0', '#7F8C8D', '#1F2937'),
}


def _service_from_node(label):
    return label.split(':', 1)[0]



def discovery_badge(label, level=None):
    label = str(label or 'INFO').upper()
    key = (level or label).upper()
    return html_status(key, '', display=label).replace('</span> ', '</span>')

def _details_html(summary, items):
    clean_items = [html_escape(str(item)) for item in items if str(item).strip()]
    if not clean_items:
        return '-'
    body = ''.join(
        '<div style="padding:4px 0; border-bottom:1px solid #eeeeee; overflow-wrap:anywhere;">'
        f'{item}'
        '</div>'
        for item in clean_items
    )
    return (
        '<details style="cursor:pointer;">'
        f'<summary style="color:#2E5090; font-weight:600;">{html_escape(summary)}</summary>'
        f'<div style="margin-top:8px; color:#333; line-height:1.45;">{body}</div>'
        '</details>'
    )



def _key_value_details_html(summary, pairs):
    clean_pairs = [(str(key), str(value)) for key, value in pairs if str(key).strip()]
    if not clean_pairs:
        return '-'
    rows = ''.join(
        '<div style="display:grid; grid-template-columns:minmax(150px, 35%) 1fr; gap:8px; padding:4px 0; border-bottom:1px solid #eeeeee;">'
        f'<strong style="color:#2E5090; overflow-wrap:anywhere;">{html_escape(key)}</strong>'
        f'<span style="overflow-wrap:anywhere;">{html_escape(value)}</span>'
        '</div>'
        for key, value in clean_pairs
    )
    return (
        '<details style="cursor:pointer;">'
        f'<summary style="color:#2E5090; font-weight:600;">{html_escape(summary)}</summary>'
        f'<div style="margin-top:8px; line-height:1.45;">{rows}</div>'
        '</details>'
    )

def render_inventory_details(details):
    if not details:
        return '-'
    if len(details) == 1:
        return html_escape(str(details[0]))
    return _details_html(f'View {len(details)} resources', details)


def render_relation_evidence(evidence):
    items = [item.strip() for item in str(evidence or '').split(';') if item.strip()]
    if not items:
        return '-'
    if len(items) == 1:
        return html_escape(items[0])
    return _details_html(f'View {len(items)} evidence items', items)


def render_inventory_chart(inventory):
    active = [row for row in inventory if isinstance(row[2], int) and row[2] > 0]
    if not active:
        return html_empty('No discovered resources available for the inventory chart.')

    services = [row[0] for row in active]
    counts = [row[2] for row in active]
    colors = [DISCOVERY_COLORS.get(service, '#7F8C8D') for service in services]

    fig = go.Figure(go.Bar(
        x=counts,
        y=services,
        orientation='h',
        marker_color=colors,
        text=counts,
        textposition='outside',
        hovertemplate='<b>%{y}</b><br>Resources: %{x}<extra></extra>',
    ))
    fig.update_layout(
        title='Detected resources by service',
        height=max(320, 34 * len(services) + 100),
        margin=dict(l=140, r=40, t=55, b=40),
        xaxis_title='Resources',
        yaxis_title='',
        template='plotly_white',
        showlegend=False,
    )
    return fig.to_html(include_plotlyjs='cdn', full_html=False)


def _mermaid_text(text):
    return str(text).replace('"', "'")


def _mermaid_class_name(service):
    clean = re.sub(r'[^A-Za-z0-9_]', '_', str(service or 'resource'))
    return f'svc_{clean}'


def build_mermaid_architecture(relations):
    if not relations:
        return 'graph LR\n  A["No direct relationships detected"]'

    labels = []
    for rel in relations:
        labels.extend([rel['source'], rel['target']])
    labels = list(dict.fromkeys(labels))
    node_ids = {label: f'N{i+1}' for i, label in enumerate(labels)}

    used_services = sorted({_service_from_node(label) for label in labels})
    lines = ['graph LR']
    for service in used_services:
        fill, stroke, color = DISCOVERY_MERMAID_STYLES.get(service, ('#F0F0F0', '#7F8C8D', '#1F2937'))
        lines.append(f'  classDef {_mermaid_class_name(service)} fill:{fill},stroke:{stroke},stroke-width:2px,color:{color};')

    for label in labels:
        lines.append(f'  {node_ids[label]}["{_mermaid_text(label)}"]')
    for rel in relations:
        relation = _mermaid_text(rel['relation'])
        lines.append(f'  {node_ids[rel["source"]]} -->|"{relation}"| {node_ids[rel["target"]]}')
    for label in labels:
        service = _service_from_node(label)
        lines.append(f'  class {node_ids[label]} {_mermaid_class_name(service)};')
    return '\n'.join(lines)


def render_mermaid_diagram(relations, note):
    if not relations:
        return html_empty('No relationships detected for this diagram.')

    source = build_mermaid_architecture(relations)
    diagram_id = f"mermaid-{abs(hash(source))}"
    escaped_source = html_escape(source)
    return (
        html_note(note)
        + f'<div id="{diagram_id}" class="mermaid" style="background:#fff; border:1px solid #e6e6e6; padding:14px; overflow:auto; margin-bottom:16px;">{escaped_source}</div>'
        + '<script>'
        + '(function(){'
        + f'var node = document.getElementById("{diagram_id}");'
        + 'function renderMermaid(){'
        + 'if(!node || !window.mermaid){return;}'
        + 'try {'
        + 'window.mermaid.initialize({startOnLoad:false, securityLevel:"loose", theme:"base", themeVariables:{primaryColor:"#f8f9fb", primaryTextColor:"#1f2937", primaryBorderColor:"#2E5090", lineColor:"#2E5090", fontFamily:"Arial"}});'
        + 'window.mermaid.run({nodes:[node]});'
        + '} catch(e) {'
        + 'node.innerHTML = "<p style=\\"color:#777; font-size:13px;\\">Mermaid could not be rendered in this notebook environment.</p><pre style=\\"white-space:pre-wrap;\\">" + node.textContent + "</pre>";'
        + '}'
        + '}'
        + 'if(window.mermaid){renderMermaid();}'
        + 'else {'
        + 'var script = document.createElement("script");'
        + 'script.src = "https://cdn.jsdelivr.net/npm/mermaid@10/dist/mermaid.min.js";'
        + 'script.onload = renderMermaid;'
        + 'script.onerror = function(){ if(node){ node.innerHTML = "<p style=\\"color:#777; font-size:13px;\\">Mermaid JavaScript could not be loaded. Raw Mermaid definition:</p><pre style=\\"white-space:pre-wrap;\\">" + node.textContent + "</pre>"; } };'
        + 'document.head.appendChild(script);'
        + '}'
        + '})();'
        + '</script>'
    )


def render_architecture_diagrams(relations):
    direct_relations = [rel for rel in relations if rel.get('confidence', 'Direct') == 'Direct']
    inferred_relations = [rel for rel in relations if rel.get('confidence') == 'Inferred']

    html = html_section('Architecture diagrams')
    html += html_subsection('Direct relationships')
    html += render_mermaid_diagram(
        direct_relations,
        'Direct relationships come from explicit AWS configuration.'
    )
    html += html_subsection('Inferred relationships')
    html += render_mermaid_diagram(
        inferred_relations,
        'Inferred relationships are deduced from configuration evidence such as environment variables that reference known AWS resources.'
    )
    return html

def render_discovery_relations_table(relations):
    if not relations:
        return html_empty('No direct relationships detected.')

    rows = []
    for rel in relations:
        confidence = rel.get('confidence', 'Direct')
        rows.append([
            html_escape(rel['source']),
            html_escape(rel['relation']),
            html_escape(rel['target']),
            discovery_badge(str(confidence).upper(), 'INFO' if confidence == 'Inferred' else 'OK'),
            render_relation_evidence(rel.get('evidence', '')),
        ])
    return html_table(['Source', 'Relation', 'Target', 'Confidence', 'Evidence'], rows, widths=['24%', '18%', '24%', '12%', '22%'])
```

```python
# === Discovery renderers: Lambda ===

def render_discovery_overview():
    inventory = collect_discovery_inventory()
    relations = collect_architecture_relations()

    active_services = sum(1 for _, _, count, _ in inventory if isinstance(count, int) and count > 0)
    total_resources = sum(count for _, _, count, _ in inventory if isinstance(count, int))

    html = html_title('Discovery Overview')
    html += html_section('Inventory summary')
    html += html_table(
        ['Metric', 'Value'],
        [
            ['Services with detected resources', active_services],
            ['Total detected resources', total_resources],
            ['Detected relationships', len(relations)],
        ],
        widths=['65%', '35%']
    )
    html += html_note('Total detected resources is the sum of the Count column in the inventory table.')

    html += render_inventory_chart(inventory)

    inventory_rows = [
        [service, resource_type, count, render_inventory_details(details)]
        for service, resource_type, count, details in inventory
    ]
    html += html_section('Resource inventory')
    html += html_table(['Service', 'Resource type', 'Count', 'Details'], inventory_rows, widths=['20%', '24%', '10%', '46%'])

    html += render_architecture_diagrams(relations)

    html += html_section('Detected relationships')
    html += render_discovery_relations_table(relations)

    return html



def _lambda_role_name(role_arn):
    if not role_arn:
        return '-'
    return str(role_arn).split('/')[-1]


def _lambda_vpc_label(function):
    vpc_config = function.get('VpcConfig') or {}
    subnet_count = len(vpc_config.get('SubnetIds') or [])
    sg_count = len(vpc_config.get('SecurityGroupIds') or [])
    if subnet_count or sg_count:
        return f'{subnet_count} subnet(s), {sg_count} security group(s)'
    return 'Not configured'


def _lambda_source_name(source_arn):
    source_arn = str(source_arn or '')
    if ':sqs:' in source_arn:
        return 'SQS', _resource_name_from_arn(source_arn)
    if ':dynamodb:' in source_arn:
        return 'DynamoDB', source_arn.split(':table/')[-1].split('/')[0]
    if ':kinesis:' in source_arn:
        return 'Kinesis', _resource_name_from_arn(source_arn)
    return 'AWS resource', _resource_name_from_arn(source_arn) or source_arn


def render_lambda_runtime_chart(functions):
    counts = {}
    for function in functions:
        runtime = function.get('Runtime') or function.get('PackageType') or 'Unknown'
        counts[runtime] = counts.get(runtime, 0) + 1
    if not counts:
        return html_empty('No Lambda runtimes detected.')

    labels = list(counts.keys())
    values = [counts[label] for label in labels]
    fig = go.Figure(go.Bar(
        x=labels,
        y=values,
        marker_color='#F58536',
        text=values,
        textposition='outside',
        hovertemplate='<b>%{x}</b><br>Functions: %{y}<extra></extra>',
    ))
    fig.update_layout(
        title='Runtime distribution',
        height=320,
        margin=dict(l=50, r=30, t=55, b=80),
        xaxis_title='Runtime',
        yaxis_title='Functions',
        template='plotly_white',
        showlegend=False,
    )
    return fig.to_html(include_plotlyjs='cdn', full_html=False)


def render_lambda_discovery_summary(functions, mappings, relations):
    runtime_count = len({function.get('Runtime') for function in functions if function.get('Runtime')})
    with_vpc = sum(1 for function in functions if (function.get('VpcConfig') or {}).get('SubnetIds'))
    with_env = 0
    for function in functions:
        variables = _environment_variables_from_lambda(function.get('FunctionName'))
        if variables:
            with_env += 1
    lambda_relations = [rel for rel in relations if rel.get('source_service') == 'Lambda' or rel.get('target_service') == 'Lambda']
    rows = [
        ['Functions detected', len(functions)],
        ['Runtime families', runtime_count],
        ['Functions with VPC config', with_vpc],
        ['Functions with environment variables', with_env],
        ['Event source mappings', len(mappings)],
        ['Lambda relationships detected', len(lambda_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_lambda_functions_inventory(functions):
    if not functions:
        return html_empty('No Lambda functions detected.')
    rows = []
    for function in sorted(functions, key=lambda item: item.get('FunctionName', '')):
        rows.append([
            html_escape(function.get('FunctionName', '-')),
            html_escape(function.get('Runtime') or function.get('PackageType') or '-'),
            html_escape(function.get('Handler') or '-'),
            html_escape(function.get('PackageType', 'Zip')),
            function.get('MemorySize', '-'),
            function.get('Timeout', '-'),
            round((function.get('CodeSize') or 0) / 1024, 1),
            html_escape(_lambda_role_name(function.get('Role'))),
            html_escape(_lambda_vpc_label(function)),
            html_escape(str(function.get('LastModified', '-'))[:10]),
        ])
    return html_table(
        ['Function', 'Runtime', 'Handler', 'Package', 'Memory MB', 'Timeout s', 'Code KB', 'Role', 'VPC', 'Last modified'],
        rows,
        widths=['15%', '10%', '13%', '8%', '8%', '8%', '8%', '10%', '10%', '10%']
    )


def render_lambda_event_source_mappings(mappings):
    if not mappings:
        return html_empty('No Lambda event source mappings detected.')
    rows = []
    for mapping in sorted(mappings, key=lambda item: item.get('FunctionArn') or item.get('FunctionName') or ''):
        source_service, source_name = _lambda_source_name(mapping.get('EventSourceArn'))
        function_name = _resource_name_from_arn(mapping.get('FunctionArn') or mapping.get('FunctionName'))
        rows.append([
            html_escape(source_service),
            html_escape(source_name),
            html_escape(function_name),
            discovery_badge(mapping.get('State', '-'), 'OK' if mapping.get('State') == 'Enabled' else 'INFO'),
            mapping.get('BatchSize', '-'),
            html_escape(mapping.get('StartingPosition', '-') or '-'),
        ])
    return html_table(['Source service', 'Source', 'Target function', 'State', 'Batch size', 'Starting position'], rows, widths=['16%', '24%', '24%', '14%', '10%', '12%'])


def render_lambda_environment_references(functions):
    candidates = _discovery_resource_candidates()
    rows = []
    for function in sorted(functions, key=lambda item: item.get('FunctionName', '')):
        function_name = function.get('FunctionName')
        variables = _environment_variables_from_lambda(function_name)
        variable_pairs = sorted(variables.items())
        references = []
        for var_name, var_value in variables.items():
            for target_service, target_candidates in candidates.items():
                for resource_name, matched_identifier in _value_mentions(var_value, target_candidates):
                    references.append(f'{var_name}: {target_service} {resource_name}')
        if variable_pairs or references:
            rows.append([
                html_escape(function_name),
                len(variable_pairs),
                _key_value_details_html(f'View {len(variable_pairs)} variables', variable_pairs) if variable_pairs else '-',
                render_inventory_details(sorted(set(references))) if references else html_empty('No known resource references detected.'),
            ])
    if not rows:
        return html_empty('No Lambda environment variables detected.')
    return html_table(['Function', 'Variable count', 'Environment variables', 'Detected references'], rows, widths=['24%', '12%', '32%', '32%'])


def render_lambda_async_config(functions):
    rows = []
    for function in sorted(functions, key=lambda item: item.get('FunctionName', '')):
        function_name = function.get('FunctionName')
        config = _safe_aws(None, lambda function_name=function_name: clients['lambda'].get_function_event_invoke_config(FunctionName=function_name))
        if not config:
            continue
        destination = config.get('DestinationConfig') or {}
        on_success = (destination.get('OnSuccess') or {}).get('Destination', '-')
        on_failure = (destination.get('OnFailure') or {}).get('Destination', '-')
        rows.append([
            html_escape(function_name),
            config.get('MaximumRetryAttempts', 'default'),
            config.get('MaximumEventAgeInSeconds', 'default'),
            html_escape(on_success),
            html_escape(on_failure),
        ])
    if not rows:
        return html_empty('No custom asynchronous invocation configuration detected.')
    return html_table(['Function', 'Max retries', 'Max event age', 'On success', 'On failure'], rows, widths=['25%', '12%', '14%', '24%', '25%'])


def render_lambda_tags(functions):
    rows = []
    for function in sorted(functions, key=lambda item: item.get('FunctionName', '')):
        function_name = function.get('FunctionName')
        arn = function.get('FunctionArn')
        tags = _safe_aws({}, lambda arn=arn: clients['lambda'].list_tags(Resource=arn).get('Tags', {})) if arn else {}
        if tags:
            rows.append([
                html_escape(function_name),
                len(tags),
                _key_value_details_html(f'View {len(tags)} tags', sorted(tags.items()))
            ])
    if not rows:
        return html_empty('No Lambda tags detected.')
    return html_table(['Function', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])



def render_lambda_architecture_diagrams(relations):
    incoming = [
        rel for rel in relations
        if rel.get('target_service') == 'Lambda' and rel.get('confidence', 'Direct') == 'Direct'
    ]
    outgoing = [
        rel for rel in relations
        if rel.get('source_service') == 'Lambda' and rel.get('confidence') == 'Inferred'
    ]

    html = html_section('Lambda architecture diagrams')
    html += html_subsection('Incoming triggers and integrations')
    html += render_mermaid_diagram(
        incoming,
        'Direct relationships where Lambda is the target, such as API Gateway integrations, SQS event source mappings, SNS subscriptions, S3 notifications or Step Functions tasks.'
    )
    html += html_subsection('Outgoing inferred dependencies')
    html += render_mermaid_diagram(
        outgoing,
        'Inferred relationships where Lambda references known AWS resources through environment variables.'
    )
    return html

def render_lambda_relationships(relations):
    lambda_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'Lambda' or rel.get('target_service') == 'Lambda'
    ]
    if not lambda_relations:
        return html_empty('No Lambda relationships detected.')
    return render_discovery_relations_table(lambda_relations)


def render_discovery_lambda():
    functions = _safe_aws([], lambda: clients['lambda'].list_functions().get('Functions', []))
    mappings = _safe_aws([], lambda: clients['lambda'].list_event_source_mappings().get('EventSourceMappings', []))
    relations = collect_architecture_relations()

    html = html_title('Lambda - Discovery')
    html += html_section('Summary')
    html += render_lambda_discovery_summary(functions, mappings, relations)
    html += render_lambda_runtime_chart(functions)

    html += render_lambda_architecture_diagrams(relations)

    html += html_section('Functions inventory')
    html += render_lambda_functions_inventory(functions)

    html += html_section('Event source mappings')
    html += render_lambda_event_source_mappings(mappings)

    html += html_section('Environment references')
    html += render_lambda_environment_references(functions)

    html += html_section('Asynchronous invocation configuration')
    html += render_lambda_async_config(functions)

    html += html_section('Tags')
    html += render_lambda_tags(functions)

    html += html_section('Lambda relationships')
    html += render_lambda_relationships(relations)
    return html
```

```python
# === Discovery renderers: DynamoDB ===

def _dynamodb_key_schema_label(key_schema):
    parts = []
    for key in key_schema or []:
        key_type = 'PK' if key.get('KeyType') == 'HASH' else 'SK'
        parts.append(f"{key.get('AttributeName')} ({key_type})")
    return ' | '.join(parts) if parts else '-'


def _dynamodb_projection_label(projection):
    projection = projection or {}
    projection_type = projection.get('ProjectionType', '-')
    non_key = projection.get('NonKeyAttributes') or []
    if non_key:
        return f"{projection_type}: {', '.join(non_key)}"
    return projection_type


def _dynamodb_describe_tables():
    table_names = _safe_aws([], lambda: clients['dynamodb'].list_tables().get('TableNames', []))
    tables = []
    for table_name in table_names:
        table = _safe_aws(None, lambda table_name=table_name: clients['dynamodb'].describe_table(TableName=table_name).get('Table'))
        if table:
            tables.append(table)
    return tables


def _dynamodb_scalar(value):
    if isinstance(value, dict):
        if len(value) == 1:
            data_type, data_value = next(iter(value.items()))
            if data_type in ('S', 'N', 'BOOL', 'NULL'):
                return data_value
            if data_type in ('SS', 'NS', 'BS'):
                return ', '.join(str(item) for item in data_value)
        return json.dumps(value, ensure_ascii=False)
    return value


def _dynamodb_item_details(items):
    if not items:
        return '-'
    rendered = []
    for idx, item in enumerate(items, start=1):
        pairs = []
        for key, value in sorted(item.items()):
            pairs.append((key, _dynamodb_scalar(value)))
        rendered.append(
            '<div style="margin:8px 0 10px 0; padding:8px; border:1px solid #eeeeee; background:#fafafa;">'
            f'<div style="font-weight:700; color:#2E5090; margin-bottom:6px;">Item {idx}</div>'
            + _key_value_details_html('View attributes', pairs)
            + '</div>'
        )
    return '<div>' + ''.join(rendered) + '</div>'


def render_dynamodb_summary(tables):
    active = sum(1 for table in tables if table.get('TableStatus') == 'ACTIVE')
    total_items = sum(table.get('ItemCount') or 0 for table in tables)
    total_size = sum(table.get('TableSizeBytes') or 0 for table in tables)
    with_indexes = sum(1 for table in tables if table.get('GlobalSecondaryIndexes') or table.get('LocalSecondaryIndexes'))

    ttl_enabled = 0
    pitr_enabled = 0
    for table in tables:
        table_name = table.get('TableName')
        ttl = _safe_aws({}, lambda table_name=table_name: clients['dynamodb'].describe_time_to_live(TableName=table_name).get('TimeToLiveDescription', {}))
        if ttl.get('TimeToLiveStatus') == 'ENABLED':
            ttl_enabled += 1
        backups = _safe_aws({}, lambda table_name=table_name: clients['dynamodb'].describe_continuous_backups(TableName=table_name).get('ContinuousBackupsDescription', {}))
        pitr = (backups.get('PointInTimeRecoveryDescription') or {}).get('PointInTimeRecoveryStatus')
        if pitr == 'ENABLED':
            pitr_enabled += 1

    rows = [
        ['Tables detected', len(tables)],
        ['Active tables', active],
        ['Total approx items (AWS metadata)', f'{total_items:,}'],
        ['Total size (AWS metadata)', f'{total_size:,} bytes'],
        ['Tables with indexes', with_indexes],
        ['Tables with TTL enabled', ttl_enabled],
        ['Tables with PITR enabled', pitr_enabled],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_dynamodb_distribution_chart(tables):
    if not tables:
        return html_empty('No DynamoDB tables detected for charts.')
    names = [table.get('TableName') for table in tables]
    items = [table.get('ItemCount') or 0 for table in tables]
    sizes_kb = [round((table.get('TableSizeBytes') or 0) / 1024, 2) for table in tables]

    fig = make_subplots(rows=1, cols=2, subplot_titles=('Approx items by table', 'Table size by table'))
    fig.add_trace(go.Bar(x=names, y=items, marker_color='#3B6FB6', name='Items'), row=1, col=1)
    fig.add_trace(go.Bar(x=names, y=sizes_kb, marker_color='#2FA36B', name='Size KB'), row=1, col=2)
    fig.update_layout(
        height=360,
        margin=dict(l=45, r=30, t=65, b=95),
        template='plotly_white',
        showlegend=False,
    )
    fig.update_xaxes(tickangle=-35)
    fig.update_yaxes(title_text='Items', row=1, col=1)
    fig.update_yaxes(title_text='KB', row=1, col=2)
    return fig.to_html(include_plotlyjs='cdn', full_html=False)


def render_dynamodb_relationship_diagram(relations):
    dynamodb_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'DynamoDB' or rel.get('target_service') == 'DynamoDB'
    ]
    html = html_section('DynamoDB architecture diagram')
    html += render_mermaid_diagram(
        dynamodb_relations,
        'Relationships where DynamoDB appears as source or target. Direct links usually come from streams; inferred links come from environment references.'
    )
    return html



def _dynamodb_billing_label(billing):
    if billing == 'PAY_PER_REQUEST':
        return 'On-demand'
    if billing == 'PROVISIONED':
        return 'Provisioned'
    return billing or '-'


def _compact_metadata_lines(pairs):
    return ''.join(
        '<div style="white-space:nowrap; line-height:1.55;">'
        f'<strong style="color:#2E5090;">{html_escape(str(key))}:</strong> {html_escape(str(value))}'
        '</div>'
        for key, value in pairs
    )

def render_dynamodb_tables_inventory(tables):
    if not tables:
        return html_empty('No DynamoDB tables detected.')
    rows = []
    for table in sorted(tables, key=lambda item: item.get('TableName', '')):
        gsis = table.get('GlobalSecondaryIndexes') or []
        lsis = table.get('LocalSecondaryIndexes') or []
        billing = (table.get('BillingModeSummary') or {}).get('BillingMode', 'PROVISIONED')
        metadata = [
            ('Created', str(table.get('CreationDateTime', '-'))[:10]),
            ('Indexes', len(gsis) + len(lsis)),
            ('Approx items', f"{table.get('ItemCount') or 0:,}"),
            ('Size bytes', f"{table.get('TableSizeBytes') or 0:,}"),
        ]
        rows.append([
            html_escape(table.get('TableName', '-')),
            discovery_badge(table.get('TableStatus', 'INFO')),
            html_escape(_dynamodb_key_schema_label(table.get('KeySchema'))),
            html_escape(_dynamodb_billing_label(billing)),
            _compact_metadata_lines(metadata),
        ])
    return html_table(['Table', 'Status', 'Keys', 'Billing', 'Metadata'], rows, widths=['24%', '12%', '22%', '14%', '28%'])


def render_dynamodb_secondary_indexes(tables):
    rows = []
    for table in sorted(tables, key=lambda item: item.get('TableName', '')):
        table_name = table.get('TableName')
        for index in table.get('GlobalSecondaryIndexes') or []:
            rows.append([
                html_escape(table_name),
                'GSI',
                html_escape(index.get('IndexName', '-')),
                html_escape(_dynamodb_key_schema_label(index.get('KeySchema'))),
                discovery_badge(index.get('IndexStatus', '-'), 'OK' if index.get('IndexStatus') == 'ACTIVE' else 'INFO'),
                html_escape(_dynamodb_projection_label(index.get('Projection'))),
            ])
        for index in table.get('LocalSecondaryIndexes') or []:
            rows.append([
                html_escape(table_name),
                'LSI',
                html_escape(index.get('IndexName', '-')),
                html_escape(_dynamodb_key_schema_label(index.get('KeySchema'))),
                discovery_badge('Local', 'OK'),
                html_escape(_dynamodb_projection_label(index.get('Projection'))),
            ])
    if not rows:
        return html_empty('No secondary indexes detected.')
    return html_table(['Table', 'Type', 'Index', 'Keys', 'Status', 'Projection'], rows, widths=['20%', '8%', '20%', '22%', '12%', '18%'])


def render_dynamodb_table_configuration(tables):
    rows = []
    for table in sorted(tables, key=lambda item: item.get('TableName', '')):
        table_name = table.get('TableName')
        ttl = _safe_aws({}, lambda table_name=table_name: clients['dynamodb'].describe_time_to_live(TableName=table_name).get('TimeToLiveDescription', {}))
        ttl_status = ttl.get('TimeToLiveStatus', 'DISABLED')
        ttl_attr = ttl.get('AttributeName', '-') if ttl_status == 'ENABLED' else '-'
        backups = _safe_aws({}, lambda table_name=table_name: clients['dynamodb'].describe_continuous_backups(TableName=table_name).get('ContinuousBackupsDescription', {}))
        pitr = (backups.get('PointInTimeRecoveryDescription') or {}).get('PointInTimeRecoveryStatus', 'DISABLED')
        rows.append([
            html_escape(table_name),
            discovery_badge(ttl_status, 'OK' if ttl_status == 'ENABLED' else 'INFO'),
            html_escape(ttl_attr),
            discovery_badge(pitr, 'OK' if pitr == 'ENABLED' else 'WARN'),
        ])
    if not rows:
        return html_empty('No DynamoDB table configuration available.')
    return html_table(['Table', 'TTL status', 'TTL attribute', 'Point-in-time recovery'], rows, widths=['34%', '18%', '24%', '24%'])


def render_dynamodb_item_sample(tables, limit=5):
    rows = []
    for table in sorted(tables, key=lambda item: item.get('TableName', '')):
        table_name = table.get('TableName')
        response = _safe_aws({}, lambda table_name=table_name, limit=limit: clients['dynamodb'].scan(TableName=table_name, Limit=limit))
        items = response.get('Items', []) if response else []
        rows.append([
            html_escape(table_name),
            len(items),
            _dynamodb_item_details(items) if items else html_empty('Table empty or no sample returned.'),
        ])
    if not rows:
        return html_empty('No DynamoDB sample data available.')
    return html_table(['Table', 'Items shown', 'Sample'], rows, widths=['24%', '12%', '64%'])


def render_dynamodb_tags(tables):
    rows = []
    for table in sorted(tables, key=lambda item: item.get('TableName', '')):
        table_name = table.get('TableName')
        arn = table.get('TableArn')
        tags = _safe_aws([], lambda arn=arn: clients['dynamodb'].list_tags_of_resource(ResourceArn=arn).get('Tags', [])) if arn else []
        if tags:
            tag_pairs = sorted((tag.get('Key'), tag.get('Value')) for tag in tags)
            rows.append([html_escape(table_name), len(tags), _key_value_details_html(f'View {len(tags)} tags', tag_pairs)])
    if not rows:
        return html_empty('No DynamoDB tags detected.')
    return html_table(['Table', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_dynamodb_relationships(relations):
    dynamodb_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'DynamoDB' or rel.get('target_service') == 'DynamoDB'
    ]
    if not dynamodb_relations:
        return html_empty('No DynamoDB relationships detected.')
    return render_discovery_relations_table(dynamodb_relations)


def render_discovery_dynamodb():
    tables = _dynamodb_describe_tables()
    relations = collect_architecture_relations()

    html = html_title('DynamoDB - Discovery')
    html += html_section('Summary')
    html += render_dynamodb_summary(tables)
    html += render_dynamodb_distribution_chart(tables)

    html += render_dynamodb_relationship_diagram(relations)

    html += html_section('Tables inventory')
    html += html_note('Approx item count and table size come from DynamoDB table metadata. AWS updates these values periodically, so recently modified tables can show 0 even when the limited sample below returns items.')
    html += render_dynamodb_tables_inventory(tables)

    html += html_section('Secondary indexes')
    html += render_dynamodb_secondary_indexes(tables)

    html += html_section('Table configuration')
    html += render_dynamodb_table_configuration(tables)

    html += html_section('Item sample')
    html += html_note('This is a limited scan sample for structural inspection, not a complete data inventory.')
    html += render_dynamodb_item_sample(tables, limit=5)

    html += html_section('Tags')
    html += render_dynamodb_tags(tables)

    html += html_section('DynamoDB relationships')
    html += render_dynamodb_relationships(relations)
    return html
```

```python
# === Discovery renderers: SQS ===

def _sqs_queue_name(url):
    return str(url or '').rstrip('/').split('/')[-1]


def _sqs_int(attrs, key, default=0):
    try:
        return int((attrs or {}).get(key, default))
    except Exception:
        return default


def _sqs_created_date(attrs):
    created_ts = (attrs or {}).get('CreatedTimestamp')
    try:
        return datetime.fromtimestamp(int(created_ts)).strftime('%Y-%m-%d')
    except Exception:
        return 'N/A'


def _sqs_json(value, fallback=None):
    if not value:
        return fallback
    try:
        return json.loads(value)
    except Exception:
        return fallback


def _sqs_encryption_label(attrs):
    attrs = attrs or {}
    if str(attrs.get('SqsManagedSseEnabled', '')).lower() == 'true':
        return 'SSE-SQS'
    if attrs.get('KmsMasterKeyId'):
        return 'SSE-KMS'
    return 'NONE'


def _sqs_policy_statements(policy_str):
    policy = _sqs_json(policy_str, {}) or {}
    statements = policy.get('Statement', [])
    if isinstance(statements, dict):
        statements = [statements]
    return statements if isinstance(statements, list) else []


def _sqs_statement_summary(statement):
    effect = statement.get('Effect', 'N/A')
    principal = statement.get('Principal', 'N/A')
    action = statement.get('Action', 'N/A')
    condition = 'yes' if statement.get('Condition') else 'no'
    if isinstance(action, list):
        visible = ', '.join(str(item) for item in action[:3])
        if len(action) > 3:
            visible += f' (+{len(action) - 3} more)'
        action = visible
    return f'Effect: {effect} | Principal: {principal} | Action: {action} | Condition: {condition}'


def _sqs_collect_details():
    urls = _safe_aws([], lambda: clients['sqs'].list_queues().get('QueueUrls', []))
    queues = []
    for url in urls:
        name = _sqs_queue_name(url)
        attrs = _safe_aws({}, lambda url=url: clients['sqs'].get_queue_attributes(QueueUrl=url, AttributeNames=['All']).get('Attributes', {}))
        redrive = _sqs_json(attrs.get('RedrivePolicy'), {}) or {}
        dlq_arn = redrive.get('deadLetterTargetArn', '')
        dlq_name = _resource_name_from_arn(dlq_arn) if dlq_arn else ''
        incoming_sources = _safe_aws([], lambda url=url: clients['sqs'].list_dead_letter_source_queues(QueueUrl=url).get('queueUrls', []))
        incoming_source_names = [_sqs_queue_name(source_url) for source_url in incoming_sources]
        tags = _safe_aws({}, lambda url=url: clients['sqs'].list_queue_tags(QueueUrl=url).get('Tags', {}))
        policy_str = attrs.get('Policy', '')
        statements = _sqs_policy_statements(policy_str)
        queues.append({
            'name': name,
            'url': url,
            'arn': attrs.get('QueueArn', ''),
            'attributes': attrs,
            'type': 'FIFO' if attrs.get('FifoQueue') == 'true' else 'Standard',
            'created': _sqs_created_date(attrs),
            'available': _sqs_int(attrs, 'ApproximateNumberOfMessages'),
            'in_flight': _sqs_int(attrs, 'ApproximateNumberOfMessagesNotVisible'),
            'delayed': _sqs_int(attrs, 'ApproximateNumberOfMessagesDelayed'),
            'retention': attrs.get('MessageRetentionPeriod', 'N/A'),
            'visibility': attrs.get('VisibilityTimeout', 'N/A'),
            'delay': attrs.get('DelaySeconds', '0'),
            'max_size': attrs.get('MaximumMessageSize', 'N/A'),
            'encryption': _sqs_encryption_label(attrs),
            'dlq_name': dlq_name,
            'dlq_arn': dlq_arn,
            'max_receive_count': redrive.get('maxReceiveCount', '-'),
            'incoming_sources': incoming_source_names,
            'is_dlq': bool(incoming_source_names) or 'dlq' in name.lower(),
            'policy': policy_str,
            'policy_statements': statements,
            'tags': tags,
        })
    return queues


def _sqs_dlq_relations_from_details(queues):
    relations = []
    for queue in queues:
        if queue.get('dlq_name'):
            _add_relation(
                relations,
                'SQS', f"SQS: {queue['name']}",
                'SQS', f"SQS: {queue['dlq_name']}",
                'Dead-letter queue',
                f"maxReceiveCount={queue.get('max_receive_count', '-')}",
            )
    return relations


def render_sqs_summary(queues, relations):
    standard = sum(1 for queue in queues if queue.get('type') == 'Standard')
    fifo = sum(1 for queue in queues if queue.get('type') == 'FIFO')
    with_dlq = sum(1 for queue in queues if queue.get('dlq_name'))
    dlqs = sum(1 for queue in queues if queue.get('is_dlq'))
    encrypted = sum(1 for queue in queues if queue.get('encryption') != 'NONE')
    sqs_relations = [rel for rel in relations if rel.get('source_service') == 'SQS' or rel.get('target_service') == 'SQS']
    rows = [
        ['Queues detected', len(queues)],
        ['Standard queues', standard],
        ['FIFO queues', fifo],
        ['Queues with DLQ configured', with_dlq],
        ['Dead-letter queues detected', dlqs],
        ['Encrypted queues', encrypted],
        ['SQS relationships detected', len(sqs_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_sqs_relationship_diagram(relations):
    sqs_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'SQS' or rel.get('target_service') == 'SQS'
    ]
    html = html_section('SQS architecture diagram')
    html += render_mermaid_diagram(
        sqs_relations,
        'Relationships where SQS appears as source or target, including Lambda event source mappings, SNS subscriptions, S3 notifications and inferred environment references.'
    )
    return html


def render_sqs_dlq_diagram(queues):
    dlq_relations = _sqs_dlq_relations_from_details(queues)
    html = html_section('Dead-letter queue diagram')
    html += render_mermaid_diagram(
        dlq_relations,
        'Redrive relationships from source queues to their configured dead-letter queues.'
    )
    return html


def render_sqs_inventory(queues):
    if not queues:
        return html_empty('No SQS queues detected.')
    rows = []
    for queue in sorted(queues, key=lambda item: item.get('name', '')):
        message_details = _compact_metadata_lines([
            ('Available', queue.get('available', 0)),
            ('In flight', queue.get('in_flight', 0)),
            ('Delayed', queue.get('delayed', 0)),
        ])
        max_size = f"{int(queue.get('max_size', 0)):,} bytes" if str(queue.get('max_size', '')).isdigit() else queue.get('max_size', 'N/A')
        config_details = _compact_metadata_lines([
            ('Retention', f"{queue.get('retention', 'N/A')} s"),
            ('Visibility', f"{queue.get('visibility', 'N/A')} s"),
            ('Delay', f"{queue.get('delay', '0')} s"),
            ('Max size', max_size),
        ])
        encryption = queue.get('encryption', 'NONE')
        rows.append([
            html_escape(queue.get('name', '-')),
            discovery_badge(queue.get('type', 'Standard'), 'OK'),
            html_escape(queue.get('created', 'N/A')),
            message_details,
            config_details,
            discovery_badge(encryption, 'OK' if encryption != 'NONE' else 'WARN'),
            html_escape(queue.get('dlq_name') or 'None'),
        ])
    return html_table(['Queue', 'Type', 'Created', 'Messages', 'Settings', 'Encryption', 'DLQ'], rows, widths=['22%', '11%', '11%', '15%', '19%', '11%', '11%'])


def render_sqs_identifiers(queues):
    if not queues:
        return html_empty('No SQS queue identifiers available.')
    rows = []
    for queue in sorted(queues, key=lambda item: item.get('name', '')):
        rows.append([
            html_escape(queue.get('name', '-')),
            _details_html('View identifiers', [
                f"ARN: {queue.get('arn', '-')}",
                f"URL: {queue.get('url', '-')}",
            ]),
        ])
    return html_table(['Queue', 'Identifiers'], rows, widths=['30%', '70%'])


def render_sqs_dlq_relationships(queues):
    rows = []
    for queue in sorted(queues, key=lambda item: item.get('name', '')):
        if queue.get('dlq_name'):
            rows.append([
                html_escape(queue.get('name', '-')),
                html_escape(queue.get('dlq_name')),
                html_escape(queue.get('max_receive_count', '-')),
                discovery_badge('Configured', 'OK'),
            ])
    for queue in sorted(queues, key=lambda item: item.get('name', '')):
        if queue.get('is_dlq'):
            rows.append([
                html_escape(', '.join(queue.get('incoming_sources') or []) or 'No source queues detected'),
                html_escape(queue.get('name', '-')),
                '-',
                discovery_badge('DLQ', 'INFO'),
            ])
    if not rows:
        return html_empty('No SQS dead-letter queue relationships detected.')
    return html_table(['Source queue', 'Dead-letter queue', 'Max receive count', 'Role'], rows, widths=['34%', '34%', '16%', '16%'])


def render_sqs_lambda_mappings():
    mappings = _safe_aws([], lambda: clients['lambda'].list_event_source_mappings().get('EventSourceMappings', []))
    rows = []
    for mapping in sorted(mappings, key=lambda item: item.get('FunctionArn') or item.get('FunctionName') or ''):
        source_arn = str(mapping.get('EventSourceArn') or '')
        if ':sqs:' not in source_arn:
            continue
        function_name = _resource_name_from_arn(mapping.get('FunctionArn') or mapping.get('FunctionName'))
        rows.append([
            html_escape(_resource_name_from_arn(source_arn)),
            html_escape(function_name),
            discovery_badge(mapping.get('State', '-'), 'OK' if mapping.get('State') == 'Enabled' else 'INFO'),
            mapping.get('BatchSize', '-'),
            html_escape(mapping.get('UUID', '-')),
        ])
    if not rows:
        return html_empty('No SQS to Lambda event source mappings detected.')
    return html_table(['Queue', 'Target Lambda', 'State', 'Batch size', 'Mapping UUID'], rows, widths=['24%', '24%', '12%', '12%', '28%'])


def render_sqs_policies(queues):
    if not queues:
        return html_empty('No SQS queues detected.')
    rows = []
    for queue in sorted(queues, key=lambda item: item.get('name', '')):
        statements = queue.get('policy_statements') or []
        if statements:
            details = _details_html(f'View {len(statements)} statements', [_sqs_statement_summary(stmt) for stmt in statements])
            status = discovery_badge(f'{len(statements)} statements', 'INFO')
        else:
            details = 'Access managed by IAM; no queue policy configured.'
            status = discovery_badge('No policy', 'INFO')
        rows.append([
            html_escape(queue.get('name', '-')),
            status,
            details,
        ])
    return html_table(['Queue', 'Policy', 'Details'], rows, widths=['30%', '18%', '52%'])


def render_sqs_tags(queues):
    rows = []
    for queue in sorted(queues, key=lambda item: item.get('name', '')):
        tags = queue.get('tags') or {}
        if tags:
            rows.append([
                html_escape(queue.get('name', '-')),
                len(tags),
                _key_value_details_html(f'View {len(tags)} tags', sorted(tags.items())),
            ])
    if not rows:
        return html_empty('No SQS tags detected.')
    return html_table(['Queue', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_sqs_relationships(relations):
    sqs_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'SQS' or rel.get('target_service') == 'SQS'
    ]
    if not sqs_relations:
        return html_empty('No SQS relationships detected.')
    return render_discovery_relations_table(sqs_relations)


def render_discovery_sqs():
    queues = _sqs_collect_details()
    relations = collect_architecture_relations()

    html = html_title('SQS - Discovery')
    html += html_section('Summary')
    html += render_sqs_summary(queues, relations)

    html += render_sqs_relationship_diagram(relations)
    html += render_sqs_dlq_diagram(queues)

    html += html_section('Queues inventory')
    html += render_sqs_inventory(queues)

    html += html_section('Queue identifiers')
    html += render_sqs_identifiers(queues)

    html += html_section('Dead-letter queue relationships')
    html += render_sqs_dlq_relationships(queues)

    html += html_section('Lambda event source mappings')
    html += render_sqs_lambda_mappings()

    html += html_section('Queue policies')
    html += render_sqs_policies(queues)

    html += html_section('Tags')
    html += render_sqs_tags(queues)

    html += html_section('SQS relationships')
    html += render_sqs_relationships(relations)
    return html
```

```python
# === Discovery renderers: SNS ===

def _sns_topic_name(arn):
    return _resource_name_from_arn(arn)


def _sns_int(attrs, key, default=0):
    try:
        return int((attrs or {}).get(key, default))
    except Exception:
        return default


def _sns_json(value, fallback=None):
    if not value:
        return fallback
    try:
        return json.loads(value)
    except Exception:
        return fallback


def _sns_policy_statements(policy_str):
    policy = _sns_json(policy_str, {}) or {}
    statements = policy.get('Statement', [])
    if isinstance(statements, dict):
        statements = [statements]
    return statements if isinstance(statements, list) else []


def _sns_statement_summary(statement):
    effect = statement.get('Effect', 'N/A')
    principal = statement.get('Principal', 'N/A')
    action = statement.get('Action', 'N/A')
    condition = 'yes' if statement.get('Condition') else 'no'
    if isinstance(action, list):
        visible = ', '.join(str(item) for item in action[:3])
        if len(action) > 3:
            visible += f' (+{len(action) - 3} more)'
        action = visible
    return f'Effect: {effect} | Principal: {principal} | Action: {action} | Condition: {condition}'


def _sns_collect_details():
    topics = _safe_aws([], lambda: clients['sns'].list_topics().get('Topics', []))
    details = []
    for topic in topics:
        arn = topic.get('TopicArn', '')
        name = _sns_topic_name(arn)
        attrs = _safe_aws({}, lambda arn=arn: clients['sns'].get_topic_attributes(TopicArn=arn).get('Attributes', {}))
        access_error = not bool(attrs)
        subscriptions = []
        if not access_error:
            raw_subs = _safe_aws([], lambda arn=arn: clients['sns'].list_subscriptions_by_topic(TopicArn=arn).get('Subscriptions', []))
            for sub in raw_subs:
                sub_arn = sub.get('SubscriptionArn', '')
                sub_attrs = {}
                if sub_arn and sub_arn != 'PendingConfirmation':
                    sub_attrs = _safe_aws({}, lambda sub_arn=sub_arn: clients['sns'].get_subscription_attributes(SubscriptionArn=sub_arn).get('Attributes', {}))
                redrive = _sns_json(sub_attrs.get('RedrivePolicy'), {}) or {}
                filter_policy = sub_attrs.get('FilterPolicy', 'none')
                subscriptions.append({
                    'protocol': sub.get('Protocol', '-'),
                    'endpoint': sub.get('Endpoint', '-'),
                    'arn': sub_arn,
                    'status': 'Pending' if sub_arn == 'PendingConfirmation' else 'Confirmed',
                    'raw': sub_attrs.get('RawMessageDelivery', '-'),
                    'filter_policy': filter_policy or 'none',
                    'dlq': _resource_name_from_arn(redrive.get('deadLetterTargetArn', '')) if redrive else '',
                })
        tags = _safe_aws([], lambda arn=arn: clients['sns'].list_tags_for_resource(ResourceArn=arn).get('Tags', [])) if arn else []
        policy_str = attrs.get('Policy', '') if attrs else ''
        details.append({
            'name': name,
            'arn': arn,
            'attributes': attrs,
            'access_error': access_error,
            'display_name': attrs.get('DisplayName') or 'Not configured',
            'owner': attrs.get('Owner', '-'),
            'kms': attrs.get('KmsMasterKeyId') or 'Not configured',
            'subscriptions_confirmed': _sns_int(attrs, 'SubscriptionsConfirmed'),
            'subscriptions_pending': _sns_int(attrs, 'SubscriptionsPending'),
            'subscriptions_deleted': _sns_int(attrs, 'SubscriptionsDeleted'),
            'subscriptions': subscriptions,
            'policy': policy_str,
            'policy_statements': _sns_policy_statements(policy_str),
            'tags': tags,
        })
    return details


def render_sns_summary(topics, relations):
    subscriptions = [sub for topic in topics for sub in topic.get('subscriptions', [])]
    confirmed = sum(1 for sub in subscriptions if sub.get('status') == 'Confirmed')
    pending = sum(1 for sub in subscriptions if sub.get('status') == 'Pending')
    encrypted = sum(1 for topic in topics if topic.get('kms') not in ('', 'Not configured'))
    with_policy = sum(1 for topic in topics if topic.get('policy_statements'))
    no_access = sum(1 for topic in topics if topic.get('access_error'))
    sns_relations = [rel for rel in relations if rel.get('source_service') == 'SNS' or rel.get('target_service') == 'SNS']
    rows = [
        ['Topics detected', len(topics)],
        ['Subscriptions detected', len(subscriptions)],
        ['Confirmed subscriptions', confirmed],
        ['Pending subscriptions', pending],
        ['Topics with policies', with_policy],
        ['Topics with KMS encryption', encrypted],
        ['Topics without attribute access', no_access],
        ['SNS relationships detected', len(sns_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_sns_relationship_diagram(relations):
    sns_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'SNS' or rel.get('target_service') == 'SNS'
    ]
    html = html_section('SNS architecture diagram')
    html += render_mermaid_diagram(
        sns_relations,
        'Relationships where SNS appears as source or target, including subscriptions and inferred environment references.'
    )
    return html


def render_sns_topics_inventory(topics):
    if not topics:
        return html_empty('No SNS topics detected.')
    rows = []
    for topic in sorted(topics, key=lambda item: item.get('name', '')):
        if topic.get('access_error'):
            rows.append([
                html_escape(topic.get('name', '-')),
                discovery_badge('No access', 'INFO'),
                '-',
                '-',
                '-',
            ])
            continue
        subscription_metadata = _compact_metadata_lines([
            ('Confirmed', topic.get('subscriptions_confirmed', 0)),
            ('Pending', topic.get('subscriptions_pending', 0)),
            ('Deleted count', topic.get('subscriptions_deleted', 0)),
        ])
        topic_metadata = _compact_metadata_lines([
            ('Display name', topic.get('display_name', 'Not configured')),
            ('Owner', topic.get('owner', '-')),
        ])
        kms = topic.get('kms', 'Not configured')
        statements = topic.get('policy_statements') or []
        policy_label = f"{len(statements)} STATEMENT" if len(statements) == 1 else f"{len(statements)} STATEMENTS" if statements else 'NO POLICY'
        rows.append([
            html_escape(topic.get('name', '-')),
            topic_metadata,
            subscription_metadata,
            discovery_badge('KMS', 'OK') if kms != 'Not configured' else discovery_badge('NO KMS', 'WARN'),
            discovery_badge(policy_label, 'INFO'),
        ])
    return html_table(['Topic', 'Metadata', 'Subscriptions', 'Encryption', 'Policy'], rows, widths=['22%', '30%', '22%', '12%', '14%'])


def render_sns_topic_identifiers(topics):
    if not topics:
        return html_empty('No SNS topic identifiers available.')
    rows = []
    for topic in sorted(topics, key=lambda item: item.get('name', '')):
        rows.append([
            html_escape(topic.get('name', '-')),
            _details_html('View identifier', [f"ARN: {topic.get('arn', '-')}"]),
        ])
    return html_table(['Topic', 'Identifier'], rows, widths=['30%', '70%'])


def render_sns_subscriptions(topics):
    html = ''
    any_subs = False
    for topic in sorted(topics, key=lambda item: item.get('name', '')):
        subs = topic.get('subscriptions') or []
        if not subs:
            continue
        any_subs = True
        rows = []
        for sub in sorted(subs, key=lambda item: (item.get('protocol', ''), item.get('endpoint', ''))):
            identifier = sub.get('arn', '-') or '-'
            details = _compact_metadata_lines([
                ('Raw delivery', sub.get('raw', '-')),
                ('Filter policy', sub.get('filter_policy', 'none')),
                ('DLQ', sub.get('dlq') or 'None'),
                ('Identifier', identifier),
            ])
            rows.append([
                html_escape(sub.get('protocol', '-')),
                html_escape(sub.get('endpoint', '-')),
                discovery_badge(sub.get('status', 'INFO'), 'OK' if sub.get('status') == 'Confirmed' else 'WARN'),
                details,
            ])
        html += html_subsection(topic.get('name', '-'))
        html += html_table(['Protocol', 'Endpoint', 'Status', 'Delivery settings'], rows, widths=['14%', '40%', '14%', '32%'])
    if not any_subs:
        return html_empty('No SNS subscriptions detected.')
    return html


def render_sns_policies(topics):
    if not topics:
        return html_empty('No SNS topics detected.')
    rows = []
    for topic in sorted(topics, key=lambda item: item.get('name', '')):
        if topic.get('access_error'):
            rows.append([html_escape(topic.get('name', '-')), discovery_badge('No access', 'INFO'), 'Could not inspect topic attributes.'])
            continue
        statements = topic.get('policy_statements') or []
        if statements:
            details = _details_html(f'View {len(statements)} statements', [_sns_statement_summary(stmt) for stmt in statements])
            status = discovery_badge(f'{len(statements)} statements', 'INFO')
        else:
            details = 'No topic policy configured.'
            status = discovery_badge('No policy', 'INFO')
        rows.append([html_escape(topic.get('name', '-')), status, details])
    return html_table(['Topic', 'Policy', 'Details'], rows, widths=['30%', '18%', '52%'])


def render_sns_tags(topics):
    rows = []
    for topic in sorted(topics, key=lambda item: item.get('name', '')):
        tags = topic.get('tags') or []
        if tags:
            tag_pairs = sorted((tag.get('Key'), tag.get('Value')) for tag in tags)
            rows.append([html_escape(topic.get('name', '-')), len(tags), _key_value_details_html(f'View {len(tags)} tags', tag_pairs)])
    if not rows:
        return html_empty('No SNS tags detected.')
    return html_table(['Topic', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_sns_relationships(relations):
    sns_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'SNS' or rel.get('target_service') == 'SNS'
    ]
    if not sns_relations:
        return html_empty('No SNS relationships detected.')
    return render_discovery_relations_table(sns_relations)


def render_discovery_sns():
    topics = _sns_collect_details()
    relations = collect_architecture_relations()

    html = html_title('SNS - Discovery')
    html += html_section('Summary')
    html += render_sns_summary(topics, relations)

    html += render_sns_relationship_diagram(relations)

    html += html_section('Topics inventory')
    html += render_sns_topics_inventory(topics)

    html += html_section('Topic identifiers')
    html += render_sns_topic_identifiers(topics)

    html += html_section('Subscriptions')
    html += render_sns_subscriptions(topics)

    html += html_section('Topic policies')
    html += render_sns_policies(topics)

    html += html_section('Tags')
    html += render_sns_tags(topics)

    html += html_section('SNS relationships')
    html += render_sns_relationships(relations)
    return html
```

```python
# === Discovery renderers: S3 ===
S3_OBJECT_SAMPLE_LIMIT = 100

def _s3_region(bucket_name):
    location = _safe_aws({}, lambda bucket_name=bucket_name: clients['s3'].get_bucket_location(Bucket=bucket_name))
    if not location:
        return 'No access'
    return location.get('LocationConstraint') or 'us-east-1'


def _s3_website_url(bucket_name, region):
    if region == 'us-east-1':
        return f'http://{bucket_name}.s3-website-{region}.amazonaws.com'
    return f'http://{bucket_name}.s3-website.{region}.amazonaws.com'


def _s3_bytes_label(size_bytes):
    try:
        size = float(size_bytes or 0)
    except Exception:
        return '0 B'
    if size >= 1024 * 1024:
        return f'{size / (1024 * 1024):.2f} MB'
    if size >= 1024:
        return f'{size / 1024:.1f} KB'
    return f'{int(size)} B'


def _s3_statement_summary(statement):
    effect = statement.get('Effect', 'N/A')
    principal = statement.get('Principal', 'N/A')
    action = statement.get('Action', 'N/A')
    condition = 'yes' if statement.get('Condition') else 'no'
    if isinstance(action, list):
        visible = ', '.join(str(item) for item in action[:3])
        if len(action) > 3:
            visible += f' (+{len(action) - 3} more)'
        action = visible
    return f'Effect: {effect} | Principal: {principal} | Action: {action} | Condition: {condition}'


def _s3_policy_statements(policy_str):
    if not policy_str:
        return []
    try:
        policy = json.loads(policy_str)
    except Exception:
        return []
    statements = policy.get('Statement', [])
    if isinstance(statements, dict):
        statements = [statements]
    return statements if isinstance(statements, list) else []


def _s3_collect_details(sample_limit=S3_OBJECT_SAMPLE_LIMIT):
    buckets = _safe_aws([], lambda: clients['s3'].list_buckets().get('Buckets', []))
    details = []
    for bucket in buckets:
        name = bucket.get('Name')
        created = str(bucket.get('CreationDate', '-'))[:10]
        region = _s3_region(name)

        versioning = _safe_aws({}, lambda name=name: clients['s3'].get_bucket_versioning(Bucket=name))
        versioning_status = versioning.get('Status') or 'Disabled'

        website = _safe_aws(None, lambda name=name: clients['s3'].get_bucket_website(Bucket=name))
        website_enabled = bool(website)
        website_index = ((website or {}).get('IndexDocument') or {}).get('Suffix', '-')
        website_error = ((website or {}).get('ErrorDocument') or {}).get('Key', '-')
        website_url = _s3_website_url(name, region) if website_enabled and region != 'No access' else '-'

        cors = _safe_aws({}, lambda name=name: clients['s3'].get_bucket_cors(Bucket=name))
        cors_rules = cors.get('CORSRules', []) if cors else []

        encryption = _safe_aws({}, lambda name=name: clients['s3'].get_bucket_encryption(Bucket=name))
        encryption_rules = (encryption.get('ServerSideEncryptionConfiguration') or {}).get('Rules', []) if encryption else []
        encryption_label = 'Configured' if encryption_rules else 'Not configured'

        objects_response = _safe_aws({}, lambda name=name, sample_limit=sample_limit: clients['s3'].list_objects_v2(Bucket=name, MaxKeys=sample_limit))
        objects = objects_response.get('Contents', []) if objects_response else []
        sample_count = objects_response.get('KeyCount', 0) if objects_response else 0
        sample_truncated = bool(objects_response.get('IsTruncated')) if objects_response else False
        sample_size = sum(obj.get('Size', 0) for obj in objects)

        tags_response = _safe_aws({}, lambda name=name: clients['s3'].get_bucket_tagging(Bucket=name))
        tags = tags_response.get('TagSet', []) if tags_response else []

        policy_response = _safe_aws({}, lambda name=name: clients['s3'].get_bucket_policy(Bucket=name))
        policy_str = policy_response.get('Policy', '') if policy_response else ''
        policy_statements = _s3_policy_statements(policy_str)

        notification = _safe_aws({}, lambda name=name: clients['s3'].get_bucket_notification_configuration(Bucket=name))
        notification_counts = {
            'Lambda': len(notification.get('LambdaFunctionConfigurations', []) or []),
            'SQS': len(notification.get('QueueConfigurations', []) or []),
            'SNS': len(notification.get('TopicConfigurations', []) or []),
        }

        details.append({
            'name': name,
            'created': created,
            'region': region,
            'versioning': versioning_status,
            'website_enabled': website_enabled,
            'website_index': website_index,
            'website_error': website_error,
            'website_url': website_url,
            'cors_rules': cors_rules,
            'encryption': encryption_label,
            'encryption_rules': encryption_rules,
            'objects': objects,
            'sample_count': sample_count,
            'sample_limit': sample_limit,
            'sample_truncated': sample_truncated,
            'sample_size': sample_size,
            'tags': tags,
            'policy_statements': policy_statements,
            'notification_counts': notification_counts,
        })
    return details


def render_s3_summary(buckets, relations):
    versioned = sum(1 for bucket in buckets if bucket.get('versioning') == 'Enabled')
    websites = sum(1 for bucket in buckets if bucket.get('website_enabled'))
    encrypted = sum(1 for bucket in buckets if bucket.get('encryption') == 'Configured')
    with_cors = sum(1 for bucket in buckets if bucket.get('cors_rules'))
    with_tags = sum(1 for bucket in buckets if bucket.get('tags'))
    with_policy = sum(1 for bucket in buckets if bucket.get('policy_statements'))
    s3_relations = [rel for rel in relations if rel.get('source_service') == 'S3' or rel.get('target_service') == 'S3']
    rows = [
        ['Buckets detected', len(buckets)],
        ['Buckets with versioning enabled', versioned],
        ['Buckets with static website hosting', websites],
        ['Buckets with encryption configured', encrypted],
        ['Buckets with CORS configured', with_cors],
        ['Buckets with tags', with_tags],
        ['Buckets with bucket policy', with_policy],
        ['S3 relationships detected', len(s3_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_s3_relationship_diagram(relations):
    s3_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'S3' or rel.get('target_service') == 'S3'
    ]
    html = html_section('S3 architecture diagram')
    html += render_mermaid_diagram(
        s3_relations,
        'Relationships where S3 appears as source or target, including bucket notifications and inferred environment references.'
    )
    return html


def render_s3_buckets_inventory(buckets):
    if not buckets:
        return html_empty('No S3 buckets detected.')
    rows = []
    for bucket in sorted(buckets, key=lambda item: item.get('name', '')):
        object_note = f"{bucket.get('sample_count', 0)} shown"
        if bucket.get('sample_truncated'):
            object_note += f"; sample limited to {bucket.get('sample_limit')}"
        metadata = _compact_metadata_lines([
            ('Region', bucket.get('region', '-')),
            ('Created', bucket.get('created', '-')),
            ('CORS rules', len(bucket.get('cors_rules') or [])),
            ('Objects sample', object_note),
            ('Sample size', _s3_bytes_label(bucket.get('sample_size'))),
        ])
        states = _compact_metadata_lines([
            ('Versioning', bucket.get('versioning', 'Disabled')),
            ('Website', 'Enabled' if bucket.get('website_enabled') else 'Not configured'),
            ('Encryption', bucket.get('encryption', 'Not configured')),
        ])
        rows.append([
            html_escape(bucket.get('name', '-')),
            metadata,
            states,
        ])
    return html_table(['Bucket', 'Metadata', 'Configuration state'], rows, widths=['28%', '44%', '28%'])


def render_s3_bucket_configuration(buckets):
    if not buckets:
        return html_empty('No S3 bucket configuration available.')
    rows = []
    for bucket in sorted(buckets, key=lambda item: item.get('name', '')):
        website_summary = discovery_badge('WEBSITE', 'OK') if bucket.get('website_enabled') else discovery_badge('NO WEBSITE', 'INFO')
        website_items = [
            f"Status: {'Enabled' if bucket.get('website_enabled') else 'Not configured'}",
            f"Index document: {bucket.get('website_index', '-')}",
            f"Error document: {bucket.get('website_error', '-')}",
            f"URL: {bucket.get('website_url', '-')}",
        ]
        website_details = website_summary + '<br>' + _details_html('View website details', website_items)
        notification_counts = bucket.get('notification_counts') or {}
        notification_details = _compact_metadata_lines([
            ('Lambda', notification_counts.get('Lambda', 0)),
            ('SQS', notification_counts.get('SQS', 0)),
            ('SNS', notification_counts.get('SNS', 0)),
        ])
        rows.append([
            html_escape(bucket.get('name', '-')),
            website_details,
            notification_details,
            _details_html('View CORS rules', [json.dumps(rule, default=str) for rule in bucket.get('cors_rules') or []]) if bucket.get('cors_rules') else 'No CORS configuration',
        ])
    return html_table(['Bucket', 'Website hosting', 'Notifications', 'CORS'], rows, widths=['28%', '30%', '20%', '22%'])


def render_s3_object_sample(buckets):
    rows = []
    for bucket in sorted(buckets, key=lambda item: item.get('name', '')):
        objects = bucket.get('objects') or []
        if not objects:
            rows.append([html_escape(bucket.get('name', '-')), 0, html_empty('No objects returned in the sample.')])
            continue
        items = []
        for obj in objects:
            key = obj.get('Key', '-')
            size = _s3_bytes_label(obj.get('Size', 0))
            modified = str(obj.get('LastModified', '-'))[:19]
            storage = obj.get('StorageClass', 'STANDARD')
            items.append(f'{key} | {size} | {modified} | {storage}')
        summary = f"View {len(objects)} sampled objects"
        if bucket.get('sample_truncated'):
            summary += f" (limited to {bucket.get('sample_limit')})"
        rows.append([html_escape(bucket.get('name', '-')), len(objects), _details_html(summary, items)])
    if not rows:
        return html_empty('No S3 object sample available.')
    return html_table(['Bucket', 'Objects shown', 'Sample'], rows, widths=['28%', '12%', '60%'])


def render_s3_bucket_policies(buckets):
    if not buckets:
        return html_empty('No S3 buckets detected.')
    rows = []
    for bucket in sorted(buckets, key=lambda item: item.get('name', '')):
        statements = bucket.get('policy_statements') or []
        if statements:
            details = _details_html(f'View {len(statements)} statements', [_s3_statement_summary(stmt) for stmt in statements])
            status = discovery_badge(f'{len(statements)} statement' if len(statements) == 1 else f'{len(statements)} statements', 'INFO')
        else:
            details = 'No bucket policy configured.'
            status = discovery_badge('No policy', 'INFO')
        rows.append([html_escape(bucket.get('name', '-')), status, details])
    return html_table(['Bucket', 'Policy', 'Details'], rows, widths=['30%', '18%', '52%'])


def render_s3_tags(buckets):
    rows = []
    for bucket in sorted(buckets, key=lambda item: item.get('name', '')):
        tags = bucket.get('tags') or []
        if tags:
            tag_pairs = sorted((tag.get('Key'), tag.get('Value')) for tag in tags)
            rows.append([html_escape(bucket.get('name', '-')), len(tags), _key_value_details_html(f'View {len(tags)} tags', tag_pairs)])
    if not rows:
        return html_empty('No S3 tags detected.')
    return html_table(['Bucket', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_s3_relationships(relations):
    s3_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'S3' or rel.get('target_service') == 'S3'
    ]
    if not s3_relations:
        return html_empty('No S3 relationships detected.')
    return render_discovery_relations_table(s3_relations)


def render_discovery_s3():
    buckets = _s3_collect_details()
    relations = collect_architecture_relations()

    html = html_title('S3 - Discovery')
    html += html_section('Summary')
    html += render_s3_summary(buckets, relations)

    html += render_s3_relationship_diagram(relations)

    html += html_section('Buckets inventory')
    html += html_note(f'Object sample uses up to {S3_OBJECT_SAMPLE_LIMIT} objects per bucket via ListObjectsV2. Counts and sizes shown here describe only that sample, not the full bucket inventory.')
    html += render_s3_buckets_inventory(buckets)

    html += html_section('Bucket configuration')
    html += render_s3_bucket_configuration(buckets)

    html += html_section('Object sample')
    html += render_s3_object_sample(buckets)

    html += html_section('Bucket policies')
    html += render_s3_bucket_policies(buckets)

    html += html_section('Tags')
    html += render_s3_tags(buckets)

    html += html_section('S3 relationships')
    html += render_s3_relationships(relations)
    return html
```

```python
# === Discovery renderers: API Gateway ===

def method_badge(method):
    method = str(method or '-')
    level = 'INFO' if method == 'OPTIONS' else 'OK'
    return discovery_badge(method, level)


def api_key_badge(required):
    if required is True:
        return discovery_badge('Required', 'OK')
    if required == '-':
        return discovery_badge('Unknown', 'INFO')
    return discovery_badge('Not required', 'INFO')


def auth_badge(auth_type):
    auth_type = str(auth_type or '-')
    if auth_type == 'NONE':
        return discovery_badge(auth_type, 'WARN')
    if auth_type in ('N/A', 'No access'):
        return discovery_badge(auth_type, 'INFO')
    return discovery_badge(auth_type, 'OK')

def _apigw_endpoint_types(api):
    config = api.get('endpointConfiguration') or {}
    types = config.get('types') or []
    return ', '.join(types) if types else '-'


def _apigw_collect_details():
    apis = _safe_aws([], lambda: clients['apigateway'].get_rest_apis().get('items', []))
    details = []
    for api in apis:
        api_id = api.get('id')
        stages = _safe_aws([], lambda api_id=api_id: clients['apigateway'].get_stages(restApiId=api_id).get('item', []))
        resources = _safe_aws([], lambda api_id=api_id: clients['apigateway'].get_resources(restApiId=api_id, limit=500).get('items', []))
        endpoints = []
        integrations = []
        auth_methods = []
        for resource in resources:
            path = resource.get('resourcePath', resource.get('path', '/'))
            resource_id = resource.get('id')
            for method in (resource.get('resourceMethods') or {}).keys():
                endpoint_type = 'CORS' if method == 'OPTIONS' else 'Functional'
                endpoints.append({'method': method, 'path': path, 'type': endpoint_type})

                method_cfg = _safe_aws({}, lambda api_id=api_id, resource_id=resource_id, method=method: clients['apigateway'].get_method(
                    restApiId=api_id,
                    resourceId=resource_id,
                    httpMethod=method
                ))
                auth_methods.append({
                    'method': method,
                    'path': path,
                    'auth': method_cfg.get('authorizationType', 'N/A') if method_cfg else 'No access',
                    'api_key_required': method_cfg.get('apiKeyRequired', False) if method_cfg else '-',
                    'request_validator': method_cfg.get('requestValidatorId', '-') if method_cfg else '-',
                })

                if method == 'OPTIONS':
                    continue
                integration = _safe_aws({}, lambda api_id=api_id, resource_id=resource_id, method=method: clients['apigateway'].get_integration(
                    restApiId=api_id,
                    resourceId=resource_id,
                    httpMethod=method
                ))
                if integration:
                    uri = integration.get('uri', '')
                    lambda_name = _lambda_name_from_uri(uri) or '-'
                    integrations.append({
                        'method': method,
                        'path': path,
                        'type': integration.get('type', '-'),
                        'lambda': lambda_name,
                    })

        deployments = _safe_aws([], lambda api_id=api_id: clients['apigateway'].get_deployments(restApiId=api_id).get('items', []))
        arn = f"arn:aws:apigateway:{clients['apigateway'].meta.region_name}::/restapis/{api_id}"
        tags = _safe_aws({}, lambda arn=arn: clients['apigateway'].get_tags(resourceArn=arn).get('tags', {}))

        details.append({
            'api': api,
            'stages': stages,
            'resources': resources,
            'endpoints': endpoints,
            'integrations': integrations,
            'auth_methods': auth_methods,
            'deployments': deployments,
            'tags': tags,
        })
    return details


def render_apigateway_summary(api_details):
    stages = sum(len(item['stages']) for item in api_details)
    methods = sum(len(item['endpoints']) for item in api_details)
    functional = sum(1 for item in api_details for endpoint in item['endpoints'] if endpoint['type'] == 'Functional')
    cors = sum(1 for item in api_details for endpoint in item['endpoints'] if endpoint['type'] == 'CORS')
    integrations = sum(len(item['integrations']) for item in api_details)
    authorized = sum(1 for item in api_details for auth in item['auth_methods'] if auth.get('auth') not in ('NONE', 'N/A', 'No access'))
    api_key_required = sum(1 for item in api_details for auth in item['auth_methods'] if auth.get('api_key_required') is True)
    validators = len({auth.get('request_validator') for item in api_details for auth in item['auth_methods'] if auth.get('request_validator') not in ('-', None)})
    deployments = sum(len(item['deployments']) for item in api_details)
    rows = [
        ['REST APIs detected', len(api_details)],
        ['Stages detected', stages],
        ['Methods detected', methods],
        ['Functional methods', functional],
        ['CORS OPTIONS methods', cors],
        ['Lambda integrations', integrations],
        ['Methods with authorization', authorized],
        ['Methods requiring API key', api_key_required],
        ['Request validators used', validators],
        ['Deployments detected', deployments],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_apigateway_relationship_diagram(relations):
    api_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'API Gateway' or rel.get('target_service') == 'API Gateway'
    ]
    html = html_section('API Gateway architecture diagram')
    html += render_mermaid_diagram(
        api_relations,
        'Relationships where API Gateway appears as source or target, including Lambda integrations, Cognito authorizers and WAF associations.'
    )
    return html


def render_apigateway_inventory(api_details):
    if not api_details:
        return html_empty('No API Gateway REST APIs detected.')
    rows = []
    for item in sorted(api_details, key=lambda data: data['api'].get('name', '')):
        api = item['api']
        stage_names = [stage.get('stageName') for stage in item['stages']]
        functional = sum(1 for endpoint in item['endpoints'] if endpoint['type'] == 'Functional')
        cors = sum(1 for endpoint in item['endpoints'] if endpoint['type'] == 'CORS')
        rows.append([
            html_escape(api.get('name', '-')),
            html_escape(api.get('id', '-')),
            html_escape(str(api.get('createdDate', '-'))[:10]),
            html_escape(_apigw_endpoint_types(api)),
            render_inventory_details(stage_names) if stage_names else '-',
            len(item['endpoints']),
            functional,
            cors,
        ])
    return html_table(['API', 'ID', 'Created', 'Endpoint type', 'Stages', 'Methods', 'Functional', 'CORS'], rows, widths=['18%', '11%', '11%', '13%', '17%', '10%', '10%', '10%'])


def render_apigateway_endpoints(api_details):
    if not any(item['endpoints'] for item in api_details):
        return html_empty('No API Gateway endpoints detected.')
    html = ''
    for item in sorted(api_details, key=lambda data: data['api'].get('name', '')):
        api_name = item['api'].get('name', '-')
        endpoints = sorted(item['endpoints'], key=lambda ep: (ep['path'], ep['method']))
        if not endpoints:
            continue
        rows = []
        for endpoint in endpoints:
            rows.append([
                method_badge(endpoint['method']),
                html_escape(endpoint['path']),
                html_escape(endpoint['type']),
            ])
        html += html_subsection(api_name)
        html += html_table(['Method', 'Path', 'Type'], rows, widths=['16%', '58%', '26%'])
    return html


def render_apigateway_integrations(api_details):
    if not any(item['integrations'] for item in api_details):
        return html_empty('No API Gateway Lambda integrations detected.')
    html = ''
    for item in sorted(api_details, key=lambda data: data['api'].get('name', '')):
        api_name = item['api'].get('name', '-')
        integrations = sorted(item['integrations'], key=lambda row: (row['path'], row['method']))
        if not integrations:
            continue
        rows = []
        for integration in integrations:
            rows.append([
                method_badge(integration['method']),
                html_escape(integration['path']),
                html_escape(integration['type']),
                html_escape(integration['lambda']),
            ])
        html += html_subsection(api_name)
        html += html_table(['Method', 'Path', 'Integration', 'Lambda'], rows, widths=['14%', '42%', '20%', '24%'])
    return html


def render_apigateway_authorization(api_details):
    if not any(item['auth_methods'] for item in api_details):
        return html_empty('No API Gateway method authorization data detected.')
    html = ''
    for item in sorted(api_details, key=lambda data: data['api'].get('name', '')):
        api_name = item['api'].get('name', '-')
        auth_methods = sorted(item['auth_methods'], key=lambda row: (row['path'], row['method']))
        if not auth_methods:
            continue
        rows = []
        for auth in auth_methods:
            auth_type = auth.get('auth', '-')
            rows.append([
                method_badge(auth['method']),
                html_escape(auth['path']),
                auth_badge(auth_type),
                api_key_badge(auth.get('api_key_required')),
                html_escape(auth.get('request_validator', '-')),
            ])
        html += html_subsection(api_name)
        html += html_table(['Method', 'Path', 'Auth', 'API key', 'Request validator'], rows, widths=['12%', '38%', '20%', '15%', '15%'])
    return html


def render_apigateway_deployments(api_details):
    rows = []
    for item in sorted(api_details, key=lambda data: data['api'].get('name', '')):
        api_name = item['api'].get('name', '-')
        for deployment in sorted(item['deployments'], key=lambda dep: str(dep.get('createdDate', '')), reverse=True):
            rows.append([
                html_escape(api_name),
                html_escape(str(deployment.get('createdDate', '-'))[:19]),
                html_escape(deployment.get('description', '-') or '-'),
            ])
    if not rows:
        return html_empty('No API Gateway deployments detected.')
    return html_table(['API', 'Created', 'Description'], rows, widths=['28%', '22%', '50%'])


def render_apigateway_tags(api_details):
    rows = []
    for item in sorted(api_details, key=lambda data: data['api'].get('name', '')):
        api_name = item['api'].get('name', '-')
        tags = item.get('tags') or {}
        if tags:
            rows.append([html_escape(api_name), len(tags), _key_value_details_html(f'View {len(tags)} tags', sorted(tags.items()))])
    if not rows:
        return html_empty('No API Gateway tags detected.')
    return html_table(['API', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_apigateway_relationships(relations):
    api_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'API Gateway' or rel.get('target_service') == 'API Gateway'
    ]
    if not api_relations:
        return html_empty('No API Gateway relationships detected.')
    return render_discovery_relations_table(api_relations)


def render_discovery_apigateway():
    api_details = _apigw_collect_details()
    relations = collect_architecture_relations()

    html = html_title('API Gateway - Discovery')
    html += html_section('Summary')
    html += render_apigateway_summary(api_details)

    html += render_apigateway_relationship_diagram(relations)

    html += html_section('APIs inventory')
    html += render_apigateway_inventory(api_details)

    html += html_section('Endpoints')
    html += render_apigateway_endpoints(api_details)

    html += html_section('Lambda integrations')
    html += render_apigateway_integrations(api_details)

    html += html_section('Authorization by method')
    html += render_apigateway_authorization(api_details)

    html += html_section('Deployments')
    html += render_apigateway_deployments(api_details)

    html += html_section('Tags')
    html += render_apigateway_tags(api_details)

    html += html_section('API Gateway relationships')
    html += render_apigateway_relationships(relations)
    return html
```

```python
# === Discovery renderers: RDS + Cognito + Step Functions ===

# ---------------------------------------------------------------------------
# RDS Discovery
# ---------------------------------------------------------------------------

def _rds_collect_details():
    clusters = _safe_aws([], lambda: clients['rds'].describe_db_clusters().get('DBClusters', []))
    instances = _safe_aws([], lambda: clients['rds'].describe_db_instances().get('DBInstances', []))
    return clusters, instances


def _rds_endpoint_label(endpoint):
    endpoint = endpoint or {}
    address = endpoint.get('Address', 'N/A')
    port = endpoint.get('Port', 'N/A')
    return f'{address}:{port}' if address != 'N/A' else 'N/A'


def _rds_yes_no(value):
    return 'Yes' if bool(value) else 'No'


def _rds_instance_subnet_details(instance):
    subnet_group = instance.get('DBSubnetGroup') or {}
    subnets = []
    for subnet in subnet_group.get('Subnets', []) or []:
        subnet_id = subnet.get('SubnetIdentifier', 'N/A')
        subnet_az = (subnet.get('SubnetAvailabilityZone') or {}).get('Name', 'N/A')
        subnets.append(f'{subnet_id} ({subnet_az})')
    return subnet_group, subnets


def _rds_tag_pairs(arn):
    if not arn:
        return []
    tags = _safe_aws([], lambda arn=arn: clients['rds'].list_tags_for_resource(ResourceName=arn).get('TagList', []))
    return sorted((tag.get('Key'), tag.get('Value')) for tag in tags)


def render_rds_summary(clusters, instances, relations):
    available_instances = sum(1 for inst in instances if inst.get('DBInstanceStatus') == 'available')
    encrypted_instances = sum(1 for inst in instances if inst.get('StorageEncrypted'))
    public_instances = sum(1 for inst in instances if inst.get('PubliclyAccessible'))
    cluster_members = sum(len(cluster.get('DBClusterMembers') or []) for cluster in clusters)
    rds_relations = [rel for rel in relations if rel.get('source_service') == 'RDS' or rel.get('target_service') == 'RDS']
    rows = [
        ['Clusters detected', len(clusters)],
        ['Instances detected', len(instances)],
        ['Available instances', available_instances],
        ['Encrypted instances', encrypted_instances],
        ['Public instances', public_instances],
        ['Cluster members', cluster_members],
        ['RDS relationships detected', len(rds_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_rds_relationship_diagram(relations):
    rds_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'RDS' or rel.get('target_service') == 'RDS'
    ]
    html = html_section('RDS architecture diagram')
    html += render_mermaid_diagram(
        rds_relations,
        'Relationships where RDS appears as source or target. In this dashboard these are usually inferred from environment variables that reference known RDS instances or endpoints.'
    )
    return html


def render_rds_clusters(clusters):
    if not clusters:
        return html_empty('No RDS clusters detected.')
    rows = []
    for cluster in sorted(clusters, key=lambda item: item.get('DBClusterIdentifier', '')):
        metadata = _compact_metadata_lines([
            ('Created', str(cluster.get('ClusterCreateTime', 'N/A'))[:10]),
            ('Engine', f"{cluster.get('Engine', '-')} {cluster.get('EngineVersion', '')}".strip()),
            ('Database', cluster.get('DatabaseName', 'N/A')),
            ('Admin user', cluster.get('MasterUsername', 'N/A')),
            ('Port', cluster.get('Port', 'N/A')),
        ])
        operations = _compact_metadata_lines([
            ('Backup', f"{cluster.get('BackupRetentionPeriod', 'N/A')} days"),
            ('Backup window', cluster.get('PreferredBackupWindow', 'N/A')),
            ('Maintenance', cluster.get('PreferredMaintenanceWindow', 'N/A')),
            ('Deletion protection', _rds_yes_no(cluster.get('DeletionProtection'))),
            ('Encrypted', _rds_yes_no(cluster.get('StorageEncrypted'))),
        ])
        rows.append([
            html_escape(cluster.get('DBClusterIdentifier', '-')),
            discovery_badge(cluster.get('Status', 'INFO'), 'OK' if cluster.get('Status') == 'available' else 'INFO'),
            metadata,
            operations,
        ])
    return html_table(['Cluster', 'Status', 'Metadata', 'Operations'], rows, widths=['22%', '12%', '34%', '32%'])


def render_rds_instances(instances):
    if not instances:
        return html_empty('No RDS instances detected.')
    rows = []
    for inst in sorted(instances, key=lambda item: item.get('DBInstanceIdentifier', '')):
        subnet_group, _ = _rds_instance_subnet_details(inst)
        metadata = _compact_metadata_lines([
            ('Created', str(inst.get('InstanceCreateTime', 'N/A'))[:10]),
            ('Class', inst.get('DBInstanceClass', 'N/A')),
            ('Engine', f"{inst.get('Engine', '-')} {inst.get('EngineVersion', '')}".strip()),
            ('AZ', inst.get('AvailabilityZone', 'N/A')),
            ('Storage', f"{inst.get('AllocatedStorage', 'N/A')} GB"),
        ])
        network = _compact_metadata_lines([
            ('Public access', _rds_yes_no(inst.get('PubliclyAccessible'))),
            ('Subnet group', subnet_group.get('DBSubnetGroupName', 'N/A')),
            ('VPC', subnet_group.get('VpcId', 'N/A')),
        ])
        rows.append([
            html_escape(inst.get('DBInstanceIdentifier', '-')),
            discovery_badge(inst.get('DBInstanceStatus', 'INFO'), 'OK' if inst.get('DBInstanceStatus') == 'available' else 'INFO'),
            metadata,
            network,
        ])
    return html_table(['Instance', 'Status', 'Metadata', 'Network'], rows, widths=['24%', '12%', '36%', '28%'])


def render_rds_connectivity(clusters, instances):
    rows = []
    for cluster in sorted(clusters, key=lambda item: item.get('DBClusterIdentifier', '')):
        members = []
        for member in cluster.get('DBClusterMembers') or []:
            role = 'Writer' if member.get('IsClusterWriter') else 'Reader'
            members.append(f"{member.get('DBInstanceIdentifier', '-')} ({role})")
        rows.append([
            html_escape(cluster.get('DBClusterIdentifier', '-')),
            'Cluster',
            _details_html('View endpoints', [
                f"Writer: {cluster.get('Endpoint', 'N/A')}",
                f"Reader: {cluster.get('ReaderEndpoint', 'N/A')}",
            ]),
            _details_html(f'View {len(members)} members', members) if members else 'No members detected',
            '-',
        ])
    for inst in sorted(instances, key=lambda item: item.get('DBInstanceIdentifier', '')):
        _, subnets = _rds_instance_subnet_details(inst)
        groups = []
        groups.extend(f"Security group: {sg.get('VpcSecurityGroupId', 'N/A')} ({sg.get('Status', 'N/A')})" for sg in inst.get('VpcSecurityGroups') or [])
        groups.extend(f"Parameter group: {pg.get('DBParameterGroupName', 'N/A')} ({pg.get('ParameterApplyStatus', 'N/A')})" for pg in inst.get('DBParameterGroups') or [])
        groups.extend(f"Option group: {og.get('OptionGroupName', 'N/A')} ({og.get('Status', 'N/A')})" for og in inst.get('OptionGroupMemberships') or [])
        rows.append([
            html_escape(inst.get('DBInstanceIdentifier', '-')),
            'Instance',
            _details_html('View endpoint', [_rds_endpoint_label(inst.get('Endpoint'))]),
            _details_html('View groups', groups) if groups else 'No groups detected',
            _details_html(f'View {len(subnets)} subnets', subnets) if subnets else 'No subnets detected',
        ])
    if not rows:
        return html_empty('No RDS connectivity details available.')
    return html_table(['Resource', 'Type', 'Endpoints', 'Groups / members', 'Subnets'], rows, widths=['28%', '10%', '24%', '24%', '14%'])


def render_rds_tags(clusters, instances):
    rows = []
    for cluster in sorted(clusters, key=lambda item: item.get('DBClusterIdentifier', '')):
        pairs = _rds_tag_pairs(cluster.get('DBClusterArn'))
        if pairs:
            rows.append([html_escape(cluster.get('DBClusterIdentifier', '-')), 'Cluster', len(pairs), _key_value_details_html(f'View {len(pairs)} tags', pairs)])
    for inst in sorted(instances, key=lambda item: item.get('DBInstanceIdentifier', '')):
        pairs = _rds_tag_pairs(inst.get('DBInstanceArn'))
        if pairs:
            rows.append([html_escape(inst.get('DBInstanceIdentifier', '-')), 'Instance', len(pairs), _key_value_details_html(f'View {len(pairs)} tags', pairs)])
    if not rows:
        return html_empty('No RDS tags detected.')
    return html_table(['Resource', 'Type', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '12%', '46%'])


def render_rds_relationships(relations):
    rds_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'RDS' or rel.get('target_service') == 'RDS'
    ]
    if not rds_relations:
        return html_empty('No RDS relationships detected.')
    return render_discovery_relations_table(rds_relations)


def render_discovery_rds():
    clusters, instances = _rds_collect_details()
    relations = collect_architecture_relations()

    html = html_title('RDS - Discovery')
    html += html_section('Summary')
    html += render_rds_summary(clusters, instances, relations)

    html += render_rds_relationship_diagram(relations)

    html += html_section('Clusters')
    html += render_rds_clusters(clusters)

    html += html_section('Instances')
    html += render_rds_instances(instances)

    html += html_section('Connectivity details')
    html += render_rds_connectivity(clusters, instances)

    html += html_section('Tags')
    html += render_rds_tags(clusters, instances)

    html += html_section('RDS relationships')
    html += render_rds_relationships(relations)
    return html

# ---------------------------------------------------------------------------
# Cognito Discovery
# ---------------------------------------------------------------------------

def _cognito_collect_details():
    pool_summaries = _safe_aws([], lambda: clients['cognito'].list_user_pools(MaxResults=60).get('UserPools', []))
    pools = []
    for summary in pool_summaries:
        pool_id = summary.get('Id')
        pool = _safe_aws({}, lambda pool_id=pool_id: clients['cognito'].describe_user_pool(UserPoolId=pool_id).get('UserPool', {}))
        access_error = not bool(pool)
        clients_list = []
        users = []
        groups = []
        providers = []
        tags = {}
        if not access_error:
            clients_summary = _safe_aws([], lambda pool_id=pool_id: clients['cognito'].list_user_pool_clients(UserPoolId=pool_id, MaxResults=60).get('UserPoolClients', []))
            for client_summary in clients_summary:
                client_id = client_summary.get('ClientId')
                client = _safe_aws({}, lambda pool_id=pool_id, client_id=client_id: clients['cognito'].describe_user_pool_client(UserPoolId=pool_id, ClientId=client_id).get('UserPoolClient', {}))
                clients_list.append(client or client_summary)
            users = _safe_aws([], lambda pool_id=pool_id: clients['cognito'].list_users(UserPoolId=pool_id, Limit=60).get('Users', []))
            groups = _safe_aws([], lambda pool_id=pool_id: clients['cognito'].list_groups(UserPoolId=pool_id, Limit=60).get('Groups', []))
            providers = _safe_aws([], lambda pool_id=pool_id: clients['cognito'].list_identity_providers(UserPoolId=pool_id, MaxResults=60).get('Providers', []))
            arn = pool.get('Arn', '')
            tags = _safe_aws({}, lambda arn=arn: clients['cognito'].list_tags_for_resource(ResourceArn=arn).get('Tags', {})) if arn else {}
        pools.append({
            'summary': summary,
            'pool': pool,
            'id': pool_id,
            'name': pool.get('Name') or summary.get('Name') or pool_id,
            'access_error': access_error,
            'clients': clients_list,
            'users': users,
            'groups': groups,
            'providers': providers,
            'tags': tags,
        })
    return pools


def _cognito_password_policy_label(pool):
    policy = ((pool or {}).get('Policies') or {}).get('PasswordPolicy') or {}
    return _compact_metadata_lines([
        ('Minimum length', policy.get('MinimumLength', 'N/A')),
        ('Uppercase', _rds_yes_no(policy.get('RequireUppercase'))),
        ('Lowercase', _rds_yes_no(policy.get('RequireLowercase'))),
        ('Numbers', _rds_yes_no(policy.get('RequireNumbers'))),
        ('Symbols', _rds_yes_no(policy.get('RequireSymbols'))),
    ])


def _cognito_user_attrs(user):
    return {attr.get('Name'): attr.get('Value') for attr in user.get('Attributes', [])}


def render_cognito_summary(pools, relations):
    clients_count = sum(len(pool.get('clients') or []) for pool in pools)
    users_count = sum(len(pool.get('users') or []) for pool in pools)
    groups_count = sum(len(pool.get('groups') or []) for pool in pools)
    providers_count = sum(len(pool.get('providers') or []) for pool in pools)
    mfa_enabled = sum(1 for item in pools if (item.get('pool') or {}).get('MfaConfiguration') not in (None, 'OFF'))
    cognito_relations = [rel for rel in relations if rel.get('source_service') == 'Cognito' or rel.get('target_service') == 'Cognito']
    rows = [
        ['User pools detected', len(pools)],
        ['App clients detected', clients_count],
        ['Users shown', users_count],
        ['Groups detected', groups_count],
        ['Identity providers detected', providers_count],
        ['Pools with MFA enabled', mfa_enabled],
        ['Cognito relationships detected', len(cognito_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_cognito_relationship_diagram(relations):
    cognito_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'Cognito' or rel.get('target_service') == 'Cognito'
    ]
    html = html_section('Cognito architecture diagram')
    html += render_mermaid_diagram(
        cognito_relations,
        'Relationships where Cognito appears as source or target, including API Gateway authorizers and Lambda user pool triggers.'
    )
    return html


def render_cognito_user_pools(pools):
    if not pools:
        return html_empty('No Cognito user pools detected.')
    rows = []
    for item in sorted(pools, key=lambda data: data.get('name', '')):
        pool = item.get('pool') or {}
        if item.get('access_error'):
            rows.append([html_escape(item.get('name', '-')), html_escape(item.get('id', '-')), discovery_badge('No access', 'INFO'), '-', '-'])
            continue
        schema = pool.get('SchemaAttributes') or []
        custom_attrs = [attr.get('Name') for attr in schema if str(attr.get('Name', '')).startswith('custom:')]
        standard_attrs = [attr.get('Name') for attr in schema if not str(attr.get('Name', '')).startswith('custom:')]
        metadata = _compact_metadata_lines([
            ('Created', str(pool.get('CreationDate', 'N/A'))[:10]),
            ('Modified', str(pool.get('LastModifiedDate', 'N/A'))[:10]),
            ('Estimated users', pool.get('EstimatedNumberOfUsers', 0)),
            ('Domain', pool.get('Domain') or 'Not configured'),
        ])
        verification_items = [
            ('MFA', pool.get('MfaConfiguration', 'OFF')),
            ('Auto-verified', ', '.join(pool.get('AutoVerifiedAttributes') or []) or 'None'),
            ('Standard attributes', len(standard_attrs)),
            ('Custom attributes', len(custom_attrs)),
        ]
        verification = _compact_metadata_lines(verification_items)
        if custom_attrs:
            verification += _details_html('View custom attributes', custom_attrs)
        rows.append([
            html_escape(item.get('name', '-')),
            html_escape(item.get('id', '-')),
            metadata,
            verification,
            _cognito_password_policy_label(pool),
        ])
    return html_table(['User pool', 'ID', 'Metadata', 'Verification', 'Password policy'], rows, widths=['18%', '18%', '27%', '22%', '15%'])


def render_cognito_app_clients(pools):
    rows = []
    for item in sorted(pools, key=lambda data: data.get('name', '')):
        for client in sorted(item.get('clients') or [], key=lambda data: data.get('ClientName', data.get('ClientId', ''))):
            flows = client.get('ExplicitAuthFlows') or []
            oauth_flows = client.get('AllowedOAuthFlows') or []
            scopes = client.get('AllowedOAuthScopes') or []
            callbacks = client.get('CallbackURLs') or []
            logouts = client.get('LogoutURLs') or []
            token_validity = _compact_metadata_lines([
                ('Access token', client.get('AccessTokenValidity', 'default')),
                ('ID token', client.get('IdTokenValidity', 'default')),
                ('Refresh token', client.get('RefreshTokenValidity', 'default')),
            ])
            auth_details = _details_html('View auth settings', [
                f"Auth flows: {', '.join(flows) if flows else 'None'}",
                f"OAuth flows: {', '.join(oauth_flows) if oauth_flows else 'None'}",
                f"OAuth scopes: {', '.join(scopes) if scopes else 'None'}",
                f"Callback URLs: {', '.join(callbacks) if callbacks else 'None'}",
                f"Logout URLs: {', '.join(logouts) if logouts else 'None'}",
            ])
            rows.append([
                html_escape(item.get('name', '-')),
                html_escape(client.get('ClientName', '-')),
                html_escape(client.get('ClientId', '-')),
                token_validity,
                auth_details,
            ])
    if not rows:
        return html_empty('No Cognito app clients detected.')
    return html_table(['User pool', 'Client', 'Client ID', 'Token validity', 'Auth and URLs'], rows, widths=['20%', '20%', '20%', '20%', '20%'])


def render_cognito_users(pools):
    rows = []
    for item in sorted(pools, key=lambda data: data.get('name', '')):
        for user in sorted(item.get('users') or [], key=lambda data: data.get('Username', '')):
            attrs = _cognito_user_attrs(user)
            details = _compact_metadata_lines([
                ('Enabled', _rds_yes_no(user.get('Enabled'))),
                ('Email', attrs.get('email', '-')),
                ('Email verified', attrs.get('email_verified', '-')),
                ('Created', str(user.get('UserCreateDate', 'N/A'))[:19]),
                ('Modified', str(user.get('UserLastModifiedDate', 'N/A'))[:19]),
            ])
            rows.append([
                html_escape(item.get('name', '-')),
                html_escape(user.get('Username', '-')),
                discovery_badge(user.get('UserStatus', 'INFO'), 'OK' if user.get('UserStatus') in ('CONFIRMED', 'FORCE_CHANGE_PASSWORD') else 'INFO'),
                details,
            ])
    if not rows:
        return html_empty('No Cognito users returned in the sample.')
    return html_table(['User pool', 'Username', 'Status', 'Details'], rows, widths=['22%', '24%', '14%', '40%'])


def render_cognito_groups_and_providers(pools):
    rows = []
    for item in sorted(pools, key=lambda data: data.get('name', '')):
        groups = item.get('groups') or []
        providers = item.get('providers') or []
        group_items = [f"{group.get('GroupName', '-')}: {group.get('Description', '-') or '-'}" for group in groups]
        provider_items = [f"{provider.get('ProviderName', '-')} ({provider.get('ProviderType', 'N/A')})" for provider in providers]
        rows.append([
            html_escape(item.get('name', '-')),
            len(groups),
            _details_html(f'View {len(groups)} groups', group_items) if group_items else 'No groups',
            len(providers),
            _details_html(f'View {len(providers)} providers', provider_items) if provider_items else 'No external providers',
        ])
    return html_table(['User pool', 'Groups', 'Group details', 'Providers', 'Provider details'], rows, widths=['24%', '10%', '28%', '10%', '28%'])


def render_cognito_tags(pools):
    rows = []
    for item in sorted(pools, key=lambda data: data.get('name', '')):
        tags = item.get('tags') or {}
        if tags:
            rows.append([html_escape(item.get('name', '-')), len(tags), _key_value_details_html(f'View {len(tags)} tags', sorted(tags.items()))])
    if not rows:
        return html_empty('No Cognito tags detected.')
    return html_table(['User pool', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_cognito_relationships(relations):
    cognito_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'Cognito' or rel.get('target_service') == 'Cognito'
    ]
    if not cognito_relations:
        return html_empty('No Cognito relationships detected.')
    return render_discovery_relations_table(cognito_relations)


def render_discovery_cognito():
    pools = _cognito_collect_details()
    relations = collect_architecture_relations()

    html = html_title('Cognito - Discovery')
    html += html_section('Summary')
    html += render_cognito_summary(pools, relations)

    html += render_cognito_relationship_diagram(relations)

    html += html_section('User pools')
    html += render_cognito_user_pools(pools)

    html += html_section('Application clients')
    html += render_cognito_app_clients(pools)

    html += html_section('Users')
    html += html_note('Users are listed with Cognito list_users using a limit of 60 per user pool.')
    html += render_cognito_users(pools)

    html += html_section('Groups and identity providers')
    html += render_cognito_groups_and_providers(pools)

    html += html_section('Tags')
    html += render_cognito_tags(pools)

    html += html_section('Cognito relationships')
    html += render_cognito_relationships(relations)
    return html

# ---------------------------------------------------------------------------
# Step Functions Discovery
# ---------------------------------------------------------------------------

def _sfn_collect_details():
    machines = _safe_aws([], lambda: clients['stepfunctions'].list_state_machines().get('stateMachines', []))
    details = []
    for sm in machines:
        arn = sm.get('stateMachineArn')
        detail = _safe_aws({}, lambda arn=arn: clients['stepfunctions'].describe_state_machine(stateMachineArn=arn))
        definition_text = detail.get('definition', '{}') if detail else '{}'
        try:
            definition = json.loads(definition_text)
        except Exception:
            definition = {}
        executions = []
        if sm.get('type') != 'EXPRESS':
            executions = _safe_aws([], lambda arn=arn: clients['stepfunctions'].list_executions(stateMachineArn=arn, maxResults=10).get('executions', []))
        tags = _safe_aws([], lambda arn=arn: clients['stepfunctions'].list_tags_for_resource(resourceArn=arn).get('tags', []))
        details.append({
            'summary': sm,
            'detail': detail,
            'definition': definition,
            'executions': executions,
            'tags': tags,
        })
    return details


def _sfn_role_name(role_arn):
    role_arn = str(role_arn or 'N/A')
    return role_arn.split('/')[-1] if role_arn != 'N/A' else 'N/A'


def _sfn_state_target(state_def):
    resource = str(state_def.get('Resource', ''))
    params = state_def.get('Parameters') or {}
    if 'lambda:invoke' in resource:
        fn_arn = params.get('FunctionName') or params.get('FunctionName.$') or ''
        return 'Lambda', _resource_name_from_arn(fn_arn) if fn_arn else '-'
    if ':function:' in resource:
        return 'Lambda', _resource_name_from_arn(resource)
    if 'sns' in resource.lower():
        topic_arn = params.get('TopicArn') or params.get('TopicArn.$') or ''
        return 'SNS', _resource_name_from_arn(topic_arn) if topic_arn else 'SNS integration'
    if 'sqs' in resource.lower():
        queue_url = params.get('QueueUrl') or params.get('QueueUrl.$') or ''
        return 'SQS', _sqs_queue_name(queue_url) if queue_url else 'SQS integration'
    if 'dynamodb' in resource.lower():
        return 'DynamoDB', 'DynamoDB integration'
    return 'Task', resource.split(':')[-1] if resource else '-'


def _sfn_next_or_end(state_def):
    if state_def.get('Next'):
        return state_def.get('Next')
    if state_def.get('End') is True:
        return 'End'
    return '-'


def _sfn_extract_states(definition):
    rows = []
    states = (definition or {}).get('States') or {}
    for state_name, state_def in states.items():
        state_type = state_def.get('Type', 'N/A')
        if state_type == 'Task':
            target_service, target = _sfn_state_target(state_def)
            rows.append((state_name, state_type, target_service, target, _sfn_next_or_end(state_def)))
        elif state_type == 'Parallel':
            branches = state_def.get('Branches') or []
            rows.append((state_name, state_type, 'Branches', f'{len(branches)} branch(es)', _sfn_next_or_end(state_def)))
            for idx, branch in enumerate(branches, 1):
                for branch_state, branch_def in (branch.get('States') or {}).items():
                    target_service, target = _sfn_state_target(branch_def) if branch_def.get('Type') == 'Task' else (branch_def.get('Type', 'N/A'), '-')
                    rows.append((f'Branch {idx}: {branch_state}', branch_def.get('Type', 'N/A'), target_service, target, _sfn_next_or_end(branch_def)))
        elif state_type == 'Choice':
            choices = state_def.get('Choices') or []
            rows.append((state_name, state_type, 'Choices', f'{len(choices)} option(s)', state_def.get('Default', '-')))
        elif state_type == 'Wait':
            rows.append((state_name, state_type, 'Wait', f"{state_def.get('Seconds', 'N/A')} seconds", _sfn_next_or_end(state_def)))
        elif state_type in ('Succeed', 'Fail'):
            rows.append((state_name, state_type, state_type, state_def.get('Error', 'End'), '-'))
        else:
            rows.append((state_name, state_type, '-', '-', _sfn_next_or_end(state_def)))
    return rows


def render_sfn_summary(machines, relations):
    express = sum(1 for item in machines if item.get('summary', {}).get('type') == 'EXPRESS')
    standard = sum(1 for item in machines if item.get('summary', {}).get('type') == 'STANDARD')
    logging_enabled = sum(1 for item in machines if (item.get('detail', {}).get('loggingConfiguration') or {}).get('level', 'OFF') != 'OFF')
    total_states = sum(len((item.get('definition') or {}).get('States') or {}) for item in machines)
    sfn_relations = [rel for rel in relations if rel.get('source_service') == 'Step Functions' or rel.get('target_service') == 'Step Functions']
    rows = [
        ['State machines detected', len(machines)],
        ['STANDARD machines', standard],
        ['EXPRESS machines', express],
        ['Machines with logging enabled', logging_enabled],
        ['Top-level states detected', total_states],
        ['Step Functions relationships detected', len(sfn_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_sfn_relationship_diagram(relations):
    sfn_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'Step Functions' or rel.get('target_service') == 'Step Functions'
    ]
    html = html_section('Step Functions architecture diagram')
    html += render_mermaid_diagram(
        sfn_relations,
        'Relationships where Step Functions appears as source or target, usually from task resources in state machine definitions.'
    )
    return html


def render_sfn_state_machines(machines):
    if not machines:
        return html_empty('No Step Functions state machines detected.')
    rows = []
    for item in sorted(machines, key=lambda data: data.get('summary', {}).get('name', '')):
        sm = item.get('summary') or {}
        detail = item.get('detail') or {}
        logging_config = detail.get('loggingConfiguration') or {}
        tracing_config = detail.get('tracingConfiguration') or {}
        metadata = _compact_metadata_lines([
            ('Created', str(sm.get('creationDate', 'N/A'))[:10]),
            ('Type', sm.get('type', 'N/A')),
            ('Role', _sfn_role_name(detail.get('roleArn'))),
            ('Logging', logging_config.get('level', 'OFF')),
            ('Tracing', _rds_yes_no(tracing_config.get('enabled'))),
        ])
        states_count = len((item.get('definition') or {}).get('States') or {})
        rows.append([
            html_escape(sm.get('name', '-')),
            discovery_badge(detail.get('status', 'ACTIVE'), 'OK' if detail.get('status', 'ACTIVE') == 'ACTIVE' else 'INFO'),
            metadata,
            states_count,
            _details_html('View ARN', [sm.get('stateMachineArn', '-')]),
        ])
    return html_table(['State machine', 'Status', 'Metadata', 'States', 'Identifier'], rows, widths=['24%', '12%', '38%', '10%', '16%'])


def render_sfn_workflow_states(machines):
    html = ''
    any_states = False
    for item in sorted(machines, key=lambda data: data.get('summary', {}).get('name', '')):
        sm_name = item.get('summary', {}).get('name', '-')
        state_rows = _sfn_extract_states(item.get('definition') or {})
        if not state_rows:
            continue
        any_states = True
        rows = [
            [html_escape(name), html_escape(kind), html_escape(target_service), html_escape(target), html_escape(str(next_value))]
            for name, kind, target_service, target, next_value in state_rows
        ]
        html += html_subsection(sm_name)
        html += html_table(['State', 'Type', 'Target service', 'Target', 'Next / End'], rows, widths=['24%', '14%', '18%', '28%', '16%'])
    if not any_states:
        return html_empty('No workflow states detected in Step Functions definitions.')
    return html


def render_sfn_recent_executions(machines):
    rows = []
    notes = []
    for item in sorted(machines, key=lambda data: data.get('summary', {}).get('name', '')):
        sm = item.get('summary') or {}
        if sm.get('type') == 'EXPRESS':
            notes.append(f"{sm.get('name', '-')}: EXPRESS machines do not support list_executions; executions require CloudWatch Logs when logging is enabled.")
            continue
        for ex in item.get('executions') or []:
            rows.append([
                html_escape(sm.get('name', '-')),
                html_escape(ex.get('name', '-')),
                discovery_badge(ex.get('status', 'INFO'), 'OK' if ex.get('status') == 'SUCCEEDED' else 'WARN' if ex.get('status') in ('FAILED', 'TIMED_OUT', 'ABORTED') else 'INFO'),
                html_escape(str(ex.get('startDate', 'N/A'))[:19]),
                html_escape(str(ex.get('stopDate', 'Running'))[:19]),
            ])
    html = ''
    if notes:
        html += html_note(' '.join(notes))
    if rows:
        html += html_table(['State machine', 'Execution', 'Status', 'Start', 'Stop'], rows, widths=['22%', '30%', '12%', '18%', '18%'])
    elif not notes:
        html += html_empty('No recent Step Functions executions detected.')
    return html


def render_sfn_tags(machines):
    rows = []
    for item in sorted(machines, key=lambda data: data.get('summary', {}).get('name', '')):
        sm_name = item.get('summary', {}).get('name', '-')
        tags = item.get('tags') or []
        if tags:
            pairs = sorted((tag.get('key'), tag.get('value')) for tag in tags)
            rows.append([html_escape(sm_name), len(tags), _key_value_details_html(f'View {len(tags)} tags', pairs)])
    if not rows:
        return html_empty('No Step Functions tags detected.')
    return html_table(['State machine', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_sfn_relationships(relations):
    sfn_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'Step Functions' or rel.get('target_service') == 'Step Functions'
    ]
    if not sfn_relations:
        return html_empty('No Step Functions relationships detected.')
    return render_discovery_relations_table(sfn_relations)


def render_discovery_stepfunctions():
    machines = _sfn_collect_details()
    relations = collect_architecture_relations()

    html = html_title('Step Functions - Discovery')
    html += html_section('Summary')
    html += render_sfn_summary(machines, relations)

    html += render_sfn_relationship_diagram(relations)

    html += html_section('State machines')
    html += render_sfn_state_machines(machines)

    html += html_section('Workflow states')
    html += render_sfn_workflow_states(machines)

    html += html_section('Recent executions')
    html += render_sfn_recent_executions(machines)

    html += html_section('Tags')
    html += render_sfn_tags(machines)

    html += html_section('Step Functions relationships')
    html += render_sfn_relationships(relations)
    return html
```

```python
# === Discovery renderers: Elastic Beanstalk + ElastiCache ===

# ---------------------------------------------------------------------------
# Elastic Beanstalk Discovery
# ---------------------------------------------------------------------------

def _eb_collect_details():
    apps = _safe_aws([], lambda: clients['elasticbeanstalk'].describe_applications().get('Applications', []))
    envs = _safe_aws([], lambda: clients['elasticbeanstalk'].describe_environments(IncludeDeleted=False).get('Environments', []))
    enriched = []
    for env in envs:
        env_name = env.get('EnvironmentName')
        app_name = env.get('ApplicationName')
        resources = _safe_aws({}, lambda env_name=env_name: clients['elasticbeanstalk'].describe_environment_resources(EnvironmentName=env_name).get('EnvironmentResources', {}))
        settings = _safe_aws([], lambda app_name=app_name, env_name=env_name: clients['elasticbeanstalk'].describe_configuration_settings(
            ApplicationName=app_name,
            EnvironmentName=env_name
        ).get('ConfigurationSettings', [])) if app_name and env_name else []
        options = settings[0].get('OptionSettings', []) if settings else []
        env_vars = [opt for opt in options if opt.get('Namespace') == 'aws:elasticbeanstalk:application:environment']
        instance_type = next((opt.get('Value') for opt in options if opt.get('OptionName') == 'InstanceType'), 'N/A')
        detailed_health = _safe_aws({}, lambda env_name=env_name: clients['elasticbeanstalk'].describe_environment_health(
            EnvironmentName=env_name,
            AttributeNames=['All']
        )) if env_name else {}
        instance_health = _safe_aws([], lambda env_name=env_name: clients['elasticbeanstalk'].describe_instances_health(
            EnvironmentName=env_name,
            AttributeNames=['All']
        ).get('InstanceHealthList', [])) if env_name else []
        events = _safe_aws([], lambda env_name=env_name: clients['elasticbeanstalk'].describe_events(
            EnvironmentName=env_name,
            MaxRecords=10
        ).get('Events', [])) if env_name else []
        tags = _safe_aws([], lambda arn=env.get('EnvironmentArn'): clients['elasticbeanstalk'].list_tags_for_resource(ResourceArn=arn).get('ResourceTags', [])) if env.get('EnvironmentArn') else []
        enriched.append({
            'environment': env,
            'resources': resources,
            'settings': settings,
            'options': options,
            'env_vars': env_vars,
            'instance_type': instance_type,
            'health': detailed_health,
            'instance_health': instance_health,
            'events': events,
            'tags': tags,
        })
    return apps, enriched


def _eb_status_badge(status):
    status = str(status or 'INFO')
    level = 'OK' if status.lower() in ('ready', 'available', 'green', 'ok') else 'WARN'
    if status.lower() in ('red', 'severe', 'degraded', 'terminated'):
        level = 'CRIT'
    if status in ('N/A', '-', ''):
        level = 'INFO'
    return discovery_badge(status, level)


def _eb_health_badge(color, status=None):
    label = color or status or 'N/A'
    color_lower = str(label).lower()
    if color_lower == 'green':
        level = 'OK'
    elif color_lower in ('yellow', 'warning'):
        level = 'WARN'
    elif color_lower in ('red', 'severe'):
        level = 'CRIT'
    else:
        level = 'INFO'
    return discovery_badge(label, level)


def _eb_resource_names(resources, key, field):
    return [item.get(field, '-') for item in (resources.get(key) or [])]


def render_eb_summary(apps, envs, relations):
    green = sum(1 for item in envs if str((item.get('environment') or {}).get('Health', '')).lower() == 'green')
    instances = sum(len((item.get('resources') or {}).get('Instances') or []) for item in envs)
    asgs = sum(len((item.get('resources') or {}).get('AutoScalingGroups') or []) for item in envs)
    lbs = sum(len((item.get('resources') or {}).get('LoadBalancers') or []) for item in envs)
    eb_relations = [rel for rel in relations if rel.get('source_service') == 'Elastic Beanstalk' or rel.get('target_service') == 'Elastic Beanstalk']
    rows = [
        ['Applications detected', len(apps)],
        ['Environments detected', len(envs)],
        ['Green environments', f'{green}/{len(envs)}' if envs else '0'],
        ['EC2 instances', instances],
        ['Auto Scaling groups', asgs],
        ['Load balancers', lbs],
        ['Elastic Beanstalk relationships detected', len(eb_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_eb_relationship_diagram(relations):
    eb_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'Elastic Beanstalk' or rel.get('target_service') == 'Elastic Beanstalk'
    ]
    html = html_section('Elastic Beanstalk architecture diagram')
    html += render_mermaid_diagram(
        eb_relations,
        'Relationships where Elastic Beanstalk appears as source or target, mainly inferred from environment variables.'
    )
    return html


def render_eb_applications(apps):
    if not apps:
        return html_empty('No Elastic Beanstalk applications detected.')
    rows = []
    for app in sorted(apps, key=lambda item: item.get('ApplicationName', '')):
        versions = app.get('Versions') or []
        templates = app.get('ConfigurationTemplates') or []
        metadata = _compact_metadata_lines([
            ('Created', str(app.get('DateCreated', '-'))[:10]),
            ('Updated', str(app.get('DateUpdated', '-'))[:10]),
            ('Versions', len(versions)),
            ('Config templates', len(templates)),
        ])
        rows.append([
            html_escape(app.get('ApplicationName', '-')),
            metadata,
            _details_html(f'View {len(versions)} versions', versions) if versions else '-',
            _details_html(f'View {len(templates)} templates', templates) if templates else '-',
        ])
    return html_table(['Application', 'Metadata', 'Versions', 'Configuration templates'], rows, widths=['24%', '34%', '21%', '21%'])


def render_eb_environments(envs):
    if not envs:
        return html_empty('No Elastic Beanstalk environments detected.')
    rows = []
    for item in sorted(envs, key=lambda entry: (entry.get('environment') or {}).get('EnvironmentName', '')):
        env = item.get('environment') or {}
        metadata = _compact_metadata_lines([
            ('Application', env.get('ApplicationName', 'N/A')),
            ('Platform', env.get('SolutionStackName') or env.get('PlatformArn') or 'N/A'),
            ('Version', env.get('VersionLabel', 'N/A')),
            ('Created', str(env.get('DateCreated', '-'))[:10]),
            ('Updated', str(env.get('DateUpdated', '-'))[:10]),
        ])
        endpoint_items = [
            f"URL: http://{env.get('CNAME')}" if env.get('CNAME') else 'URL: N/A',
            f"Environment ID: {env.get('EnvironmentId', 'N/A')}",
        ]
        rows.append([
            html_escape(env.get('EnvironmentName', '-')),
            _eb_status_badge(env.get('Status')),
            _eb_health_badge(env.get('Health'), env.get('HealthStatus')),
            metadata,
            _details_html('View endpoint details', endpoint_items),
        ])
    return html_table(['Environment', 'Status', 'Health', 'Metadata', 'Endpoint / ID'], rows, widths=['18%', '12%', '12%', '43%', '15%'])


def render_eb_environment_resources(envs):
    if not envs:
        return html_empty('No Elastic Beanstalk environments detected.')
    rows = []
    for item in sorted(envs, key=lambda entry: (entry.get('environment') or {}).get('EnvironmentName', '')):
        env = item.get('environment') or {}
        resources = item.get('resources') or {}
        instances = _eb_resource_names(resources, 'Instances', 'Id')
        asgs = _eb_resource_names(resources, 'AutoScalingGroups', 'Name')
        lbs = _eb_resource_names(resources, 'LoadBalancers', 'Name')
        rows.append([
            html_escape(env.get('EnvironmentName', '-')),
            _details_html(f'View {len(instances)} instances', instances) if instances else 'No instances detected',
            _details_html(f'View {len(asgs)} groups', asgs) if asgs else 'No Auto Scaling groups detected',
            _details_html(f'View {len(lbs)} load balancers', lbs) if lbs else 'No load balancers detected',
        ])
    return html_table(['Environment', 'EC2 instances', 'Auto Scaling groups', 'Load balancers'], rows, widths=['24%', '25%', '25%', '26%'])


def render_eb_configuration(envs):
    if not envs:
        return html_empty('No Elastic Beanstalk configuration detected.')
    rows = []
    for item in sorted(envs, key=lambda entry: (entry.get('environment') or {}).get('EnvironmentName', '')):
        env = item.get('environment') or {}
        env_vars = item.get('env_vars') or []
        variable_pairs = [(var.get('OptionName', '-'), var.get('Value', '-')) for var in env_vars]
        option_count = len(item.get('options') or [])
        settings_count = len(item.get('settings') or [])
        details = _compact_metadata_lines([
            ('Configuration settings', settings_count),
            ('Option settings', option_count),
            ('Environment variables', len(env_vars)),
        ])
        rows.append([
            html_escape(env.get('EnvironmentName', '-')),
            html_escape(item.get('instance_type', 'N/A')),
            details,
            _key_value_details_html(f'View {len(env_vars)} variables', variable_pairs) if variable_pairs else 'No environment variables configured',
        ])
    return html_table(['Environment', 'Instance type', 'Configuration summary', 'Environment variables'], rows, widths=['24%', '18%', '28%', '30%'])


def render_eb_health(envs):
    if not envs:
        return html_empty('No Elastic Beanstalk health information detected.')
    rows = []
    for item in sorted(envs, key=lambda entry: (entry.get('environment') or {}).get('EnvironmentName', '')):
        env = item.get('environment') or {}
        health = item.get('health') or {}
        causes = health.get('Causes') or []
        rows.append([
            html_escape(env.get('EnvironmentName', '-')),
            _eb_health_badge(health.get('Color') or env.get('Health'), health.get('Status') or env.get('HealthStatus')),
            html_escape(health.get('Status') or env.get('HealthStatus') or 'N/A'),
            _details_html(f'View {len(causes)} causes', causes) if causes else 'No health causes reported',
        ])
    return html_table(['Environment', 'Color', 'Status', 'Causes'], rows, widths=['26%', '14%', '20%', '40%'])


def render_eb_instance_health(envs):
    rows = []
    for item in sorted(envs, key=lambda entry: (entry.get('environment') or {}).get('EnvironmentName', '')):
        env = item.get('environment') or {}
        for instance in item.get('instance_health') or []:
            cpu = (instance.get('System') or {}).get('CPUUtilization') or {}
            metadata = _compact_metadata_lines([
                ('Launched', str(instance.get('LaunchedAt', '-'))[:19]),
                ('CPU user', cpu.get('User', 'N/A')),
            ])
            rows.append([
                html_escape(env.get('EnvironmentName', '-')),
                html_escape(instance.get('InstanceId', '-')),
                _eb_health_badge(instance.get('Color'), instance.get('HealthStatus')),
                html_escape(instance.get('HealthStatus', 'N/A')),
                metadata,
            ])
    if not rows:
        return html_empty('No Elastic Beanstalk instance health details returned.')
    return html_table(['Environment', 'Instance', 'Color', 'Health status', 'Metadata'], rows, widths=['22%', '22%', '12%', '18%', '26%'])


def render_eb_recent_events(envs):
    rows = []
    for item in sorted(envs, key=lambda entry: (entry.get('environment') or {}).get('EnvironmentName', '')):
        env = item.get('environment') or {}
        for event in item.get('events') or []:
            rows.append([
                str(event.get('EventDate', '-'))[:19],
                html_escape(env.get('EnvironmentName', '-')),
                discovery_badge(event.get('Severity', 'INFO'), 'INFO' if str(event.get('Severity', '')).upper() == 'INFO' else 'WARN'),
                html_escape(event.get('Message', 'N/A')),
            ])
    if not rows:
        return html_empty('No recent Elastic Beanstalk events detected.')
    rows.sort(key=lambda row: row[0], reverse=True)
    return html_table(['Date', 'Environment', 'Severity', 'Message'], rows[:20], widths=['18%', '22%', '12%', '48%'])


def render_eb_tags(envs):
    rows = []
    for item in sorted(envs, key=lambda entry: (entry.get('environment') or {}).get('EnvironmentName', '')):
        env = item.get('environment') or {}
        tags = item.get('tags') or []
        if tags:
            pairs = sorted((tag.get('Key'), tag.get('Value')) for tag in tags)
            rows.append([html_escape(env.get('EnvironmentName', '-')), len(tags), _key_value_details_html(f'View {len(tags)} tags', pairs)])
    if not rows:
        return html_empty('No Elastic Beanstalk tags detected.')
    return html_table(['Environment', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_eb_relationships(relations):
    eb_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'Elastic Beanstalk' or rel.get('target_service') == 'Elastic Beanstalk'
    ]
    if not eb_relations:
        return html_empty('No Elastic Beanstalk relationships detected.')
    return render_discovery_relations_table(eb_relations)


def render_discovery_elasticbeanstalk():
    apps, envs = _eb_collect_details()
    relations = collect_architecture_relations()

    html = html_title('Elastic Beanstalk - Discovery')
    html += html_section('Summary')
    html += render_eb_summary(apps, envs, relations)

    html += render_eb_relationship_diagram(relations)

    html += html_section('Applications')
    html += render_eb_applications(apps)

    html += html_section('Environments')
    html += render_eb_environments(envs)

    html += html_section('Environment resources')
    html += render_eb_environment_resources(envs)

    html += html_section('Configuration')
    html += render_eb_configuration(envs)

    html += html_section('Environment health')
    html += render_eb_health(envs)

    html += html_section('Instance health')
    html += render_eb_instance_health(envs)

    html += html_section('Recent events')
    html += render_eb_recent_events(envs)

    html += html_section('Tags')
    html += render_eb_tags(envs)

    html += html_section('Elastic Beanstalk relationships')
    html += render_eb_relationships(relations)
    return html


# ---------------------------------------------------------------------------
# ElastiCache Discovery
# ---------------------------------------------------------------------------

def _elasticache_collect_details():
    clusters = _safe_aws([], lambda: clients['elasticache'].describe_cache_clusters(ShowCacheNodeInfo=True).get('CacheClusters', []))
    subnet_groups = _safe_aws([], lambda: clients['elasticache'].describe_cache_subnet_groups().get('CacheSubnetGroups', []))
    subnet_groups_by_name = {sg.get('CacheSubnetGroupName'): sg for sg in subnet_groups}
    events = _safe_aws([], lambda: clients['elasticache'].describe_events(SourceType='cache-cluster', MaxRecords=20).get('Events', []))
    enriched = []
    for cluster in clusters:
        arn = cluster.get('ARN')
        tags = _safe_aws([], lambda arn=arn: clients['elasticache'].list_tags_for_resource(ResourceName=arn).get('TagList', [])) if arn else []
        subnet_group = subnet_groups_by_name.get(cluster.get('CacheSubnetGroupName'), {})
        enriched.append({
            'cluster': cluster,
            'subnet_group': subnet_group,
            'tags': tags,
        })
    return enriched, subnet_groups, events


def _elasticache_status_badge(status):
    status = str(status or 'INFO')
    level = 'OK' if status.lower() == 'available' else 'WARN'
    if status.lower() in ('deleting', 'deleted', 'incompatible-network', 'restore-failed'):
        level = 'CRIT'
    if status in ('N/A', '-', ''):
        level = 'INFO'
    return discovery_badge(status, level)


def _elasticache_endpoint_label(endpoint):
    endpoint = endpoint or {}
    address = endpoint.get('Address', 'N/A')
    port = endpoint.get('Port', 'N/A')
    return f'{address}:{port}' if address != 'N/A' else 'N/A'


def _elasticache_subnet_summary(subnet_group):
    subnets = []
    azs = []
    for subnet in subnet_group.get('Subnets', []) or []:
        subnet_id = subnet.get('SubnetIdentifier', 'N/A')
        az = (subnet.get('SubnetAvailabilityZone') or {}).get('Name', 'N/A')
        subnets.append(subnet_id)
        azs.append(az)
    return subnets, sorted(set(azs))


def render_elasticache_summary(enriched, subnet_groups, relations):
    available = sum(1 for item in enriched if (item.get('cluster') or {}).get('CacheClusterStatus') == 'available')
    nodes = sum(len((item.get('cluster') or {}).get('CacheNodes') or []) for item in enriched)
    engines = sorted(set((item.get('cluster') or {}).get('Engine', 'N/A') for item in enriched))
    ec_relations = [rel for rel in relations if rel.get('source_service') == 'ElastiCache' or rel.get('target_service') == 'ElastiCache']
    rows = [
        ['Clusters detected', len(enriched)],
        ['Available clusters', f'{available}/{len(enriched)}' if enriched else '0'],
        ['Cache nodes detected', nodes],
        ['Engines detected', ', '.join(engines) if engines else 'N/A'],
        ['Subnet groups detected', len(subnet_groups)],
        ['ElastiCache relationships detected', len(ec_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_elasticache_relationship_diagram(relations):
    ec_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'ElastiCache' or rel.get('target_service') == 'ElastiCache'
    ]
    html = html_section('ElastiCache architecture diagram')
    html += render_mermaid_diagram(
        ec_relations,
        'Relationships where ElastiCache appears as source or target, mainly inferred from environment variables.'
    )
    return html


def render_elasticache_clusters(enriched):
    if not enriched:
        return html_empty('No ElastiCache clusters detected.')
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('cluster') or {}).get('CacheClusterId', '')):
        cluster = item.get('cluster') or {}
        nodes = cluster.get('CacheNodes') or []
        security_groups = [sg.get('SecurityGroupId', '-') for sg in cluster.get('SecurityGroups') or []]
        metadata = _compact_metadata_lines([
            ('Engine', f"{cluster.get('Engine', 'N/A')} {cluster.get('EngineVersion', '')}".strip()),
            ('Node type', cluster.get('CacheNodeType', 'N/A')),
            ('Created', str(cluster.get('CacheClusterCreateTime', '-'))[:10]),
            ('Preferred AZ', cluster.get('PreferredAvailabilityZone', 'N/A')),
            ('Parameter group', (cluster.get('CacheParameterGroup') or {}).get('CacheParameterGroupName', 'N/A')),
            ('Subnet group', cluster.get('CacheSubnetGroupName', 'N/A')),
            ('Auto minor upgrade', _rds_yes_no(cluster.get('AutoMinorVersionUpgrade'))),
        ])
        counts = _compact_metadata_lines([
            ('Security groups', len(security_groups)),
            ('Nodes', len(nodes)),
        ])
        rows.append([
            html_escape(cluster.get('CacheClusterId', '-')),
            _elasticache_status_badge(cluster.get('CacheClusterStatus')),
            metadata,
            counts,
        ])
    return html_table(['Cluster', 'Status', 'Metadata', 'Counts'], rows, widths=['22%', '12%', '48%', '18%'])


def render_elasticache_endpoints(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('cluster') or {}).get('CacheClusterId', '')):
        cluster = item.get('cluster') or {}
        cluster_id = cluster.get('CacheClusterId', '-')
        config_endpoint = cluster.get('ConfigurationEndpoint')
        if config_endpoint:
            rows.append([
                html_escape(cluster_id),
                'Configuration endpoint',
                '-',
                discovery_badge('AVAILABLE', 'OK'),
                html_escape(_elasticache_endpoint_label(config_endpoint)),
                '-',
            ])
        for node in cluster.get('CacheNodes') or []:
            rows.append([
                html_escape(cluster_id),
                'Cache node',
                html_escape(node.get('CacheNodeId', 'N/A')),
                _elasticache_status_badge(node.get('CacheNodeStatus')),
                html_escape(_elasticache_endpoint_label(node.get('Endpoint'))),
                str(node.get('CacheNodeCreateTime', '-'))[:19],
            ])
    if not rows:
        return html_empty('No ElastiCache endpoints or nodes detected.')
    return html_table(['Cluster', 'Type', 'Node', 'Status', 'Endpoint', 'Created'], rows, widths=['22%', '18%', '10%', '12%', '28%', '10%'])


def render_elasticache_subnet_groups(enriched, subnet_groups):
    rows = []
    for sg in sorted(subnet_groups, key=lambda item: item.get('CacheSubnetGroupName', '')):
        subnets, azs = _elasticache_subnet_summary(sg)
        rows.append([
            html_escape(sg.get('CacheSubnetGroupName', '-')),
            html_escape(sg.get('CacheSubnetGroupDescription', 'N/A')),
            html_escape(sg.get('VpcId', 'N/A')),
            _details_html(f'View {len(subnets)} subnets', subnets) if subnets else '-',
            ', '.join(azs) if azs else 'N/A',
        ])
    if not rows:
        return html_empty('No ElastiCache subnet groups detected.')
    return html_table(['Subnet group', 'Description', 'VPC', 'Subnets', 'AZs'], rows, widths=['22%', '28%', '18%', '17%', '15%'])


def render_elasticache_recent_events(events):
    if not events:
        return html_empty('No recent ElastiCache events detected.')
    rows = []
    for event in sorted(events, key=lambda item: str(item.get('Date', '')), reverse=True):
        rows.append([
            str(event.get('Date', '-'))[:19],
            html_escape(event.get('SourceIdentifier', 'N/A')),
            html_escape(event.get('SourceType', 'N/A')),
            html_escape(event.get('Message', 'N/A')),
        ])
    return html_table(['Date', 'Source', 'Type', 'Message'], rows[:20], widths=['18%', '24%', '14%', '44%'])


def render_elasticache_tags(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('cluster') or {}).get('CacheClusterId', '')):
        cluster = item.get('cluster') or {}
        tags = item.get('tags') or []
        if tags:
            pairs = sorted((tag.get('Key'), tag.get('Value')) for tag in tags)
            rows.append([html_escape(cluster.get('CacheClusterId', '-')), len(tags), _key_value_details_html(f'View {len(tags)} tags', pairs)])
    if not rows:
        return html_empty('No ElastiCache tags detected.')
    return html_table(['Cluster', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_elasticache_relationships(relations):
    ec_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'ElastiCache' or rel.get('target_service') == 'ElastiCache'
    ]
    if not ec_relations:
        return html_empty('No ElastiCache relationships detected.')
    return render_discovery_relations_table(ec_relations)


def render_discovery_elasticache():
    enriched, subnet_groups, events = _elasticache_collect_details()
    relations = collect_architecture_relations()

    html = html_title('ElastiCache - Discovery')
    html += html_section('Summary')
    html += render_elasticache_summary(enriched, subnet_groups, relations)

    html += render_elasticache_relationship_diagram(relations)

    html += html_section('Clusters')
    html += render_elasticache_clusters(enriched)

    html += html_section('Endpoints and nodes')
    html += render_elasticache_endpoints(enriched)

    html += html_section('Subnet groups')
    html += render_elasticache_subnet_groups(enriched, subnet_groups)

    html += html_section('Recent events')
    html += render_elasticache_recent_events(events)

    html += html_section('Tags')
    html += render_elasticache_tags(enriched)

    html += html_section('ElastiCache relationships')
    html += render_elasticache_relationships(relations)
    return html
```

```python
# === Discovery renderers: ECR + CodeCommit + WAF ===

# ---------------------------------------------------------------------------
# ECR Discovery
# ---------------------------------------------------------------------------

def _ecr_collect_details():
    repos = _safe_aws([], lambda: clients['ecr'].describe_repositories().get('repositories', []))
    enriched = []
    for repo in repos:
        name = repo.get('repositoryName')
        images = _safe_aws([], lambda name=name: clients['ecr'].describe_images(repositoryName=name).get('imageDetails', [])) if name else []
        try:
            lifecycle = clients['ecr'].get_lifecycle_policy(repositoryName=name) if name else {}
            lifecycle_status = 'Configured'
        except Exception as exc:
            lifecycle = {}
            lifecycle_status = 'Not configured' if _error_code(exc) == 'LifecyclePolicyNotFoundException' else f'No access: {_error_code(exc) or exc}'
        arn = repo.get('repositoryArn')
        tags = _safe_aws([], lambda arn=arn: clients['ecr'].list_tags_for_resource(resourceArn=arn).get('tags', [])) if arn else []
        enriched.append({
            'repository': repo,
            'images': images,
            'lifecycle': lifecycle,
            'lifecycle_status': lifecycle_status,
            'tags': tags,
        })
    return enriched


def _ecr_scan_badge(enabled):
    return discovery_badge('ENABLED' if enabled else 'DISABLED', 'OK' if enabled else 'WARN')


def _ecr_mutability_badge(value):
    value = str(value or 'N/A')
    return discovery_badge(value, 'OK' if value == 'IMMUTABLE' else 'WARN' if value == 'MUTABLE' else 'INFO')


def _ecr_lifecycle_badge(status):
    status = str(status or 'N/A')
    if status == 'Configured':
        return discovery_badge('CONFIGURED', 'OK')
    if status == 'Not configured':
        return discovery_badge('NOT CONFIGURED', 'WARN')
    return discovery_badge('NO ACCESS', 'INFO')


def _ecr_image_tags_label(image):
    tags = image.get('imageTags') or []
    return ', '.join(tags) if tags else '<untagged>'


def _ecr_size_mb(bytes_value):
    return round((bytes_value or 0) / (1024 * 1024), 1)


def render_ecr_summary(enriched, relations):
    images = sum(len(item.get('images') or []) for item in enriched)
    untagged = sum(1 for item in enriched for image in item.get('images') or [] if not image.get('imageTags'))
    scan_enabled = sum(1 for item in enriched if ((item.get('repository') or {}).get('imageScanningConfiguration') or {}).get('scanOnPush'))
    lifecycle_configured = sum(1 for item in enriched if item.get('lifecycle_status') == 'Configured')
    ecr_relations = [rel for rel in relations if rel.get('source_service') == 'ECR' or rel.get('target_service') == 'ECR']
    rows = [
        ['Repositories detected', len(enriched)],
        ['Images detected', images],
        ['Untagged images', untagged],
        ['Repositories with scan on push', f'{scan_enabled}/{len(enriched)}' if enriched else '0'],
        ['Repositories with lifecycle policy', f'{lifecycle_configured}/{len(enriched)}' if enriched else '0'],
        ['ECR relationships detected', len(ecr_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_ecr_relationship_diagram(relations):
    ecr_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'ECR' or rel.get('target_service') == 'ECR'
    ]
    html = html_section('ECR architecture diagram')
    html += render_mermaid_diagram(
        ecr_relations,
        'Relationships where ECR appears as source or target, usually Lambda container image dependencies.'
    )
    return html


def render_ecr_repositories(enriched):
    if not enriched:
        return html_empty('No ECR repositories detected.')
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('repository') or {}).get('repositoryName', '')):
        repo = item.get('repository') or {}
        images = item.get('images') or []
        metadata = _compact_metadata_lines([
            ('Created', str(repo.get('createdAt', '-'))[:10]),
            ('Encryption', (repo.get('encryptionConfiguration') or {}).get('encryptionType', 'N/A')),
            ('Images', len(images)),
            ('Registry ID', repo.get('registryId', 'N/A')),
        ])
        config = ''.join([
            '<div style="display:flex; align-items:center; gap:8px; line-height:1.8; white-space:nowrap;">'
            '<strong style="color:#2E5090; min-width:110px;">Tag mutability:</strong>'
            f'{_ecr_mutability_badge(repo.get("imageTagMutability"))}'
            '</div>',
            '<div style="display:flex; align-items:center; gap:8px; line-height:1.8; white-space:nowrap;">'
            '<strong style="color:#2E5090; min-width:110px;">Scan on push:</strong>'
            f'{_ecr_scan_badge(((repo.get("imageScanningConfiguration") or {}).get("scanOnPush")))}'
            '</div>',
            '<div style="display:flex; align-items:center; gap:8px; line-height:1.8; white-space:nowrap;">'
            '<strong style="color:#2E5090; min-width:110px;">Lifecycle:</strong>'
            f'{_ecr_lifecycle_badge(item.get("lifecycle_status"))}'
            '</div>',
        ])
        rows.append([
            html_escape(repo.get('repositoryName', '-')),
            metadata,
            config,
            _details_html('View URI', [repo.get('repositoryUri', '-')]),
        ])
    return html_table(['Repository', 'Metadata', 'Configuration', 'URI'], rows, widths=['24%', '32%', '32%', '12%'])


def render_ecr_images(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('repository') or {}).get('repositoryName', '')):
        repo = item.get('repository') or {}
        repo_name = repo.get('repositoryName', '-')
        for image in sorted(item.get('images') or [], key=lambda img: str(img.get('imagePushedAt', '')), reverse=True):
            scan_status = (image.get('imageScanStatus') or {}).get('status', 'N/A')
            findings = (image.get('imageScanFindingsSummary') or {}).get('findingSeverityCounts') or {}
            metadata = _compact_metadata_lines([
                ('Size', f"{_ecr_size_mb(image.get('imageSizeInBytes'))} MB"),
                ('Pushed', str(image.get('imagePushedAt', '-'))[:19]),
                ('Scan', scan_status),
            ])
            finding_items = [f'{severity}: {count}' for severity, count in sorted(findings.items())]
            digest = image.get('imageDigest', '-')
            rows.append([
                html_escape(repo_name),
                html_escape(_ecr_image_tags_label(image)),
                metadata,
                _details_html('View findings', finding_items) if finding_items else 'No scan findings summary',
                _details_html('View digest', [digest]),
            ])
    if not rows:
        return html_empty('No ECR images detected.')
    return html_table(['Repository', 'Tags', 'Metadata', 'Scan findings', 'Digest'], rows, widths=['20%', '20%', '25%', '20%', '15%'])


def render_ecr_lifecycle_policies(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('repository') or {}).get('repositoryName', '')):
        repo = item.get('repository') or {}
        lifecycle = item.get('lifecycle') or {}
        policy_text = lifecycle.get('lifecyclePolicyText')
        rows.append([
            html_escape(repo.get('repositoryName', '-')),
            _ecr_lifecycle_badge(item.get('lifecycle_status')),
            _details_html('View policy JSON', [policy_text]) if policy_text else html_escape(item.get('lifecycle_status', 'N/A')),
        ])
    if not rows:
        return html_empty('No ECR lifecycle policy information detected.')
    return html_table(['Repository', 'Status', 'Policy'], rows, widths=['32%', '18%', '50%'])


def render_ecr_tags(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('repository') or {}).get('repositoryName', '')):
        repo = item.get('repository') or {}
        tags = item.get('tags') or []
        if tags:
            pairs = sorted((tag.get('Key'), tag.get('Value')) for tag in tags)
            rows.append([html_escape(repo.get('repositoryName', '-')), len(tags), _key_value_details_html(f'View {len(tags)} tags', pairs)])
    if not rows:
        return html_empty('No ECR tags detected.')
    return html_table(['Repository', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_ecr_relationships(relations):
    ecr_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'ECR' or rel.get('target_service') == 'ECR'
    ]
    if not ecr_relations:
        return html_empty('No ECR relationships detected.')
    return render_discovery_relations_table(ecr_relations)


def render_discovery_ecr():
    enriched = _ecr_collect_details()
    relations = collect_architecture_relations()

    html = html_title('ECR - Discovery')
    html += html_section('Summary')
    html += render_ecr_summary(enriched, relations)

    html += render_ecr_relationship_diagram(relations)

    html += html_section('Repositories')
    html += render_ecr_repositories(enriched)

    html += html_section('Images')
    html += render_ecr_images(enriched)

    html += html_section('Lifecycle policies')
    html += render_ecr_lifecycle_policies(enriched)

    html += html_section('Tags')
    html += render_ecr_tags(enriched)

    html += html_section('ECR relationships')
    html += render_ecr_relationships(relations)
    return html


# ---------------------------------------------------------------------------
# CodeCommit Discovery
# ---------------------------------------------------------------------------

def _codecommit_commit_date(raw_date):
    try:
        timestamp = int(str(raw_date).split(' ')[0])
        return datetime.fromtimestamp(timestamp).strftime('%Y-%m-%d %H:%M:%S')
    except Exception:
        return str(raw_date or 'N/A')


def _codecommit_collect_details():
    repos = _safe_aws([], lambda: clients['codecommit'].list_repositories().get('repositories', []))
    enriched = []
    for summary in repos:
        name = summary.get('repositoryName')
        metadata = _safe_aws({}, lambda name=name: clients['codecommit'].get_repository(repositoryName=name).get('repositoryMetadata', {})) if name else {}
        branches = _safe_aws([], lambda name=name: clients['codecommit'].list_branches(repositoryName=name).get('branches', [])) if name else []
        default_branch = metadata.get('defaultBranch')
        branch_info = _safe_aws({}, lambda name=name, branch=default_branch: clients['codecommit'].get_branch(repositoryName=name, branchName=branch).get('branch', {})) if name and default_branch else {}
        commit_id = branch_info.get('commitId')
        commit = _safe_aws({}, lambda name=name, commit_id=commit_id: clients['codecommit'].get_commit(repositoryName=name, commitId=commit_id).get('commit', {})) if name and commit_id else {}
        folder = _safe_aws({}, lambda name=name: clients['codecommit'].get_folder(repositoryName=name, folderPath='/')) if name else {}
        triggers = _safe_aws([], lambda name=name: clients['codecommit'].get_repository_triggers(repositoryName=name).get('triggers', [])) if name else []
        arn = metadata.get('Arn')
        tags = _safe_aws({}, lambda arn=arn: clients['codecommit'].list_tags_for_resource(resourceArn=arn).get('tags', {})) if arn else {}
        enriched.append({
            'summary': summary,
            'metadata': metadata,
            'branches': branches,
            'branch_info': branch_info,
            'commit': commit,
            'folder': folder,
            'triggers': triggers,
            'tags': tags,
        })
    return enriched


def render_codecommit_summary(enriched):
    branches = sum(len(item.get('branches') or []) for item in enriched)
    triggers = sum(len(item.get('triggers') or []) for item in enriched)
    files = sum(len((item.get('folder') or {}).get('files') or []) for item in enriched)
    folders = sum(len((item.get('folder') or {}).get('subFolders') or []) for item in enriched)
    rows = [
        ['Repositories detected', len(enriched)],
        ['Branches detected', branches],
        ['Triggers configured', triggers],
        ['Root files detected', files],
        ['Root folders detected', folders],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_codecommit_repositories(enriched):
    if not enriched:
        return html_empty('No CodeCommit repositories detected.')
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('metadata') or entry.get('summary') or {}).get('repositoryName', '')):
        meta = item.get('metadata') or {}
        name = meta.get('repositoryName') or (item.get('summary') or {}).get('repositoryName', '-')
        metadata = _compact_metadata_lines([
            ('Created', str(meta.get('creationDate', '-'))[:10]),
            ('Updated', str(meta.get('lastModifiedDate', '-'))[:10]),
            ('Default branch', meta.get('defaultBranch', 'N/A')),
            ('Branches', len(item.get('branches') or [])),
        ])
        identifiers = [
            f"ARN: {meta.get('Arn', 'N/A')}",
            f"Clone URL: {meta.get('cloneUrlHttp', 'N/A')}",
        ]
        rows.append([
            html_escape(name),
            html_escape(meta.get('repositoryDescription') or '-'),
            metadata,
            _details_html('View identifiers', identifiers),
        ])
    return html_table(['Repository', 'Description', 'Metadata', 'Identifiers'], rows, widths=['24%', '28%', '32%', '16%'])


def render_codecommit_branches(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('metadata') or entry.get('summary') or {}).get('repositoryName', '')):
        meta = item.get('metadata') or {}
        name = meta.get('repositoryName') or (item.get('summary') or {}).get('repositoryName', '-')
        branches = item.get('branches') or []
        rows.append([
            html_escape(name),
            discovery_badge(meta.get('defaultBranch', 'N/A'), 'OK' if meta.get('defaultBranch') else 'INFO'),
            len(branches),
            _details_html(f'View {len(branches)} branches', branches) if branches else 'No branches returned',
        ])
    if not rows:
        return html_empty('No CodeCommit branch information detected.')
    return html_table(['Repository', 'Default branch', 'Branch count', 'Branches'], rows, widths=['30%', '18%', '12%', '40%'])


def render_codecommit_latest_commits(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('metadata') or entry.get('summary') or {}).get('repositoryName', '')):
        meta = item.get('metadata') or {}
        name = meta.get('repositoryName') or (item.get('summary') or {}).get('repositoryName', '-')
        commit = item.get('commit') or {}
        if not commit:
            rows.append([html_escape(name), html_escape(meta.get('defaultBranch', 'N/A')), '-', '-', 'No latest commit returned'])
            continue
        author = commit.get('author') or {}
        rows.append([
            html_escape(name),
            html_escape(meta.get('defaultBranch', 'N/A')),
            html_escape(str(commit.get('commitId', '-'))[:12]),
            html_escape(author.get('name', 'N/A')),
            _compact_metadata_lines([
                ('Date', _codecommit_commit_date(author.get('date'))),
                ('Message', str(commit.get('message', 'N/A')).strip()[:160]),
            ]),
        ])
    if not rows:
        return html_empty('No CodeCommit commit information detected.')
    return html_table(['Repository', 'Branch', 'Commit', 'Author', 'Details'], rows, widths=['24%', '16%', '14%', '16%', '30%'])


def render_codecommit_root_content(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('metadata') or entry.get('summary') or {}).get('repositoryName', '')):
        meta = item.get('metadata') or {}
        name = meta.get('repositoryName') or (item.get('summary') or {}).get('repositoryName', '-')
        folder = item.get('folder') or {}
        files = folder.get('files') or []
        subfolders = folder.get('subFolders') or []
        file_items = []
        for subfolder in subfolders:
            file_items.append(f"folder: {subfolder.get('relativePath', 'N/A')}/")
        for file in files:
            size = file.get('blobSize')
            size_label = f" ({round(size / 1024, 1)} KB)" if size else ''
            file_items.append(f"file: {file.get('relativePath', 'N/A')}{size_label}")
        rows.append([
            html_escape(name),
            len(files),
            len(subfolders),
            _details_html(f'View {len(file_items)} root items', file_items) if file_items else 'No root content returned',
        ])
    if not rows:
        return html_empty('No CodeCommit root content detected.')
    return html_table(['Repository', 'Files', 'Folders', 'Root content'], rows, widths=['30%', '12%', '12%', '46%'])


def render_codecommit_triggers(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('metadata') or entry.get('summary') or {}).get('repositoryName', '')):
        meta = item.get('metadata') or {}
        name = meta.get('repositoryName') or (item.get('summary') or {}).get('repositoryName', '-')
        triggers = item.get('triggers') or []
        if not triggers:
            rows.append([html_escape(name), discovery_badge('NONE', 'INFO'), 'No triggers configured'])
            continue
        details = []
        for trigger in triggers:
            destination = trigger.get('destinationArn', 'N/A')
            events = ', '.join(trigger.get('events') or []) or 'N/A'
            details.append(f"{trigger.get('name', 'N/A')} | {destination} | {events}")
        rows.append([html_escape(name), discovery_badge(f'{len(triggers)} TRIGGERS', 'OK'), _details_html(f'View {len(triggers)} triggers', details)])
    if not rows:
        return html_empty('No CodeCommit trigger information detected.')
    return html_table(['Repository', 'Triggers', 'Details'], rows, widths=['30%', '18%', '52%'])


def render_codecommit_tags(enriched):
    rows = []
    for item in sorted(enriched, key=lambda entry: (entry.get('metadata') or entry.get('summary') or {}).get('repositoryName', '')):
        meta = item.get('metadata') or {}
        name = meta.get('repositoryName') or (item.get('summary') or {}).get('repositoryName', '-')
        tags = item.get('tags') or {}
        if tags:
            pairs = sorted(tags.items())
            rows.append([html_escape(name), len(tags), _key_value_details_html(f'View {len(tags)} tags', pairs)])
    if not rows:
        return html_empty('No CodeCommit tags detected.')
    return html_table(['Repository', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_discovery_codecommit():
    enriched = _codecommit_collect_details()

    html = html_title('CodeCommit - Discovery')
    html += html_section('Summary')
    html += render_codecommit_summary(enriched)

    html += html_section('Repositories')
    html += render_codecommit_repositories(enriched)

    html += html_section('Branches')
    html += render_codecommit_branches(enriched)

    html += html_section('Latest commits')
    html += render_codecommit_latest_commits(enriched)

    html += html_section('Root content')
    html += render_codecommit_root_content(enriched)

    html += html_section('Triggers')
    html += render_codecommit_triggers(enriched)

    html += html_section('Tags')
    html += render_codecommit_tags(enriched)
    return html

# ---------------------------------------------------------------------------
# WAF Discovery
# ---------------------------------------------------------------------------

def _waf_collect_details(scope='REGIONAL'):
    acl_summaries = _safe_aws([], lambda scope=scope: clients['wafv2'].list_web_acls(Scope=scope).get('WebACLs', []))
    acls = []
    for summary in acl_summaries:
        name = summary.get('Name')
        acl_id = summary.get('Id')
        arn = summary.get('ARN')
        detail = _safe_aws({}, lambda name=name, acl_id=acl_id, scope=scope: clients['wafv2'].get_web_acl(Name=name, Scope=scope, Id=acl_id).get('WebACL', {}))
        resources = []
        for resource_type in ('APPLICATION_LOAD_BALANCER', 'API_GATEWAY', 'APPSYNC', 'COGNITO_USER_POOL'):
            resource_arns = _safe_aws([], lambda arn=arn, resource_type=resource_type: clients['wafv2'].list_resources_for_web_acl(WebACLArn=arn, ResourceType=resource_type).get('ResourceArns', []))
            for resource_arn in resource_arns:
                resources.append({'type': resource_type, 'arn': resource_arn, 'name': resource_arn.split('/')[-1]})
        logging_config = _safe_aws({}, lambda arn=arn: clients['wafv2'].get_logging_configuration(ResourceArn=arn).get('LoggingConfiguration', {}))
        tags = _safe_aws([], lambda arn=arn: clients['wafv2'].list_tags_for_resource(ResourceARN=arn).get('TagInfoForResource', {}).get('TagList', []))
        acls.append({
            'summary': summary,
            'detail': detail,
            'resources': resources,
            'logging': logging_config,
            'tags': tags,
            'scope': scope,
        })
    ip_sets = _safe_aws([], lambda scope=scope: clients['wafv2'].list_ip_sets(Scope=scope).get('IPSets', []))
    rule_groups = _safe_aws([], lambda scope=scope: clients['wafv2'].list_rule_groups(Scope=scope).get('RuleGroups', []))
    return acls, ip_sets, rule_groups


def _waf_default_action(web_acl):
    action = (web_acl or {}).get('DefaultAction') or {}
    if 'Allow' in action:
        return 'Allow'
    if 'Block' in action:
        return 'Block'
    return 'N/A'


def _waf_rule_action(rule):
    action = rule.get('Action') or rule.get('OverrideAction') or {}
    if 'Block' in action:
        return 'Block'
    if 'Allow' in action:
        return 'Allow'
    if 'Count' in action:
        return 'Count'
    if 'None' in action:
        return 'None'
    return 'Group default'


def _waf_rule_type(rule):
    statement = rule.get('Statement') or {}
    managed = statement.get('ManagedRuleGroupStatement') or {}
    if managed:
        return f"Managed rule group ({managed.get('VendorName', 'N/A')}/{managed.get('Name', 'N/A')})"
    if statement.get('IPSetReferenceStatement'):
        return 'IP set reference'
    if statement.get('RateBasedStatement'):
        return 'Rate based'
    if statement.get('RuleGroupReferenceStatement'):
        return 'Rule group reference'
    return ', '.join(statement.keys()) or 'Custom statement'


def render_waf_summary(acls, ip_sets, rule_groups, relations):
    total_rules = sum(len((acl.get('detail') or {}).get('Rules') or []) for acl in acls)
    managed_rules = 0
    protected_resources = sum(len(acl.get('resources') or []) for acl in acls)
    logging_enabled = sum(1 for acl in acls if acl.get('logging'))
    for acl in acls:
        for rule in (acl.get('detail') or {}).get('Rules') or []:
            if ((rule.get('Statement') or {}).get('ManagedRuleGroupStatement')):
                managed_rules += 1
    waf_relations = [rel for rel in relations if rel.get('source_service') == 'WAF' or rel.get('target_service') == 'WAF']
    rows = [
        ['Web ACLs detected', len(acls)],
        ['Rules detected', total_rules],
        ['Managed rule groups used', managed_rules],
        ['Protected resources', protected_resources],
        ['Web ACLs with logging enabled', logging_enabled],
        ['IP sets detected', len(ip_sets)],
        ['Custom rule groups detected', len(rule_groups)],
        ['WAF relationships detected', len(waf_relations)],
    ]
    return html_table(['Metric', 'Value'], rows, widths=['65%', '35%'])


def render_waf_relationship_diagram(relations):
    waf_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'WAF' or rel.get('target_service') == 'WAF'
    ]
    html = html_section('WAF architecture diagram')
    html += render_mermaid_diagram(
        waf_relations,
        'Relationships where WAF appears as source or target, usually Web ACL associations with protected AWS resources.'
    )
    return html


def render_waf_web_acls(acls):
    if not acls:
        return html_empty('No WAF Web ACLs detected.')
    rows = []
    for acl in sorted(acls, key=lambda item: item.get('summary', {}).get('Name', '')):
        detail = acl.get('detail') or {}
        summary = acl.get('summary') or {}
        visibility = detail.get('VisibilityConfig') or {}
        metadata = _compact_metadata_lines([
            ('ID', summary.get('Id', '-')),
            ('Scope', acl.get('scope', 'REGIONAL')),
            ('Default action', _waf_default_action(detail)),
            ('Capacity', f"{detail.get('Capacity', 'N/A')} WCU"),
        ])
        visibility_meta = _compact_metadata_lines([
            ('CloudWatch metrics', _rds_yes_no(visibility.get('CloudWatchMetricsEnabled'))),
            ('Metric name', visibility.get('MetricName', 'N/A')),
            ('Sampled requests', _rds_yes_no(visibility.get('SampledRequestsEnabled'))),
            ('Logging', 'Enabled' if acl.get('logging') else 'Not configured'),
        ])
        counts = _compact_metadata_lines([
            ('Rules', len(detail.get('Rules') or [])),
            ('Resources', len(acl.get('resources') or [])),
        ])
        rows.append([
            html_escape(summary.get('Name', '-')),
            metadata,
            visibility_meta,
            counts,
        ])
    return html_table(['Web ACL', 'Metadata', 'Visibility / logging', 'Counts'], rows, widths=['20%', '32%', '32%', '16%'])


def render_waf_identifiers(acls):
    rows = []
    for acl in sorted(acls, key=lambda item: item.get('summary', {}).get('Name', '')):
        summary = acl.get('summary') or {}
        rows.append([
            html_escape(summary.get('Name', '-')),
            _details_html('View ARN', [summary.get('ARN', '-')]),
        ])
    if not rows:
        return html_empty('No WAF identifiers detected.')
    return html_table(['Web ACL', 'Identifier'], rows, widths=['28%', '72%'])


def render_waf_rules(acls):
    html = ''
    any_rules = False
    for acl in sorted(acls, key=lambda item: item.get('summary', {}).get('Name', '')):
        name = acl.get('summary', {}).get('Name', '-')
        rules = (acl.get('detail') or {}).get('Rules') or []
        if not rules:
            continue
        any_rules = True
        rows = []
        for rule in sorted(rules, key=lambda item: item.get('Priority', 0)):
            visibility = rule.get('VisibilityConfig') or {}
            detail_lines = [
                f"Type: {_waf_rule_type(rule)}",
                f"Metric: {visibility.get('MetricName', 'N/A')}",
                f"CloudWatch metrics: {_rds_yes_no(visibility.get('CloudWatchMetricsEnabled'))}",
                f"Sampled requests: {_rds_yes_no(visibility.get('SampledRequestsEnabled'))}",
            ]
            rows.append([
                rule.get('Priority', 'N/A'),
                html_escape(rule.get('Name', '-')),
                html_escape(_waf_rule_action(rule)),
                _details_html('View rule details', detail_lines),
            ])
        html += html_subsection(name)
        html += html_table(['Priority', 'Rule', 'Action', 'Details'], rows, widths=['10%', '40%', '16%', '34%'])
    if not any_rules:
        return html_empty('No WAF rules detected.')
    return html


def render_waf_protected_resources(acls):
    rows = []
    for acl in sorted(acls, key=lambda item: item.get('summary', {}).get('Name', '')):
        acl_name = acl.get('summary', {}).get('Name', '-')
        for resource in acl.get('resources') or []:
            rows.append([
                html_escape(acl_name),
                html_escape(resource.get('type', '-')),
                html_escape(resource.get('name', '-')),
                _details_html('View ARN', [resource.get('arn', '-')]),
            ])
    if not rows:
        return html_empty('No WAF protected resources detected.')
    return html_table(['Web ACL', 'Resource type', 'Resource', 'Identifier'], rows, widths=['24%', '24%', '28%', '24%'])


def render_waf_ip_sets_and_rule_groups(ip_sets, rule_groups):
    rows = []
    for ip_set in sorted(ip_sets, key=lambda item: item.get('Name', '')):
        rows.append(['IP set', html_escape(ip_set.get('Name', '-')), html_escape(ip_set.get('Id', '-')), _details_html('View ARN', [ip_set.get('ARN', '-')])])
    for rule_group in sorted(rule_groups, key=lambda item: item.get('Name', '')):
        rows.append(['Rule group', html_escape(rule_group.get('Name', '-')), html_escape(rule_group.get('Id', '-')), _details_html('View ARN', [rule_group.get('ARN', '-')])])
    if not rows:
        return html_empty('No WAF IP sets or custom rule groups detected.')
    return html_table(['Type', 'Name', 'ID', 'Identifier'], rows, widths=['18%', '30%', '22%', '30%'])


def render_waf_tags(acls):
    rows = []
    for acl in sorted(acls, key=lambda item: item.get('summary', {}).get('Name', '')):
        tags = acl.get('tags') or []
        if tags:
            pairs = sorted((tag.get('Key'), tag.get('Value')) for tag in tags)
            rows.append([html_escape(acl.get('summary', {}).get('Name', '-')), len(tags), _key_value_details_html(f'View {len(tags)} tags', pairs)])
    if not rows:
        return html_empty('No WAF tags detected.')
    return html_table(['Web ACL', 'Tag count', 'Tags'], rows, widths=['30%', '12%', '58%'])


def render_waf_relationships(relations):
    waf_relations = [
        rel for rel in relations
        if rel.get('source_service') == 'WAF' or rel.get('target_service') == 'WAF'
    ]
    if not waf_relations:
        return html_empty('No WAF relationships detected.')
    return render_discovery_relations_table(waf_relations)


def render_discovery_waf():
    acls, ip_sets, rule_groups = _waf_collect_details()
    relations = collect_architecture_relations()

    html = html_title('WAF - Discovery')
    html += html_section('Summary')
    html += render_waf_summary(acls, ip_sets, rule_groups, relations)

    html += render_waf_relationship_diagram(relations)

    html += html_section('Web ACLs')
    html += render_waf_web_acls(acls)

    html += html_section('Web ACL identifiers')
    html += render_waf_identifiers(acls)

    html += html_section('Rules')
    html += render_waf_rules(acls)

    html += html_section('Protected resources')
    html += render_waf_protected_resources(acls)

    html += html_section('IP sets and custom rule groups')
    html += render_waf_ip_sets_and_rule_groups(ip_sets, rule_groups)

    html += html_section('Tags')
    html += render_waf_tags(acls)

    html += html_section('WAF relationships')
    html += render_waf_relationships(relations)
    return html
```

```python
# === Discovery router ===

def render_discovery_service(service):
    if service == 'Lambda':
        return render_discovery_lambda()
    if service == 'DynamoDB':
        return render_discovery_dynamodb()
    if service == 'API Gateway':
        return render_discovery_apigateway()
    if service == 'SQS':
        return render_discovery_sqs()
    if service == 'SNS':
        return render_discovery_sns()
    if service == 'S3':
        return render_discovery_s3()
    if service == 'RDS':
        return render_discovery_rds()
    if service == 'Cognito':
        return render_discovery_cognito()
    if service == 'Step Functions':
        return render_discovery_stepfunctions()
    if service == 'WAF':
        return render_discovery_waf()
    if service == 'Elastic Beanstalk':
        return render_discovery_elasticbeanstalk()
    if service == 'ElastiCache':
        return render_discovery_elasticache()
    if service == 'ECR':
        return render_discovery_ecr()
    if service == 'CodeCommit':
        return render_discovery_codecommit()
    return html_title(f'{service} - Discovery') + html_empty('Detailed discovery view for this service will be added next.')
```

```python
# === Dashboard router ===

def render_dashboard(service, tab):
    """Builds the dashboard HTML when the selected service or tab changes."""
    try:
        if service == 'Overview':
            return render_overview(tab)
        return render_service(service, tab)
    except Exception as e:
        return f'<p style="color: red;">Error: {e}</p>'

def render_overview(tab):
    if tab == 'Discovery':
        return render_discovery_overview()
    elif tab == 'Health':
        rows = []
        checks = {
            'Lambda': lambda: health_summary_lambda(clients),
            'DynamoDB': lambda: health_summary_dynamodb(clients),
            'API Gateway': lambda: health_summary_apigateway(clients),
            'SQS': lambda: health_summary_sqs(clients),
            'SNS': lambda: health_summary_sns(clients),
            'S3': lambda: health_summary_s3(clients),
            'RDS': lambda: health_summary_rds(clients),
            'Cognito': lambda: health_summary_cognito(clients),
            'Step Functions': lambda: health_summary_stepfunctions(clients),
            'WAF': lambda: health_summary_waf(clients),
            'Elastic Beanstalk': lambda: health_summary_eb(clients),
            'ElastiCache': lambda: health_summary_elasticache(clients),
            'ECR': lambda: health_summary_ecr(clients),
            'CodeCommit': lambda: health_summary_codecommit(clients),
        }
        category_counts = {'Healthy': 0, 'Warning': 0, 'Critical': 0, 'Info': 0}
        for svc, fn in checks.items():
            try:
                status, detail = fn()
                category_counts[health_category(status)] += 1
                rows.append((svc, html_status(status, ''), detail))
            except Exception as e:
                category_counts['Info'] += 1
                rows.append((svc, html_status('INFO', ''), str(e)))

        return (
            html_title('Health Summary')
            + render_health_donut(category_counts)
            + html_table(['Service', 'Status', 'Details'], rows)
        )
    
    elif tab == 'Metrics':
        return render_metrics_overview()
    elif tab == 'Logs':
        return render_logs_overview()
    elif tab == 'Security':
        return render_security_overview()
    else:
        return html_title(f'Overview — {tab}') + '<p style="color: #888;">Select a specific service for detailed analysis.</p>'

def render_service(service, tab):
    if tab == 'Discovery':
        return render_discovery_service(service)
    if tab == 'Health':
        func = HEALTH_FUNCTIONS.get(service)
        if not func:
            return html_title(f'{service} — Health') + '<p>No health check available.</p>'
        return render_health_output(service, func(clients))
    
    elif tab == 'Metrics':
        renderer = METRICS_RENDERERS.get(service)
        if renderer:
            return renderer()
        return render_metrics_no_data(service)
    elif tab == 'Logs':
        return render_logs_service(service)
    elif tab == 'Security':
        return render_security_service(service)
    else:
        return html_title(f'{service} — {tab}') + '<p style="color: #888;">Implementation pending.</p>'
```

```python
# === UI and callbacks ===
from pathlib import Path

header = widgets.HTML(
    value='<h2 style="color: #2E5090; margin-bottom: 0;">AWS Monitoring Dashboard</h2>'
    '<p style="color: #666; margin-top: 4px; margin-bottom: 10px;">Cloud Microservices Architecture Analysis Tool</p>'
)

service_dropdown = widgets.Dropdown(
    options=SERVICES, value='Overview',
    description='Service:', style={'description_width': '70px'},
    layout=widgets.Layout(width='280px')
)

# Los botones de pestaña se crean al final, cuando ya existen las functions de renderizado.
tab_button_widgets = []
tabs_box = widgets.HBox([])

status_bar = widgets.HTML(value='')
refresh_button = widgets.Button(
    description='Refresh',
    icon='refresh',
    tooltip='Refresh the current dashboard view',
    layout=widgets.Layout(width='110px')
)
report_button = widgets.Button(
    description='Generate report',
    icon='file',
    tooltip='Choose a report export type',
    layout=widgets.Layout(width='160px')
)
current_report_button = widgets.Button(
    description='Current page',
    icon='file',
    tooltip='Export only the currently selected service and tab',
    layout=widgets.Layout(width='140px')
)
full_report_button = widgets.Button(
    description='Full report',
    icon='files-o',
    tooltip='Export all services and all dashboard tabs',
    layout=widgets.Layout(width='140px')
)
report_options = widgets.HBox(
    [current_report_button, full_report_button],
    layout=widgets.Layout(display='none', gap='8px', align_items='center')
)
report_status = widgets.HTML(value='')
content = widgets.Output(layout=widgets.Layout(
    border='1px solid #ddd',
    min_height='300px',
    max_height='600px',
    padding='15px',
    overflow='auto',
    width='100%'
))
# --- Wire up events ---
# Explicit callbacks keep the notebook dashboard predictable and easy to debug.
current_state = {
    'service': service_dropdown.value,
    'tab': 'Discovery',
}
render_cache = {}
TIME_RANGE_TABS = {'Metrics', 'Logs'}


def update_range_visibility():
    range_dropdown.layout.display = 'flex' if current_state['tab'] in TIME_RANGE_TABS else 'none'

def set_active_tab_style():
    for button in tab_button_widgets:
        button.button_style = 'info' if button.description == current_state['tab'] else ''

def set_status(service, tab, source):
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    status_bar.value = (
        f'<p style="color: #999; font-size: 12px;">'
        f'Last updated: {timestamp} | Service: {service} | Tab: {tab} | Source: {source}</p>'
    )

def render_current(force_refresh=False):
    service = current_state['service']
    tab = current_state['tab']
    cache_key = (service, tab, RANGO if tab in TIME_RANGE_TABS else None)

    with content:
        clear_output(wait=True)
        if not force_refresh and cache_key in render_cache:
            html = render_cache[cache_key]
            source = 'cache'
        else:
            display(HTML(f'<p style="color: #777;">Loading {service} / {tab}...</p>'))
            html = render_dashboard(service, tab)
            render_cache[cache_key] = html
            source = 'AWS/API refresh'
            clear_output(wait=True)
        display(HTML(html))

    update_range_visibility()
    set_status(service, tab, source)
    set_active_tab_style()


def _report_cache_key(service, tab):
    return (service, tab, RANGO if tab in TIME_RANGE_TABS else None)


def _safe_filename_part(value):
    text = str(value).strip().lower()
    text = re.sub(r'[^a-z0-9]+', '_', text)
    return text.strip('_') or 'report'


def build_report_document(service, tab, body_html):
    generated_at = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    range_label = RANGO if tab in TIME_RANGE_TABS else 'not applicable'
    return f'''<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="utf-8">
  <meta name="viewport" content="width=device-width, initial-scale=1">
  <title>AWS Monitoring Report - {html_escape(service)} - {html_escape(tab)}</title>
  <style>
    body {{ font-family: Arial, sans-serif; color: #1f2937; margin: 32px; background: #fff; }}
    .report-header {{ border-bottom: 3px solid #2E5090; margin-bottom: 22px; padding-bottom: 14px; }}
    .report-header h1 {{ color: #2E5090; margin: 0 0 8px 0; font-size: 26px; }}
    .report-meta {{ color: #666; font-size: 13px; line-height: 1.6; }}
    .report-content {{ max-width: 1200px; }}
    table {{ page-break-inside: auto; }}
    tr {{ page-break-inside: avoid; page-break-after: auto; }}
    details {{ margin: 4px 0; }}
    @media print {{
      body {{ margin: 18mm; }}
      .report-header {{ page-break-after: avoid; }}
    }}
  </style>
</head>
<body>
  <div class="report-header">
    <h1>AWS Monitoring Report</h1>
    <div class="report-meta">
      <strong>Service:</strong> {html_escape(service)}<br>
      <strong>Tab:</strong> {html_escape(tab)}<br>
      <strong>Range:</strong> {html_escape(range_label)}<br>
      <strong>Generated at:</strong> {html_escape(generated_at)}
    </div>
  </div>
  <div class="report-content">
    {body_html}
  </div>
</body>
</html>'''



def get_report_output_dir():
    cwd = Path.cwd().resolve()
    project_root = cwd.parent if cwd.name.lower() == 'notebooks' else cwd
    output_dir = project_root / 'outputs'
    output_dir.mkdir(exist_ok=True)
    return output_dir


def _get_report_html(service, tab):
    cache_key = _report_cache_key(service, tab)
    if cache_key in render_cache:
        return render_cache[cache_key]
    html = render_dashboard(service, tab)
    render_cache[cache_key] = html
    return html


def build_full_report_body():
    sections = []
    for service in SERVICES:
        sections.append(
            f'<h2 style="color:#2E5090; border-bottom:2px solid #2E5090; '
            f'padding-bottom:6px; margin-top:34px;">{html_escape(service)}</h2>'
        )
        for tab in TABS:
            sections.append(
                f'<section style="margin-top:18px; page-break-inside:avoid;">'
                f'<h3 style="color:#333; margin-bottom:10px;">{html_escape(tab)}</h3>'
                f'{_get_report_html(service, tab)}'
                f'</section>'
            )
    return ''.join(sections)


def export_full_report():
    body_html = build_full_report_body()
    output_dir = get_report_output_dir()
    timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
    report_path = (output_dir / f'full_monitoring_report_{timestamp}.html').resolve()
    report_path.write_text(build_report_document('Full report', 'All dashboard tabs', body_html), encoding='utf-8')
    return report_path


def export_current_report():
    service = current_state['service']
    tab = current_state['tab']
    body_html = _get_report_html(service, tab)

    output_dir = get_report_output_dir()
    timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
    filename = f"{_safe_filename_part(service)}_{_safe_filename_part(tab)}_{timestamp}.html"
    report_path = (output_dir / filename).resolve()
    report_path.write_text(build_report_document(service, tab, body_html), encoding='utf-8')
    return report_path


def _set_report_success(report_path):
    report_status.value = (
        f'<p style="color:#27ae60; font-size:12px; margin:4px 0 0 0;">'
        f'Report generated: <code>{html_escape(str(report_path))}</code></p>'
    )


def _set_report_error(exc):
    report_status.value = (
        f'<p style="color:#c0392b; font-size:12px; margin:4px 0 0 0;">'
        f'Report error: {html_escape(str(exc))}</p>'
    )


def on_report_click(_):
    report_options.layout.display = 'flex' if report_options.layout.display == 'none' else 'none'
    if report_options.layout.display == 'flex':
        report_status.value = '<p style="color:#777; font-size:12px; margin:4px 0 0 0;">Choose report type.</p>'


def on_current_report_click(_):
    try:
        report_status.value = '<p style="color:#777; font-size:12px; margin:4px 0 0 0;">Generating current page report...</p>'
        report_path = export_current_report()
        report_options.layout.display = 'none'
        _set_report_success(report_path)
    except Exception as exc:
        _set_report_error(exc)


def on_full_report_click(_):
    try:
        report_status.value = '<p style="color:#777; font-size:12px; margin:4px 0 0 0;">Generating full report. This can take a few minutes...</p>'
        report_path = export_full_report()
        report_options.layout.display = 'none'
        _set_report_success(report_path)
    except Exception as exc:
        _set_report_error(exc)


def on_service_change(change):
    if change.get('name') == 'value':
        current_state['service'] = change['new']
        render_current()

def refresh_time_window():
    global start_time, end_time
    end_time = datetime.now()
    start_time = end_time - _delta

def on_refresh_click(_):
    refresh_time_window()
    render_current(force_refresh=True)

def make_tab_button(tab):
    button = widgets.Button(
        description=tab,
        layout=widgets.Layout(width='120px'),
        style={'font_weight': 'bold'}
    )

    def on_tab_click(_):
        current_state['tab'] = tab
        render_current()

    button.on_click(on_tab_click)
    return button

tab_button_widgets = [make_tab_button(tab) for tab in TABS]
tabs_box = widgets.HBox(tab_button_widgets, layout=widgets.Layout(width='100%', gap='4px'))

service_dropdown.observe(on_service_change, names='value')
range_dropdown = widgets.Dropdown(
    options=[('Last hour', 'ultima_hora'), ('Last day', 'ultimo_dia'),
             ('Last week', 'ultima_semana'), ('Last month', 'ultimo_mes')],
    value=RANGO,
    description='Range:',
    style={'description_width': '60px'},
    layout=widgets.Layout(width='200px')
)

def on_change_range(change):
    if change['name'] == 'value':
        global RANGO, PERIOD, start_time, end_time, _delta
        RANGO = change['new']
        _delta, PERIOD = _rangos.get(RANGO, _rangos['ultimo_dia'])
        end_time = datetime.now()
        start_time = end_time - _delta
        render_current()

range_dropdown.observe(on_change_range, names='value')
refresh_button.on_click(on_refresh_click)
report_button.on_click(on_report_click)
current_report_button.on_click(on_current_report_click)
full_report_button.on_click(on_full_report_click)

dashboard = widgets.VBox([
    header,
    widgets.HBox([service_dropdown, range_dropdown, refresh_button, report_button, status_bar], layout=widgets.Layout(align_items='center', gap='14px', flex_flow='row wrap')),
    report_options,
    report_status,
    tabs_box,
    content,
], layout=widgets.Layout(width='100%', max_width='900px'))

display(dashboard)
render_current()
```
